## **1. Czym jest Julia?**

### Teoria

**Julia** to nowoczesny język programowania zaprojektowany z myślą o:

- obliczeniach naukowych,
- analizie danych,
- modelowaniu matematycznym,
- symulacjach,
- optymalizacji,
- uczeniu maszynowym,
- programowaniu ogólnego przeznaczenia.

Język łączy wygodę pracy interaktywnej z wysoką wydajnością. Jedną z jego najważniejszych cech jest **multiple dispatch**, czyli dobór implementacji funkcji na podstawie typów argumentów.

Julia wykorzystuje także:

- dynamiczne typowanie,
- kompilację JIT opartą na LLVM,
- wnioskowanie o typach,
- bogate wsparcie dla matematyki i Unicode,
- bardzo dobrą interoperacyjność z innymi językami i bibliotekami.


### Ważne

Julia nie jest „tylko językiem do matematyki”. To pełnoprawny język programowania, który może być używany również do:

- budowy narzędzi,
- automatyzacji,
- przetwarzania danych,
- budowy prototypów,
- rozwiązań inżynierskich i badawczych.


### Przykład


````julia
println("Hello, Julia!")
````

### Analiza

To najprostszy możliwy program w tej lekcji.

- `println(...)` wypisuje tekst,
- tekst zapisujemy w cudzysłowach `"..."`,
- po wykonaniu instrukcji kursor przechodzi do nowej linii.


## **2. Krótka historia języka Julia i jego twórcy**

### Teoria

Prace nad językiem Julia rozpoczęły się pod koniec pierwszej dekady XXI wieku. Język został zaprojektowany po to, aby połączyć:

- prostotę i wygodę języków wysokiego poziomu,
- wysoką wydajność charakterystyczną dla języków kompilowanych,
- naturalny zapis kodu matematycznego.

Pierwsza publiczna prezentacja języka miała miejsce w **2012 roku**, kiedy opublikowano słynny tekst **“Why We Created Julia”**.


### Twórcy

Za twórców języka Julia uznaje się cztery osoby:

| Twórca | Rola |
|---|---|
| **Jeff Bezanson** | współtwórca języka, projektant i badacz |
| **Stefan Karpinski** | współtwórca, popularyzator i współautor ekosystemu |
| **Viral B. Shah** | współtwórca i współorganizator rozwoju projektu |
| **Alan Edelman** | współtwórca, matematyk i badacz związany z MIT |

Julia rozwijała się w ścisłym związku ze środowiskiem naukowym i inżynierskim, a z czasem stała się dojrzałym ekosystemem z własnym menedżerem pakietów, dokumentacją i dużą społecznością.


### Dlaczego powstała Julia?

Twórcy chcieli ograniczyć problem „dwóch języków”, czyli sytuacji, w której:

1. prototyp buduje się w języku wysokiego poziomu,
2. a fragmenty krytyczne dla wydajności trzeba później przepisywać do C lub C++.

Julia miała pozwolić pisać czytelny kod wysokiego poziomu, który jednocześnie może być bardzo szybki.


## **3. REPL i notebook**

### Teoria

Najczęściej z Julią pracujemy w dwóch trybach:

| Tryb | Znaczenie |
|---|---|
| **REPL** | interaktywny terminal Julii |
| **Notebook** | dokument łączący tekst, kod i wyniki |

**REPL** to skrót od:

**Read → Eval → Print → Loop**

czyli:

1. wczytaj polecenie,
2. wykonaj je,
3. pokaż wynik,
4. czekaj na następne.


### Ważne

Notebook zapamiętuje stan sesji.

To oznacza, że jeśli uruchomisz komórkę tworzącą zmienną `x`, to w późniejszej komórce możesz z niej korzystać. Jednak gdy uruchomisz komórki w innej kolejności, wyniki mogą być inne albo pojawi się błąd.

Dlatego w praktyce warto regularnie wykonywać notebook **od początku do końca**.


### Spróbuj sam


````julia
VERSION
````

Instrukcja `VERSION` pokazuje wersję aktualnie używanego środowiska Julia.


## **4. Wyświetlanie danych**

### Teoria

W tej sekcji poznasz cztery ważne narzędzia:

| Funkcja | Zastosowanie |
|---|---|
| `print` | wypisywanie bez nowej linii |
| `println` | wypisywanie z przejściem do nowej linii |
| `show` | techniczna reprezentacja wartości |
| `display` | prezentacja wartości przez system wyświetlania notebooka |


### Przykład — `print` i `println`


````julia
print("Julia ")
print("jest ")
print("czytelna.")
println()

println("Pierwsza linia")
println("Druga linia")
````

### Analiza

- `print` nie dodaje końca linii,
- `println` automatycznie kończy linię,
- dlatego kilka wywołań `print` może budować jeden komunikat.


### Przykład — `show` i `display`


````julia
text = "Julia\nLanguage"

println("println:")
println(text)

println("\nshow:")
show(text)

println("\n\ndisplay:")
display(text)
````

### Analiza

Warto zauważyć różnice:

- `println(text)` pokazuje tekst tak, jak ma zostać wypisany,
- `show(text)` pokazuje techniczną reprezentację napisu,
- `display(text)` przekazuje reprezentację do środowiska notebooka.


### Przykład — interpolacja napisów


````julia
name = "Julia"
major = VERSION.major
minor = VERSION.minor

println("Uczę się języka $name.")
println("Aktualna wersja główna: $major")
println("Numer wersji: $(major).$(minor)")
println("2 + 3 = $(2 + 3)")
````

### Typowy błąd

Początkujący próbują czasem pisać:

```julia
println("2 + 3 = $2 + 3")
```

Tak nie należy robić. Jeśli chcesz umieścić w napisie całe wyrażenie, użyj:

```julia
println("2 + 3 = $(2 + 3)")
```


## **5. Zmienne i przypisania**

### Teoria

Zmienna w Julia to **nazwa powiązana z wartością**.

Nie trzeba wcześniej deklarować jej typu. Wystarczy przypisać wartość:

```julia
x = 10
```


### Przykład


````julia
answer = 42
language = "Julia"
is_fast = true

println(answer)
println(language)
println(is_fast)
````

### Przykład — sprawdzanie typu


````julia
@show typeof(answer)
@show typeof(language)
@show typeof(is_fast)
````

### Ważne

Typ należy do **wartości**, a nie do samej nazwy zmiennej. Oznacza to, że ta sama nazwa może później wskazywać wartość innego typu.


````julia
x = 10
@show x typeof(x)

x = 10.5
@show x typeof(x)

x = "dziesięć"
@show x typeof(x)
````

### Analiza

Powyższy przykład pokazuje, że:

- `x` może najpierw oznaczać liczbę całkowitą,
- potem liczbę zmiennoprzecinkową,
- a później napis.

To typowe dla języka dynamicznego.

| Typ | Przykład | Opis |

| `Char` | `'A'` | pojedynczy znak |

### Przykład — liczby całkowite i zmiennoprzecinkowe

To dwa różne typy danych.

## **7. Unicode w nazwach**

### Teoria

Julia bardzo dobrze wspiera Unicode. Dzięki temu można używać wygodnych nazw matematycznych, na przykład:

- `α`,
- `β`,
- `π`,
- `σ`.


### Przykład


````julia
α = 0.25
β = 0.75
suma = α + β

@show α
@show β
@show suma
````

## **8. Komentarze**

### Teoria

Julia oferuje dwa podstawowe rodzaje komentarzy:

| Rodzaj | Składnia |
|---|---|
| Jednowierszowy | `# komentarz` |
| Wielowierszowy | `#= ... =#` |


### Przykład — komentarz jednowierszowy


````julia
# Ten komentarz nie jest wykonywany.
result = 2 + 2  # komentarz może być także na końcu linii

println(result)
````

### Przykład — komentarz wielowierszowy


````julia
#=
Komentarz zewnętrzny.

#=
Komentarz wewnętrzny.
=#

Koniec komentarza zewnętrznego.
=#

println("Kod działa.")
````

### Ważne

Dobry komentarz wyjaśnia przede wszystkim **dlaczego** coś robimy, a nie tylko powtarza dosłownie treść kodu.


## **9. Operatory arytmetyczne**

### Teoria

| Operacja | Operator | Przykład |
|---|---|---|
| dodawanie | `+` | `3 + 2` |
| odejmowanie | `-` | `10 - 4` |
| mnożenie | `*` | `6 * 3` |
| dzielenie | `/` | `10 / 3` |
| dzielenie całkowite | `÷` | `10 ÷ 3` |
| reszta z dzielenia | `%` | `10 % 3` |

### Przykład

`/` oraz `÷` nie oznaczają tego samego:

## **10. Porównania i logika**

### Teoria

| Znaczenie | Operator |
|---|---|
| równe | `==` |
| różne | `!=` |
| większe | `>` |
| mniejsze | `<` |
| większe lub równe | `>=` lub `≥` |
| mniejsze lub równe | `<=` lub `≤` |


### Przykład — porównania


````julia
@show 5 > 3
@show 4 == 4
@show 10 != 2
@show 3 <= 3
@show 8 ≥ 7
````

### Przykład — porównania łańcuchowe


````julia
x = 5
@show 1 < x < 10
````

### Teoria — logika

Najważniejsze operatory logiczne:

| Operator | Znaczenie |
|---|---|
| `&&` | logiczne AND |
| `||` | logiczne OR |
| `!` | negacja |


### Przykład


````julia
a = true
b = false

@show a && b
@show a || b
@show !a
````

### Analiza

Operatory `&&` i `||` działają z tzw. **krótkim spięciem**:

- przy `A && B`, jeśli `A` jest `false`, Julia nie musi już obliczać `B`;
- przy `A || B`, jeśli `A` jest `true`, Julia nie musi już obliczać `B`.


````julia
function positive_message(x)
    x > 0 && println("$x jest dodatnie")
    return nothing
end

positive_message(5)
positive_message(-2)
````

### Typowy błąd

Nie myl:

- `=` — przypisanie,
- `==` — porównanie.

To zupełnie różne operacje.


## **11. `convert`, `parse` i `tryparse`**

### Teoria

Początkujący bardzo często mylą konwersję wartości z parsowaniem tekstu.

| Funkcja | Zastosowanie |
|---|---|
| `convert(T, x)` | konwertuje istniejącą wartość `x` do typu `T`, jeśli jest to zdefiniowane |
| `parse(T, text)` | odczytuje tekst jako wartość typu `T` |
| `tryparse(T, text)` | próbuje sparsować tekst; przy błędzie zwraca `nothing` |


### Przykład — `convert`


````julia
converted = convert(Float64, 365)
@show converted typeof(converted)
````

### Przykład — `parse`


````julia
days_text = "365"
days_number = parse(Int, days_text)

@show days_text typeof(days_text)
@show days_number typeof(days_number)
````

### Przykład — `tryparse`


````julia
@show tryparse(Int, "365")
@show tryparse(Int, "365 dni")
````

### Typowy błąd

To nie jest poprawna droga odczytu liczby z tekstu:

```julia
convert(Int, "365")
```

Napis `"365"` jest tekstem, a nie gotową wartością liczbową w innym typie. Dlatego należy użyć:

```julia
parse(Int, "365")
```

### Zadanie 13.4
