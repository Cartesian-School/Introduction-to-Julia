## **1. Dlaczego potrzebujemy struktur danych?**

### Teoria

Pojedyncza zmienna przechowuje jedną wartość. W realnym programie często potrzebujemy jednak przechowywać:

- listę temperatur,
- dane użytkownika,
- książkę telefoniczną,
- macierz pomiarową,
- unikalny zbiór identyfikatorów,
- rekord zawierający nazwę, wiek i miasto.

**Struktura danych** organizuje wiele wartości w sposób, który ułatwia:

- przechowywanie,
- wyszukiwanie,
- modyfikację,
- iterację,
- przetwarzanie.


### Mapa najważniejszych struktur

| Struktura | Mutowalna? | Uporządkowana? | Typowe zastosowanie |
|---|---:|---:|---|
| `Tuple` | nie | tak | mały, stały zestaw wartości |
| `NamedTuple` | nie | tak | lekki rekord z nazwami pól |
| `Dict` | tak | logicznie klucz → wartość | szybki dostęp po kluczu |
| `Set` | tak | nie traktujemy kolejności jako kontraktu | unikalne wartości |
| `Vector` | tak | tak | jednowymiarowa tablica |
| `Matrix` | tak | tak | dane 2D |
| `Array{T,N}` | tak | tak | tablica N-wymiarowa |


## **2. Krotki — `Tuple`**

### Teoria

Krotka to uporządkowana, **niemutowalna** kolekcja wartości.

Tworzymy ją za pomocą nawiasów:


````julia
my_favorite_animals = ("pingwiny", "koty", "lotopałanki")

@show my_favorite_animals
@show typeof(my_favorite_animals)
````

### Indeksowanie

Julia indeksuje kolekcje od **1**.


````julia
@show my_favorite_animals[1]
@show my_favorite_animals[2]
@show my_favorite_animals[end]
````

### Ważne

`Tuple` jest niemutowalny. Nie możemy zrobić:

```julia
my_favorite_animals[1] = "wydry"
```

Jeżeli potrzebujemy zmienionej wersji, tworzymy nową krotkę.


### Przykład — krotka heterogeniczna


````julia
person = ("Anna", 30, true)

@show person
@show typeof(person)
````

### Analiza

Krotka może przechowywać elementy różnych typów bez uciekania się do `Any`. Typ krotki zachowuje informację o typie każdego elementu.


### Jednoelementowa krotka

To subtelny, ale ważny przypadek:


````julia
not_a_tuple = (42)
one_element_tuple = (42,)

@show typeof(not_a_tuple)
@show typeof(one_element_tuple)
````

### Typowy błąd

Nawiasy same w sobie nie tworzą jednoelementowej krotki.

Potrzebny jest przecinek:

```julia
(42,)
```


### Destrukturyzacja krotki

Możemy przypisać elementy do kilku zmiennych jednocześnie:


````julia
coordinates = (10, 20)
x, y = coordinates

@show x y
````

## **3. `NamedTuple` — krotka z nazwami pól**

### Teoria

`NamedTuple` działa podobnie do zwykłego `Tuple`, ale każdy element posiada nazwę.

To bardzo wygodna struktura dla małych rekordów danych.


````julia
animal_profile = (
    bird = "pingwin",
    mammal = "kot",
    marsupial = "lotopałanka",
)

@show animal_profile
@show typeof(animal_profile)
````

### Dostęp przez indeks i nazwę


````julia
@show animal_profile[1]
@show animal_profile.bird
@show animal_profile.mammal
````

### Przykład — dane osoby


````julia
person = (
    name = "Alice",
    age = 30,
    city = "New York",
)

println(person.name)
println(person.age)
println(person.city)
````

### Tworzenie z istniejących zmiennych


````julia
name = "Bob"
age = 25
city = "Los Angeles"

person2 = (; name, age, city)

@show person2
````

### Ważne

`NamedTuple` jest niemutowalny.

Nie możemy wykonać:

```julia
person.age = 31
```

Możemy natomiast zbudować nowy rekord.


### Aktualizacja przez `merge`


````julia
person_updated = merge(person, (age = 31,))

@show person
@show person_updated
````

### Zastosowanie w funkcjach


````julia
function describe_person(p)
    return "$(p.name) ma $(p.age) lat i mieszka w $(p.city)."
end

println(describe_person(person_updated))
````

## **4. Słowniki — `Dict`**

### Teoria

`Dict` przechowuje pary:

**klucz → wartość**

Przykład:

```text
"Anna" → "123-456"
```

Słownik jest mutowalny i świetnie nadaje się do danych, które chcemy odnajdywać po unikalnym kluczu.


### Tworzenie słownika


````julia
phonebook = Dict(
    "Anna" => "111-222",
    "Bartosz" => "333-444",
    "Siergiej" => "555-666",
)

@show phonebook
@show typeof(phonebook)
````

### Dostęp po kluczu


````julia
@show phonebook["Anna"]
````

### Typowy błąd

Dostęp do nieistniejącego klucza przez:

```julia
phonebook["Nieznany"]
```

zgłosi `KeyError`.

Dlatego dla danych niepewnych warto stosować bezpieczniejsze metody.


## **5. Bezpieczny dostęp do `Dict`**

### `haskey`


````julia
@show haskey(phonebook, "Anna")
@show haskey(phonebook, "Nieznany")
````

### `get`

`get(dict, key, default)` zwraca wartość albo wartość domyślną.


````julia
number = get(phonebook, "Nieznany", "brak numeru")

@show number
````

### `get!`

`get!` może dodatkowo zapisać wartość domyślną w słowniku.


````julia
settings = Dict{String, Int}()

timeout = get!(settings, "timeout", 30)

@show timeout
@show settings
````

## **6. Modyfikowanie słownika**

### Dodawanie i aktualizacja


````julia
phonebook["Ewa"] = "777-888"
phonebook["Anna"] = "999-000"

@show phonebook
````

### Usuwanie przez `delete!`


````julia
temporary = Dict("a" => 1, "b" => 2, "c" => 3)

delete!(temporary, "b")

@show temporary
````

### `pop!` — usuń i zwróć


````julia
contacts = Dict(
    "Anna" => "111",
    "Siergiej" => "222",
)

removed_number = pop!(contacts, "Siergiej")

@show removed_number
@show contacts
````

### Klucze, wartości i pary


````julia
example = Dict("a" => 10, "b" => 20)

@show collect(keys(example))
@show collect(values(example))
@show collect(pairs(example))
````

### Iteracja po słowniku


````julia
for (name, number) in phonebook
    println("$name -> $number")
end
````

### Typy kluczy i wartości

Każdy `Dict` ma typ klucza i typ wartości:

```julia
Dict{K, V}
```

Na przykład:


````julia
strict_phonebook = Dict{String, String}(
    "Anna" => "1234",
    "Boris" => "5678",
)

@show typeof(strict_phonebook)
````

### Ważne

Do `Dict{String,String}` nie możemy po prostu wstawić liczby całkowitej jako wartości.


````julia
error_detected = try
    strict_phonebook["Emergency"] = 911
    false
catch e
    e isa MethodError
end

@show error_detected
````

### Słownik heterogeniczny


````julia
flexible_phonebook = Dict{String, Any}(
    "Eugeniusz" => 8675309,
    "Konstanty" => "555-2368",
)

flexible_phonebook["Emergency"] = 911

@show flexible_phonebook
````

### Ważne

`Dict{String,Any}` daje elastyczność, ale traci część informacji typowej przydatnej kompilatorowi.

Jeżeli zestaw możliwych typów jest niewielki, często lepiej rozważyć:

```julia
Dict{String, Union{Int, String}}
```


````julia
typed_flexible_phonebook = Dict{String, Union{Int, String}}(
    "Eugeniusz" => 8675309,
    "Konstanty" => "555-2368",
)

typed_flexible_phonebook["Emergency"] = 911

@show typed_flexible_phonebook
@show typeof(typed_flexible_phonebook)
````

## **7. Zbiory — `Set`**

### Teoria

`Set` przechowuje **unikalne** wartości.

Przydatny jest wtedy, gdy interesuje nas przede wszystkim:

- przynależność elementu,
- usuwanie duplikatów,
- operacje zbiorowe.


````julia
numbers = Set([1, 2, 2, 3, 3, 3])

@show numbers
@show length(numbers)
````

### Sprawdzanie przynależności


````julia
@show 2 in numbers
@show 10 in numbers
````

### Dodawanie i usuwanie


````julia
push!(numbers, 4)
delete!(numbers, 1)

@show numbers
````

### Operacje zbiorowe


````julia
a = Set([1, 2, 3])
b = Set([3, 4, 5])

@show union(a, b)
@show intersect(a, b)
@show setdiff(a, b)
````

Any[1, "Julia", true]

## **22. `Vector{Any}` — elastyczność i koszt**

### Teoria

`Vector{Any}` może przechowywać praktycznie dowolne wartości:


````julia
mixed = Any[
    42,
    "Julia",
    true,
    [1, 2, 3],
]

@show mixed
````

### Co tracimy?

Bardzo szeroki typ elementów oznacza mniej informacji dla kompilatora.

Może to skutkować:

- częstszym dynamicznym dispatch,
- boxingiem,
- dodatkowymi alokacjami,
- gorszą przewidywalnością wydajności.

Dla krótkiego kodu edukacyjnego nie zawsze ma to znaczenie, ale w kodzie numerycznym i wydajnościowym jest istotne.


### Lepsza alternatywa — mały `Union`


````julia
id_or_name = Union{Int, String}[1, "Alice", 2, "Bob"]

@show id_or_name
@show eltype(id_or_name)
````

### Jeszcze lepsza alternatywa — struktura semantyczna

Jeżeli elementy mają stałe znaczenie, często lepszy jest:

- `Tuple`,
- `NamedTuple`,
- własny `struct`.

Nie używaj `Any` tylko dlatego, że jest najłatwiejszy.


## **23. Przypisanie nie zawsze oznacza kopię**

### Teoria

Dla mutowalnych kolekcji:


````julia
fibonacci = [1, 1, 2, 3, 5, 8]
some_more_numbers = fibonacci

some_more_numbers[1] = 999

@show fibonacci
@show some_more_numbers
````

### Analiza

Obie nazwy wskazują na **ten sam obiekt**.

To nazywamy aliasingiem.


### `copy`


````julia
fibonacci = [1, 1, 2, 3, 5, 8]
independent = copy(fibonacci)

independent[1] = 999

@show fibonacci
@show independent
````

## **24. `copy` a `deepcopy`**

### Teoria

Dla płaskiej tablicy `copy` zwykle wystarcza.

Dla struktur zagnieżdżonych `copy` kopiuje tylko zewnętrzny kontener.


````julia
nested = [[1, 2], [3, 4]]
shallow = copy(nested)

shallow[1][1] = 999

@show nested
@show shallow
````

Wewnętrzne wektory nadal były współdzielone.


### `deepcopy`


````julia
nested = [[1, 2], [3, 4]]
deep = deepcopy(nested)

deep[1][1] = 999

@show nested
@show deep
````

## **27. Iteracja po kolekcjach**

### Wektor


````julia
for value in [10, 20, 30]
    println(value)
end
````

### Indeks i wartość — `enumerate`


````julia
languages = ["Julia", "Python", "Rust"]

for (i, language) in enumerate(languages)
    println("$i -> $language")
end
````

### Słownik


````julia
d = Dict("a" => 1, "b" => 2)

for (key, value) in d
    println("$key -> $value")
end
````

### Macierz

Iteracja po macierzy przebiega po elementach w kolejności pamięci Julii, która jest **kolumnowa (column-major)**.

### Ważne — układ kolumnowy

Julia, podobnie jak Fortran i MATLAB, używa układu **column-major**.

To ma znaczenie dla wydajności w kodzie numerycznym, szczególnie w zagnieżdżonych pętlach.


## **28. `@assert` — prosta kontrola założeń**

### Teoria

Makro:

```julia
@assert warunek
```

sprawdza, czy wyrażenie jest prawdziwe.

Jeżeli jest `false`, Julia zgłasza `AssertionError`.


````julia
flexible_phonebook = Dict{String, Any}(
    "Emergency" => 911,
)

@assert flexible_phonebook["Emergency"] == 911

println("PASS — założenie jest spełnione.")
````

### Komunikat diagnostyczny

### Ważne

`@assert` jest przydatny do:

- testów,

## **29. Typowe błędy początkujących**

| Błąd | Przyczyna | Poprawne podejście |
|---|---|---|
| próba zmiany `Tuple` | krotki są niemutowalne | utwórz nową krotkę |
| `dict["brak"]` | klucz nie istnieje | `get`, `haskey` |
| dodanie `Int` do `Dict{String,String}` | typ wartości jest ograniczony | odpowiedni `V`, np. `Union` |
| mylenie `Vector{Vector}` z `Matrix` | inna struktura pamięci i indeksowania | sprawdź `typeof` i `ndims` |
| oczekiwanie, że `b = a` kopiuje tablicę | powstaje alias | `copy` lub `deepcopy` |
| używanie `Vector{Any}` bez potrzeby | tracimy informację typową | konkretny typ lub mały `Union` |
| slicing przy dużych danych bez potrzeby | powstają kopie | `view` / `@view` |
| zakładanie układu row-major | Julia jest column-major | projektuj pętle zgodnie z kolumnami |

### Przykładowe rozwiązanie 30.6

## **33. Podsumowanie lekcji**

Najważniejsze zasady Lesson 2:

1. `Tuple` i `NamedTuple` są niemutowalne.
2. `NamedTuple` jest dobrym lekkim rekordem.
3. `Dict` mapuje klucze na wartości.
4. Typ `Dict{K,V}` kontroluje legalne typy kluczy i wartości.
5. `Set` przechowuje wartości unikalne.

14. `copy` jest płytką kopią, `deepcopy` kopiuje także zagnieżdżone obiekty.
