## **1. Czym jest pętla?**

### Teoria

Pętla służy do wielokrotnego wykonywania fragmentu kodu.

Przykładowe zastosowania:

- wypisanie liczb od 1 do 100;
- przejście po wszystkich elementach tablicy;
- wielokrotne wykonywanie operacji aż do spełnienia warunku;
- budowanie macierzy;
- agregacja danych;
- przetwarzanie każdego wpisu w słowniku.


### Dwa podstawowe rodzaje pętli w tej lekcji

| Konstrukcja | Kiedy używać |
|---|---|
| `while` | gdy liczba iteracji nie jest z góry oczywista i zależy od warunku |
| `for` | gdy iterujemy po konkretnym zbiorze wartości lub iteratorze |


## **2. Pętla `while`**

### Składnia

```julia
while warunek
    instrukcje
end
```

Pętla wykonuje ciało tak długo, jak warunek ma wartość `true`.


### Przykład


````julia
n = 0

while n < 10
    n += 1
    println(n)
end

@show n
````

### Analiza

1. `n = 0` inicjalizuje licznik.
2. `while n < 10` sprawdza warunek przed każdą iteracją.
3. `n += 1` zwiększa licznik.
4. Gdy `n` osiąga `10`, kolejne sprawdzenie warunku zwraca `false`.
5. Pętla kończy działanie.


### Przykład — `while` i wektor


````julia
myfriends = ["Ted", "Robyn", "Barney", "Lily", "Marshall"]

i = 1

while i <= length(myfriends)
    println("Cześć $(myfriends[i])!")
    i += 1
end
````

### Ważne

Przy ręcznym indeksowaniu musisz sam pilnować:

- wartości początkowej indeksu;
- poprawnego warunku końcowego;
- zwiększania indeksu.

Dlatego dla typowej iteracji po kolekcji `for` jest zwykle prostszy i czytelniejszy.


## **3. Typowe pułapki pętli `while`**

### Typowy błąd — pętla nieskończona

Jeżeli zapomnisz zmienić zmienną kontrolującą warunek:

```julia
n = 0

while n < 10
    println(n)
end
```

warunek nigdy się nie zmieni i pętla będzie działać bez końca.


### Dobra praktyka

W pętli `while` zawsze sprawdź:

1. co ustala stan początkowy,
2. co zmienia stan,
3. co gwarantuje zakończenie pętli.


## **4. Pętla `for`**

### Składnia

```julia
for element in iterable
    instrukcje
end
```

`iterable` może być m.in.:

- zakresem,
- wektorem,
- napisem,
- słownikiem,
- generatorem,
- innym obiektem implementującym protokół iteracji.


### Przykład — zakres


````julia
for n in 1:10
    println(n)
end
````

### Przykład — wektor


````julia
myfriends = ["Ted", "Robyn", "Barney", "Lily", "Marshall"]

for friend in myfriends
    println("Cześć $friend, miło Cię widzieć!")
end
````

### Analiza

W tym przykładzie nie potrzebujemy ręcznie zarządzać indeksem.

Julia kolejno przypisuje każdy element `myfriends` do zmiennej `friend`.


## **5. Zakresy i iteratory**

### Zakres `1:10`


````julia
r = 1:10

@show r
@show typeof(r)
@show collect(r)
````

Zakres nie musi od razu tworzyć pełnej tablicy wartości w pamięci.


### Zakres z krokiem


````julia
for n in 2:2:10
    println(n)
end
````

### Zakres malejący


````julia
for n in 10:-2:2
    println(n)
end
````

### `range`


````julia
for x in range(0.0, 1.0; length=5)
    println(x)
end
````

## **6. Iteracja po wektorach i napisach**

### Wektor


````julia
languages = ["Julia", "Rust", "Python"]

for language in languages
    println(language)
end
````

### Napis


````julia
for c in "Julia 🚀"
    println(c)
end
````

### Ważne

Przy tekście Unicode iteracja:

```julia
for c in text
```

jest bezpieczniejsza niż ręczne odwoływanie się do `text[i]` dla `i = 1:length(text)`.


## **7. `enumerate` — indeks i wartość**

### Teoria

Jeżeli potrzebujesz jednocześnie numeru pozycji i wartości:


````julia
languages = ["Julia", "Rust", "Python"]

for (i, language) in enumerate(languages)
    println("$i -> $language")
end
````

### Ważne

`enumerate` numeruje iteracje od `1`, ale nie należy utożsamiać jego licznika ze wszystkimi możliwymi indeksami dowolnej struktury.

Dla typowych wektorów oba pojęcia często się pokrywają, lecz semantycznie są różne.


## **8. `zip` — równoległa iteracja po kilku kolekcjach**

### Przykład


````julia
names = ["Anna", "Piotr", "Ewa"]
scores = [92, 81, 96]

for (name, score) in zip(names, scores)
    println("$name -> $score")
end
````

### Analiza

`zip` łączy elementy z odpowiadających sobie pozycji.

Iteracja kończy się, gdy najkrótszy iterator zostanie wyczerpany.


## **9. `break` — natychmiastowe przerwanie pętli**

### Przykład


````julia
for n in 1:100
    if n == 7
        println("Znaleziono 7 — kończę.")
        break
    end
    println(n)
end
````

### Kiedy używać?

`break` jest przydatne np. podczas:

- wyszukiwania pierwszego pasującego elementu;
- oczekiwania na konkretny stan;
- ręcznego zakończenia pętli po wykryciu warunku.


## **10. `continue` — pominięcie bieżącej iteracji**

### Przykład


````julia
for n in 1:10
    if isodd(n)
        continue
    end
    println(n)
end
````

Pętla wypisze tylko liczby parzyste.


## **11. Zagnieżdżone pętle**

### Teoria

Pętla może znajdować się wewnątrz innej pętli.


````julia
for row in 1:3
    for col in 1:4
        println("row=$row, col=$col")
    end
end
````

### Skrócona składnia


````julia
for row in 1:3, col in 1:4
    println("row=$row, col=$col")
end
````

Obie wersje są logicznie równoważne.


## **12. Pętle i macierze**

### Cel

Zbudujemy tablicę dodawania:


````julia
m, n = 5, 5
A = fill(0, (m, n))

for j in 1:n
    for i in 1:m
        A[i, j] = i + j
    end
end

A
````

### Analiza

Wartość każdego elementu jest sumą indeksu wiersza i kolumny:

```julia
A[i, j] = i + j
```


### Column-major order

Julia przechowuje macierze kolumnami, podobnie jak Fortran i MATLAB.

Dlatego przy klasycznych zagnieżdżonych pętlach zwykle korzystniejsze jest:

```julia
for j in kolumny
    for i in wiersze
        A[i, j] = ...
    end
end
```

czyli zmienianie indeksu wiersza `i` w pętli wewnętrznej.


### Skrócona wersja tej samej pętli


````julia
B = fill(0, (m, n))

for j in 1:n, i in 1:m
    B[i, j] = i + j
end

@assert A == B
B
````

## **13. `eachindex` — idiomatyczna iteracja po indeksach**

### Teoria

Jeżeli faktycznie potrzebujesz indeksów, preferuj:


````julia
values = [10, 20, 30, 40]

for i in eachindex(values)
    println("indeks=$i, wartość=$(values[i])")
end
````

### Dlaczego `eachindex`?

`eachindex`:

- zwraca poprawne indeksy dla danej struktury;
- nie zakłada konkretnego sposobu indeksowania;
- jest dobrym wyborem w kodzie ogólnym.


## **14. Iteracja po słownikach**

### Przykład


````julia
phonebook = Dict(
    "Anna" => "111-222",
    "Piotr" => "333-444",
    "Ewa" => "555-666",
)

for (name, number) in phonebook
    println("$name -> $number")
end
````

### Ważne

Nie buduj logiki zależnej od arbitralnego „numeru pozycji” wpisu w słowniku.

Interesuje nas przede wszystkim para:

```text
klucz => wartość
```


## **15. Comprehensions — tworzenie kolekcji**

### Teoria

Comprehension pozwala tworzyć nową kolekcję za pomocą zwięzłej składni.


````julia
squares = [i^2 for i in 1:10]

@show squares
````

### Comprehension z warunkiem


````julia
even_squares = [i^2 for i in 1:20 if iseven(i)]

@show even_squares
````

### Macierz przez comprehension


````julia
C = [i + j for i in 1:m, j in 1:n]

@assert C == A
C
````

### Kiedy comprehension?

Świetnie nadaje się do tworzenia nowej kolekcji na podstawie wyrażenia.

Jeżeli natomiast wykonujesz wiele efektów ubocznych, logikę wieloetapową albo złożone sterowanie przepływem, zwykły `for` często będzie czytelniejszy.


## **16. Dict comprehension**

### Przykład


````julia
squares_dict = Dict(i => i^2 for i in 1:10)

@show squares_dict
@assert squares_dict[10] == 100
````

To bardzo naturalny sposób budowania słownika na podstawie reguły.


## **17. `map` kontra `for`**

### `map`

`map` stosuje funkcję do kolejnych elementów i zwraca wyniki.


````julia
values = 1:5
doubled = map(x -> 2x, values)

@show doubled
````

### Porównanie

Jeżeli chcesz **utworzyć wyniki**:

```julia
map(f, data)
```

lub comprehension bywają naturalne.

Jeżeli chcesz wykonywać głównie **efekty uboczne**, np. `println`, zwykły `for` jest zwykle czytelniejszy.


### Typowy błąd stylistyczny

To działa:

```julia
map(x -> println(x^2), 1:100)
```

ale `map` semantycznie służy do mapowania wartości na wyniki.

Do samego wypisywania lepiej:


````julia
for x in 1:5
    println(x^2)
end
````

## **18. Generatory i redukcje**

### Suma kwadratów

Zamiast tworzyć pośrednią tablicę:


````julia
sum_of_squares = sum(i^2 for i in 1:100)

@show sum_of_squares
@assert sum_of_squares == 338350
````

### Analiza

Wyrażenie:

```julia
(i^2 for i in 1:100)
```

jest generatorem.

`sum` może konsumować go bez konieczności tworzenia pełnej tablicy pośredniej.


## **19. Wydajność pętli w Julia**

### Ważne

W wielu językach wysokiego poziomu użytkownicy unikają pętli z obawy przed wydajnością.

W Julia pętle są pełnoprawnym, wydajnym narzędziem.

Nie musisz sztucznie „wektoryzować” każdej operacji tylko dlatego, że używasz języka wysokiego poziomu.


### Praktyczne wskazówki

Dla kodu wydajnościowego:

1. umieszczaj gorące pętle w funkcjach;
2. dbaj o stabilność typów;
3. unikaj niepotrzebnych alokacji;
4. korzystaj z `eachindex`, gdy potrzebujesz indeksów;
5. przy macierzach pamiętaj o układzie column-major.


````julia
function sum_squares_loop(n)
    total = zero(Int)
    for i in 1:n
        total += i^2
    end
    return total
end

@show sum_squares_loop(100)
````

## **20. Typowe błędy początkujących**

| Błąd | Przyczyna | Poprawne podejście |
|---|---|---|
| nieskończone `while` | stan nie jest aktualizowany | zmieniaj zmienną sterującą |
| ręczne `i = 1` bez potrzeby | komplikacja kodu | użyj `for element in collection` |
| iteracja po napisie przez `1:length(s)` | problem z Unicode | `for c in s` |
| używanie `map` tylko do `println` | niewłaściwa semantyka | zwykły `for` |
| tworzenie tablicy tylko po to, aby ją zsumować | zbędna alokacja | generator + `sum` |
| nieprawidłowa kolejność indeksów macierzy | gorsza lokalność pamięci | column-major |
| brak `break` w wyszukiwaniu pierwszego wyniku | zbędne iteracje | zakończ pętlę po znalezieniu |

## **22. Mini-projekt — analiza wyników kursantów**

````julia
passed = [
    student.name
    for student in students
    if student.score >= 80
]

@show passed
````

````julia
average_score = sum(student.score for student in students) / length(students)

@show average_score
````

````julia
first_excellent = nothing

for student in students
    if student.score > 95
        first_excellent = student
        break
    end
end

@show first_excellent
````
