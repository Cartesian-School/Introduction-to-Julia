# 1️⃣ Czym jest Julia?

## 📘 Teoria

**Julia** jest wysokopoziomowym, dynamicznym i wieloparadygmatowym językiem programowania zaprojektowanym z myślą o obliczeniach naukowych i numerycznych, ale nadającym się również do programowania ogólnego.

Język łączy:

- składnię zbliżoną do języków używanych w analizie danych i matematyce;
- **multiple dispatch** — jeden z najważniejszych mechanizmów Julii;
- wnioskowanie o typach;
- kompilację JIT opartą na LLVM;
- możliwość uzyskania wydajności zbliżonej do języków kompilowanych niskiego poziomu;
- wygodną pracę interaktywną w REPL i notebookach.

Julia jest szczególnie popularna w obszarach takich jak:

- obliczenia naukowe;
- algebra liniowa i metody numeryczne;
- symulacje;
- optymalizacja;
- analiza danych;
- uczenie maszynowe;
- modelowanie fizyczne i inżynierskie;
- high-performance computing.


## 💡 Idea — „dwa języki” nie są konieczne

W wielu środowiskach prototyp tworzy się w języku wysokiego poziomu, a krytyczne fragmenty przepisuje później do C/C++.

Julia została zaprojektowana tak, aby w wielu przypadkach ograniczyć ten podział: ten sam język może służyć zarówno do eksploracji, jak i do wydajnych obliczeń.

Nie oznacza to, że **każdy** kod Julia automatycznie jest szybki. Wydajność zależy od sposobu napisania programu, stabilności typów, alokacji pamięci i struktury algorytmu.


## 🧠 Pod maską — co dzieje się po uruchomieniu funkcji?

W dużym uproszczeniu:

1. Julia analizuje kod.
2. Wnioskuje typy wartości.
3. Generuje reprezentację pośrednią.
4. LLVM optymalizuje kod.
5. Powstaje kod maszynowy dla konkretnej specjalizacji funkcji.
6. Kolejne wywołania mogą korzystać z już skompilowanej wersji.

Dlatego **pierwsze wywołanie** funkcji może być zauważalnie wolniejsze od kolejnych — część czasu zajmuje kompilacja.

## 📘 `print` i `println`

- `print(...)` wypisuje dane bez automatycznego przejścia do nowej linii.
- `println(...)` działa podobnie, ale po wypisaniu danych dodaje znak nowej linii.


## 🧪 Przykład


````julia
print("Julia ")
print("jest ")
print("szybka.")
println()

println("Pierwsza linia")
println("Druga linia")
````

## 📘 `show` i `display`

`show` jest często używane do reprezentacji wartości w sposób bardziej techniczny, zbliżony do reprezentacji Julii.

`display` korzysta z systemu wyświetlania środowiska i jest szczególnie przydatne w notebookach, gdzie wynik może mieć reprezentację tekstową, HTML, graficzną itd.


````julia
text = "Julia\nLanguage"

println("println:")
println(text)

println("\nshow:")
show(text)

println("\n\ndisplay:")
display(text)
````

## 🔎 Analiza

Warto zwrócić uwagę, że:

- `println(text)` interpretuje znak nowej linii `\n`;
- `show(text)` pokazuje reprezentację napisu wraz z cudzysłowami i sekwencją `\n`;
- `display(text)` deleguje sposób prezentacji do systemu wyświetlania notebooka.


## 📘 Interpolacja napisów

W Julia zmienną można wstawić do napisu za pomocą `$`.

Dla prostego identyfikatora wystarczy:

```julia
"$name"
```

Dla całego wyrażenia używamy:

```julia
"$(2 + 3)"
```


````julia
name = "Julia"
version = VERSION

println("Uczę się języka $name.")
println("Wersja środowiska: $version")
println("2 + 3 = $(2 + 3)")
````

## 🐞 Typowy błąd

Początkujący często próbują pisać:

```julia
println("2 + 3 = $2 + 3")
```

`$` nie oznacza „policz wszystko do końca napisu”. Dla wyrażenia użyj `$(...)`.

## 📘 Operatory logiczne

- `&&` — logiczne AND z krótkim spięciem;
- `||` — logiczne OR z krótkim spięciem;

````julia
a = true
b = false

@show a && b
@show a || b
@show !a
````

## 🧠 Pod maską — short-circuit evaluation

`&&` i `||` nie muszą obliczać prawego argumentu.

Dla `A && B`:

- jeżeli `A` jest `false`, wynik jest już znany;
- `B` nie musi zostać wykonane.

Dla `A || B`:

- jeżeli `A` jest `true`, wynik jest już znany;
- `B` nie musi zostać wykonane.


````julia
function positive_message(x)
    x > 0 && println("$x jest dodatnie")
    return nothing
end

positive_message(5)
positive_message(-2)
````

## 🐞 Typowy błąd — `=` kontra `==`

- `=` oznacza **przypisanie**;
- `==` oznacza **porównanie wartości**.

```julia

x == 10
```

To dwie różne operacje.

# 1️⃣2️⃣ `convert` a `parse`

## 🎯 Cel

To rozróżnienie jest bardzo ważne już od pierwszej lekcji.


## 📘 `convert`

`convert(T, x)` służy do konwersji wartości `x` do typu `T`, jeżeli odpowiednia konwersja jest zdefiniowana.

Przykład:


````julia
converted = convert(Float64, 365)
@show converted typeof(converted)
````

## 📘 `parse`

`parse(T, text)` interpretuje **tekst** jako wartość typu `T`.

Przykład:


````julia
days_text = "365"
days_number = parse(Int, days_text)

@show days_text typeof(days_text)
@show days_number typeof(days_number)
````

## 🐞 Typowy błąd

To nie zadziała:

```julia
convert(Int, "365")
```

Napis `"365"` nie jest liczbą zapisaną w innym typie numerycznym. To **tekst**, który trzeba zinterpretować.

Dlatego używamy:

```julia
parse(Int, "365")
```


## 🧪 Bezpieczniejsza próba parsowania

Jeżeli dane mogą być błędne, można użyć `tryparse`.

Zamiast rzucać wyjątek, `tryparse` zwraca `nothing`, gdy tekstu nie da się poprawnie sparsować.


````julia
@show tryparse(Int, "365")
@show tryparse(Int, "365 dni")
````

# 🧾 Podsumowanie lekcji

Po tej lekcji powinieneś pamiętać:

1. Julia jest dynamicznym językiem z kompilacją JIT i silnym naciskiem na obliczenia naukowe.
2. REPL służy do interaktywnej pracy, a notebook łączy kod, tekst i wyniki.
3. `print` i `println` służą do wypisywania danych, `show` pokazuje reprezentację wartości, a `display` korzysta z systemu prezentacji środowiska.

5. `typeof(x)` pozwala sprawdzić typ wartości.
6. `"J"` i `'J'` to różne typy: `String` i `Char`.
7. Julia wspiera bogate nazwy Unicode.
8. Komentarze mają postać `# ...` oraz `#= ... =#`.
9. `/` oznacza dzielenie, a `÷` dzielenie całkowite.
10. `&&`, `||` i `!` realizują logikę warunkową.
11. `convert` konwertuje wartości między typami, a `parse` interpretuje tekst.
12. `tryparse` jest wygodne, gdy dane wejściowe mogą być niepoprawne.
