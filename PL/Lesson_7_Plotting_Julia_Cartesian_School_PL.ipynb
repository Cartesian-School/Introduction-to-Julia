## **1. Dlaczego wizualizacja danych jest ważna?**

### Teoria

Wykres pozwala zobaczyć zależności, które trudno wychwycić w surowej tabeli liczb.

Wizualizacja pomaga m.in.:

- wykrywać trendy;
- zauważać wartości odstające;
- porównywać serie;
- obserwować rozkłady;
- analizować zależności między zmiennymi;
- komunikować wyniki innym osobom.


### Ważne

Wykres nie jest dekoracją.

Dobry wykres powinien przede wszystkim:

1. odpowiadać na konkretne pytanie;
2. być czytelny;
3. nie wprowadzać w błąd;
4. mieć poprawnie opisane osie i jednostki.

## **4. Wykres liniowy**

### Teoria

Wykres liniowy jest dobrym wyborem, gdy chcemy pokazać:

- trend;
- przebieg funkcji;
- serię czasową;
- zmianę wartości względem uporządkowanej osi.


````julia
# using Plots

x = 0:0.1:2π
y = sin.(x)

# plot(x, y)
````

### Ważne — broadcasting

`sin.(x)` oznacza zastosowanie `sin` do każdego elementu kolekcji `x`.


## **5. Wykres punktowy — scatter plot**

### Teoria

Scatter plot dobrze nadaje się do obserwacji relacji między dwiema zmiennymi.


````julia
# using Plots

x = [1, 2, 3, 4, 5]
y = [2.1, 3.9, 6.2, 7.8, 10.1]

# scatter(x, y)
````

### Kiedy wybrać scatter?

Gdy dane są niezależnymi obserwacjami i łączenie ich linią mogłoby sugerować ciągłość, której w rzeczywistości nie ma.

## **15. Contour plot**

### Teoria

Contour plot pokazuje linie lub obszary o stałej wartości funkcji dwóch zmiennych.


````julia
# using Plots

x = -2:0.05:2
y = -2:0.05:2
z = [xv^2 + yv^2 for yv in y, xv in x]

# contour(x, y, z)
````

## **16. Wykres 3D**

### Surface plot


````julia
# using Plots

x = -2:0.1:2
y = -2:0.1:2
z = [exp(-(xv^2 + yv^2)) for yv in y, xv in x]

# surface(x, y, z)
````

| nadużywanie stylów | wizualny chaos | minimalizm |
| zapis wykresu bez sprawdzenia backendu | format może nie być wspierany | testuj eksport |

## **27. Mini-projekt — analiza sygnału**

### Cel

Zasymulujemy prosty sygnał będący sumą dwóch sinusoid.


````julia
t = 0:0.001:1
signal = sin.(2π * 5 .* t) .+ 0.5 .* sin.(2π * 20 .* t)

@show length(t)
@show length(signal)
````

### Wykres sygnału


````julia
# using Plots

# plot(
#     t,
#     signal,
#     xlabel="Czas [s]",
#     ylabel="Amplituda",
#     title="Sygnał złożony",
#     label="signal"
# )
````

### Fragment sygnału


````julia
mask = t .<= 0.2

short_t = t[mask]
short_signal = signal[mask]

@show length(short_t)
````

````julia
# using Plots

# scatter(
#     short_t,
#     short_signal,
#     markersize=2,
#     xlabel="Czas [s]",
#     ylabel="Amplituda",
#     title="Pierwsze 0.2 s sygnału",
#     label="samples"
# )
````
