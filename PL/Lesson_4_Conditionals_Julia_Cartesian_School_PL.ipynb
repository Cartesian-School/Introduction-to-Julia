## **1. Czym jest instrukcja warunkowa?**

### Teoria

Instrukcja warunkowa pozwala programowi wybrać różne ścieżki wykonania w zależności od wyniku wyrażenia logicznego.

Przykładowo:

- jeżeli użytkownik ma dostęp — pokaż dane;
- jeżeli liczba jest dodatnia — wykonaj obliczenie;
- jeżeli wynik przekracza próg — oznacz go jako zaliczony;
- jeżeli dane są niepoprawne — przerwij operację.


### Podstawowa idea

```text
jeżeli warunek jest prawdziwy
    wykonaj A
w przeciwnym razie
    wykonaj B
```


## **2. Typ `Bool` i warunki logiczne**

### Teoria

Warunek w Julia musi prowadzić do wartości logicznej:

- `true`,
- `false`.

Typ tych wartości to `Bool`.


````julia
@show true typeof(true)
@show false typeof(false)
````

### Przykład


````julia
x = 10

condition = x > 5

@show condition
@show typeof(condition)
````

### Ważne

Julia nie traktuje automatycznie liczb jako wartości logicznych.

To znaczy, że:

```julia
if 1
    ...
end
```

nie jest poprawnym warunkiem.

Warunek powinien jawnie zwracać `Bool`.


## **3. Instrukcja `if`**

### Składnia

```julia
if warunek
    instrukcje
end
```


### Przykład


````julia
temperature = 28

if temperature > 25
    println("Jest ciepło.")
end
````

### Analiza

Jeżeli `temperature > 25` zwróci `true`, kod wewnątrz bloku zostanie wykonany.

Jeżeli zwróci `false`, blok zostanie pominięty.


## **4. `if ... else`**

### Teoria

Jeżeli potrzebujemy dokładnie dwóch ścieżek:


````julia
age = 20

if age >= 18
    println("Osoba pełnoletnia.")
else
    println("Osoba niepełnoletnia.")
end
````

### Ważne

`else` nie ma warunku. Wykonuje się wtedy, gdy wcześniejszy warunek `if` był fałszywy.


## **5. `if ... elseif ... else`**

### Teoria

Dla kilku możliwych przypadków używamy `elseif`.


````julia
score = 87

if score >= 90
    grade = "A"
elseif score >= 80
    grade = "B"
elseif score >= 70
    grade = "C"
else
    grade = "D"
end

@show grade
````

### Analiza

Julia sprawdza warunki **od góry do dołu**.

Po znalezieniu pierwszego warunku, który jest `true`, wykonuje odpowiadający mu blok i pomija pozostałe.


## **6. Kolejność warunków ma znaczenie**

### Typowy błąd


````julia
score = 95

if score >= 60
    result = "zaliczone"
elseif score >= 90
    result = "bardzo dobry wynik"
else
    result = "niezaliczone"
end

@show result
````

### Analiza

Dla `score = 95` pierwszy warunek `score >= 60` jest już prawdziwy, więc `elseif score >= 90` nigdy nie zostanie sprawdzony.

Poprawna kolejność powinna przechodzić od najbardziej szczegółowego / najwyższego progu do bardziej ogólnego.


````julia
score = 95

if score >= 90
    result = "bardzo dobry wynik"
elseif score >= 60
    result = "zaliczone"
else
    result = "niezaliczone"
end

@show result
````

## **7. Operatory porównania**

| Znaczenie | Operator |
|---|---|
| równe | `==` |
| różne | `!=` |
| większe | `>` |
| mniejsze | `<` |
| większe lub równe | `>=` lub `≥` |
| mniejsze lub równe | `<=` lub `≤` |


````julia
@show 10 == 10
@show 10 != 5
@show 8 > 3
@show 4 <= 4
@show 7 ≥ 2
````

### Typowy błąd — `=` kontra `==`

- `=` oznacza przypisanie;
- `==` oznacza porównanie wartości.


## **8. Operatory logiczne**

| Operator | Znaczenie |
|---|---|
| `&&` | AND — oba warunki muszą być prawdziwe |
| `||` | OR — co najmniej jeden warunek ma być prawdziwy |
| `!` | NOT — negacja |


````julia
age = 25
has_ticket = true

can_enter = (age >= 18) && has_ticket

@show can_enter
````

````julia
is_weekend = false
is_holiday = true

day_off = is_weekend || is_holiday

@show day_off
````

````julia
enabled = true

@show !enabled
````

## **9. Porównania łańcuchowe**

Julia pozwala pisać warunki podobnie jak w matematyce.


````julia
x = 7

@show 1 < x < 10
````

### Zamiast

```julia
x > 1 && x < 10
```

możemy napisać:

```julia
1 < x < 10
```

To zwykle jest czytelniejsze.


## **10. Operator ternarny `? :`**

### Teoria

Operator ternarny ma postać:

```julia
warunek ? wartość_gdy_true : wartość_gdy_false
```

Jest odpowiednikiem prostego `if ... else`.


````julia
x = 10
y = 30

maximum_value = x > y ? x : y

@show maximum_value
````

### Odpowiednik przez `if`


````julia
maximum_value_if = if x > y
    x
else
    y
end

@show maximum_value_if
@assert maximum_value == maximum_value_if
````

### Ważne

Operator ternarny jest dobry dla **krótkich i prostych** warunków.

Jeżeli logika staje się złożona, zwykły `if` jest czytelniejszy.


## **11. Short-circuit evaluation**

### Teoria

Operatory `&&` i `||` działają z tzw. krótkim spięciem.

Dla:

```julia
A && B
```

jeżeli `A` jest `false`, Julia nie musi wykonywać `B`.

Dla:

```julia
A || B
```

jeżeli `A` jest `true`, Julia nie musi wykonywać `B`.


### Przykład


````julia
a = false

result = a && (println("Ten tekst nie zostanie wypisany"); true)

@show result
````

### Przykład z funkcją i efektem ubocznym


````julia
function side_effect()
    println("Funkcja side_effect została wykonana.")
    return true
end

a = false
result = a && side_effect()

@show result
````

Ponieważ `a == false`, wywołanie `side_effect()` zostaje pominięte.


## **12. Short-circuit jako bezpiecznik**

### Przykład — ochrona przed błędem


````julia
x = 0

(x != 0) && println(10 / x)

println("Program działa dalej.")
````

### Analiza

Gdy `x == 0`:

```julia
x != 0
```

jest `false`, więc prawa strona operatora `&&` nie zostaje wykonana.

Dzięki temu unikamy dzielenia przez zero.


### Przykład z `||`


````julia
username = ""

isempty(username) || println("Witaj, $username!")
````

Jeżeli napis jest pusty, prawa strona nie zostanie wykonana.


## **13. FizzBuzz — klasyczny przykład warunków**

### Cel

Dla liczby `N`:

- jeśli dzieli się przez 3 i 5 → `"FizzBuzz"`,
- jeśli tylko przez 3 → `"Fizz"`,
- jeśli tylko przez 5 → `"Buzz"`,
- w przeciwnym razie → liczba.


````julia
N = 15

if (N % 3 == 0) && (N % 5 == 0)
    println("FizzBuzz")
elseif N % 3 == 0
    println("Fizz")
elseif N % 5 == 0
    println("Buzz")
else
    println(N)
end
````

### Ważne

Najpierw sprawdzamy przypadek najbardziej szczegółowy:

```julia
N % 3 == 0 && N % 5 == 0
```

Gdybyśmy najpierw sprawdzili podzielność przez 3, liczba `15` zostałaby zaklasyfikowana jako samo `"Fizz"`.


### FizzBuzz jako funkcja


````julia
function fizzbuzz(n)
    if n % 15 == 0
        return "FizzBuzz"
    elseif n % 3 == 0
        return "Fizz"
    elseif n % 5 == 0
        return "Buzz"
    else
        return string(n)
    end
end

@show fizzbuzz(3)
@show fizzbuzz(5)
@show fizzbuzz(15)
@show fizzbuzz(7)
````

## **14. Warunki z `nothing`**

### Teoria

`nothing` jest często używane jako sygnał „braku wartości”.

Przykład: `tryparse`.


````julia
text = "123"
value = tryparse(Int, text)

if value === nothing
    println("Nie udało się odczytać liczby.")
else
    println("Odczytano liczbę: $value")
end
````

### Ważne

Do porównania z singletonem `nothing` idiomatycznie używamy:

```julia
value === nothing
```

lub:

```julia
value !== nothing
```


## **15. Warunki z kolekcjami**

### Przykład — pusty wektor


````julia
values = Int[]

if isempty(values)
    println("Brak danych.")
else
    println("Liczba elementów: $(length(values))")
end
````

### Przykład — przynależność


````julia
roles = ["admin", "editor", "viewer"]
role = "editor"

if role in roles
    println("Rola jest dozwolona.")
else
    println("Nieznana rola.")
end
````

### Przykład — słownik


````julia
config = Dict(
    "host" => "localhost",
    "port" => 8080,
)

if haskey(config, "port")
    println("Port: $(config["port"])")
else
    println("Brak konfiguracji portu.")
end
````

## **16. Warunki w funkcjach**

### Przykład


````julia
function classify_number(x)
    if x > 0
        return "dodatnia"
    elseif x < 0
        return "ujemna"
    else
        return "zero"
    end
end

@show classify_number(10)
@show classify_number(-3)
@show classify_number(0)
````

### Dobra praktyka — wcześniejszy `return`

Czasem czytelniej jest zakończyć funkcję wcześniej:


````julia
function safe_divide(a, b)
    if b == 0
        return nothing
    end

    return a / b
end

@show safe_divide(10, 2)
@show safe_divide(10, 0)
````

## **17. Warunki a multiple dispatch**

### Teoria

Nie każdy wybór zachowania powinien być zapisany jako `if`.

Julia ma bardzo silny mechanizm **multiple dispatch** — wybór metody funkcji może zależeć od typów argumentów.

Jeżeli rozgałęzienie wynika z **typu danych**, multiple dispatch często jest bardziej naturalne niż duży blok `if`.


### Podejście warunkowe


````julia
function describe_with_if(x)
    if x isa Int
        return "liczba całkowita"
    elseif x isa AbstractFloat
        return "liczba zmiennoprzecinkowa"
    elseif x isa AbstractString
        return "napis"
    else
        return "inny typ"
    end
end

@show describe_with_if(10)
@show describe_with_if(3.14)
@show describe_with_if("Julia")
````

### Podejście z multiple dispatch


````julia
describe(x::Int) = "liczba całkowita"
describe(x::AbstractFloat) = "liczba zmiennoprzecinkowa"
describe(x::AbstractString) = "napis"
describe(x) = "inny typ"

@show describe(10)
@show describe(3.14)
@show describe("Julia")
````

### Ważne

Używaj:

- `if` — gdy decyzja zależy od **wartości**;
- multiple dispatch — często wtedy, gdy zachowanie zależy od **typu**.

Nie jest to absolutna reguła, ale bardzo ważny styl projektowania w Julia.


## **18. Typowe błędy początkujących**

| Błąd | Przyczyna | Poprawne podejście |
|---|---|---|
| `if 1` | Julia oczekuje `Bool` | użyj jawnego porównania |
| `x = 5` zamiast `x == 5` | przypisanie zamiast porównania | `==` |
| zła kolejność `elseif` | zbyt ogólny warunek pojawia się za wcześnie | od szczegółowych do ogólnych |
| zbyt długi operator ternarny | spada czytelność | zwykły `if` |
| użycie `&` zamiast `&&` | brak short-circuit | `&&` dla logiki warunkowej |
| użycie `|` zamiast `||` | brak short-circuit | `||` dla logiki warunkowej |
| rozbudowany `if x isa ...` | logika zależy od typów | rozważ multiple dispatch |


### Ważne — `&&` a `&`

`&&`:

- działa na warunkach logicznych,
- wykonuje short-circuit.

`&`:

- jest operatorem bitowym / elementowym zależnie od typów,
- nie zapewnia takiego samego zachowania short-circuit.

Analogicznie:

- `||` ≠ `|`.

## **20. Mini-projekt — walidacja dostępu użytkownika**

### Dane

Mamy użytkownika opisującego:

- wiek,
- aktywność konta,
- rolę,
- posiadanie tokenu.


````julia
user = (
    name = "Anna",
    age = 29,
    active = true,
    role = "editor",
    has_token = true,
)
````

### Reguły dostępu

Użytkownik otrzymuje dostęp, jeżeli:

1. jest pełnoletni,
2. konto jest aktywne,
3. ma rolę `admin` lub `editor`,
4. posiada token.


````julia
allowed_role = user.role == "admin" || user.role == "editor"

can_access =
    user.age >= 18 &&
    user.active &&
    allowed_role &&
    user.has_token

if can_access
    println("Dostęp przyznany dla $(user.name).")
else
    println("Dostęp odrzucony.")
end
````

### Rozbudowana wersja diagnostyczna

Zamiast tylko `true/false`, możemy podać użytkownikowi konkretną przyczynę odmowy.


````julia
if user.age < 18
    reason = "użytkownik jest niepełnoletni"
elseif !user.active
    reason = "konto jest nieaktywne"
elseif !(user.role in ("admin", "editor"))
    reason = "rola nie ma dostępu"
elseif !user.has_token
    reason = "brak tokenu"
else
    reason = nothing
end

if reason === nothing
    println("Dostęp przyznany.")
else
    println("Dostęp odrzucony: $reason")
end
````

### Analiza

W projekcie wykorzystaliśmy:

- `if`,
- `elseif`,
- `else`,
- `&&`,
- `||`,
- `!`,
- `in`,
- `nothing`.

To typowy przykład, w którym czytelna logika warunkowa jest ważniejsza niż maksymalne skracanie kodu.
