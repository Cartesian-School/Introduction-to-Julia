## **1. `String` i `Char`**

### Teoria

W Julia tekst reprezentujemy przede wszystkim za pomocą dwóch typów:

| Typ | Przykład | Znaczenie |
|---|---|---|
| `String` | `"Julia"` | napis, czyli sekwencja znaków |
| `Char` | `'J'` | pojedynczy znak Unicode |

Najważniejsza różnica:

```julia
"J"   # String
'J'   # Char
```


### Przykład


````julia
word = "Julia"
letter = 'J'

@show word typeof(word)
@show letter typeof(letter)
````

### Ważne

Pojedyncze apostrofy `'...'` służą do zapisu **jednego znaku**.

Dlatego:

```julia
'Julia'
```

jest błędne.


### Typowy błąd

Początkujący często oczekują, że `'tekst'` będzie napisem — tak jak w niektórych innych językach.

W Julia:

- `"tekst"` → `String`,
- `'t'` → `Char`.

## **4. Interpolacja napisów**

### Teoria

Julia pozwala wstawiać wartości bezpośrednio do napisów za pomocą `$`.

Dla pojedynczej zmiennej:

```julia
"$name"
```

Dla całego wyrażenia:

```julia
"$(a + b)"
```


### Przykład


````julia
name = "Anna"
num_fingers = 10
num_toes = 10

println("Mam na imię $name.")
println("Mam $num_fingers palców u rąk i $num_toes palców u nóg.")
println("Łącznie to $(num_fingers + num_toes) palców.")
````

### Analiza

Wyrażenie:

```julia
"$(num_fingers + num_toes)"
```

najpierw oblicza sumę, a następnie umieszcza wynik w napisie.


### Typowy błąd

Nie zapisuj złożonego wyrażenia bez nawiasów:

```julia
"$a + $b = $a + $b"
```

Jeżeli chcesz policzyć wynik:

```julia
"$a + $b = $(a + b)"
```

## **8. Unicode i UTF-8**

### Teoria

Julia używa dla `String` kodowania **UTF-8**.

To bardzo ważne, ponieważ pojedynczy znak Unicode może zajmować różną liczbę bajtów.

Przykładowo:

- znak ASCII, np. `A`, zwykle zajmuje 1 bajt;
- wiele znaków cyrylicy zajmuje 2 bajty;
- wiele emoji zajmuje 4 bajty.

Dlatego liczba **znaków** i liczba **jednostek kodowania** nie muszą być równe.


### Przykład


````julia
en = "Hello"
pl = "Zażółć"
ru = "Привет"
emoji = "🚀"

println("Hello  -> znaki: ", length(en), ", bajty: ", ncodeunits(en))
println("Zażółć  -> znaki: ", length(pl), ", bajty: ", ncodeunits(pl))
println("Привет  -> znaki: ", length(ru), ", bajty: ", ncodeunits(ru))
println("🚀      -> znaki: ", length(emoji), ", bajty: ", ncodeunits(emoji))
````

### Ważne

- `length(s)` — liczba znaków;
- `ncodeunits(s)` — liczba jednostek kodowania UTF-8.

Nie należy traktować tych funkcji jako zamiennych.


## **9. Indeksowanie napisów**

### Teoria

To jeden z najważniejszych tematów tej lekcji.

W Julia indeks napisu nie oznacza automatycznie „numeru znaku”.

Indeks wskazuje **prawidłową pozycję w reprezentacji UTF-8**.

Dla czystego ASCII różnicy prawie nie widać:


````julia
ascii_text = "Hello"

@show ascii_text[1]
@show ascii_text[2]
@show ascii_text[5]
````

Dla tekstu Unicode kolejne poprawne indeksy mogą nie być kolejnymi liczbami całkowitymi.


````julia
ru = "Привет"

@show collect(eachindex(ru))
````

Dla `"Привет"` poprawne indeksy są zwykle:

```text
1, 3, 5, 7, 9, 11
```

Każda litera zajmuje tutaj dwa bajty.


### Przykład — poprawny indeks


````julia
ru = "Привет"
@show ru[1]
````

### Przykład — celowo błędny indeks

Poniższy kod jest przykładem błędu dydaktycznego:

```julia
ru[2]
```

Indeks `2` wpada do środka wielobajtowego znaku i Julia zgłosi `StringIndexError`.


````julia
ru = "Привет"

bad_index_detected = try
    ru[2]
    false
catch e
    e isa StringIndexError
end

@show bad_index_detected
````

### Ważne

Nie uczymy się przez celowe „rozbijanie” notebooka. Zamiast pozostawiać komórkę kończącą się nieobsłużonym wyjątkiem, przechwytujemy go i sprawdzamy jego typ.


## **10. Bezpieczne przechodzenie po tekście**

### Teoria

Jeśli potrzebujesz tylko znaków, najprostszy i najbezpieczniejszy zapis to:


````julia
for c in "Привет"
    print(c, " ")
end
println()
````

### Ważne

Nie stosuj dla ogólnego tekstu Unicode:

```julia
for i in 1:length(s)
    println(s[i])
end
```

Taki kod może działać dla ASCII, a przestać działać dla polskich liter, cyrylicy lub emoji.


### Gdy potrzebujesz indeksów — `eachindex`


````julia
text = "Zażółć 🚀"

for i in eachindex(text)
    println("indeks ", lpad(i, 2), " -> ", text[i])
end
````

`eachindex(text)` zwraca tylko takie indeksy, które są prawidłowe dla danego napisu.


## **11. `firstindex`, `lastindex`, `nextind`, `prevind`**

### Teoria

Do ręcznej nawigacji po tekście służą funkcje:

| Funkcja | Znaczenie |
|---|---|
| `firstindex(s)` | pierwszy poprawny indeks |
| `lastindex(s)` | indeks ostatniego znaku |
| `nextind(s, i)` | indeks następnego znaku |
| `prevind(s, i)` | indeks poprzedniego znaku |


````julia
text = "Привет"

@show firstindex(text)
@show lastindex(text)
@show length(text)
@show ncodeunits(text)
````

### Ważne

`lastindex(s)` i `length(s)` to nie to samo pojęcie.

Dla tekstu zawierającego znaki wielobajtowe wartości mogą być różne.


````julia
text = "Привет"

i = firstindex(text)
@show text[i]

i = nextind(text, i)
@show text[i]

i = nextind(text, i)
@show text[i]

i = prevind(text, i)
@show text[i]
````

### Ostatni znak — `end`


````julia
text = "Привет 🚀"

@show text[end]
@show text[end] == text[lastindex(text)]
````

## **20. Tekst i liczby**

## **23. Mini-projekt — normalizacja danych użytkownika**

### Cel

Załóżmy, że użytkownik podał:

```text
"   SIERGEJ.SOBolewski@Example.COM   "
```

Chcemy:

1. usunąć spacje na końcach;
2. zamienić tekst na małe litery;
3. sprawdzić, czy zawiera znak `@`;
4. rozdzielić login i domenę.


````julia
raw_email = "   SIERGEJ.SOBolewski@Example.COM   "

email = lowercase(strip(raw_email))
has_at = occursin("@", email)
parts = split(email, "@")

println("email: ", email)
println("zawiera @: ", has_at)

if has_at && length(parts) == 2
    login, domain = parts
    println("login : ", login)
    println("domena: ", domain)
end
````

### Analiza

W jednym krótkim przykładzie wykorzystaliśmy:

- `strip`,
- `lowercase`,
- `occursin`,
- `split`.

To typowa mała ścieżka przetwarzania tekstu.

### Zadanie 24.3 — Unicode i bezpieczne indeksy

Dany jest napis:

```julia
phrase = "Привет, Julia! 🚀"
```

Utwórz:

- `n_chars` — liczbę znaków;
- `n_bytes` — liczbę jednostek kodowania;
- `idxs` — wszystkie poprawne indeksy;
- `last_char` — ostatni znak;
- `bad_index` — `true`, jeśli `phrase[2]` wywołuje `StringIndexError`.


````julia
phrase = "Привет, Julia! 🚀"

# Twoje rozwiązanie:
````

### Przykładowe rozwiązanie 24.3


````julia
phrase = "Привет, Julia! 🚀"

n_chars = length(phrase)
n_bytes = ncodeunits(phrase)
idxs = collect(eachindex(phrase))
last_char = phrase[end]

bad_index = try
    phrase[2]
    false
catch e
    e isa StringIndexError
end

println("znaków          : ", n_chars)
println("jednostek UTF-8 : ", n_bytes)
println("indeksy         : ", idxs)
println("ostatni znak    : ", last_char)
println("StringIndexError: ", bad_index)

@assert length(idxs) == n_chars
@assert last_char == '🚀'
@assert bad_index === true
````

### Zadanie 24.5 — wyszukiwanie i zamiana

Dany tekst:

```julia
sentence = "Python jest szybki. Python jest wygodny."
```

1. sprawdź, czy zawiera `"Python"`;
2. zamień wszystkie wystąpienia `"Python"` na `"Julia"`;
3. sprawdź, czy nowy napis zaczyna się od `"Julia"`.


````julia
# Twoje rozwiązanie:
````

### Przykładowe rozwiązanie 24.5


````julia
sentence = "Python jest szybki. Python jest wygodny."

has_python = occursin("Python", sentence)
julia_sentence = replace(sentence, "Python" => "Julia")
starts_with_julia = startswith(julia_sentence, "Julia")

@show has_python
@show julia_sentence
@show starts_with_julia

@assert has_python
@assert starts_with_julia
````
