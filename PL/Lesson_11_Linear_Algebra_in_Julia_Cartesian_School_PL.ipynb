\[
A \in \mathbb{R}^{m\times n},\quad
B \in \mathbb{R}^{n\times p}

## **7. Transpozycja**

### `transpose(A)`


````julia
A = [
    1 2 3
    4 5 6
]

At = transpose(A)

@show At
@show typeof(At)
````

## **8. Adjoint — operator `'`**

Dla danych rzeczywistych:


````julia
A = rand(2, 3)

@show transpose(A)
@show A'
@show transpose(A) == A'
````

Dla danych zespolonych sytuacja jest inna.


````julia
Z = [
    1 + 2im  3 - 4im
    5 + 0im  6 + 7im
]

@show transpose(Z)
@show Z'
````

### Ważne

- `transpose(Z)` tylko zamienia osie,
- `Z'` wykonuje **sprzężenie zespolone + transpozycję**.

Dlatego `'` nazywamy adjoint.


## **9. `A' * A` i `A * A'`**

Te dwa iloczyny są bardzo często spotykane w analizie danych i metodach najmniejszych kwadratów.


````julia
A = rand(5, 3)

G1 = A' * A
G2 = A * A'

@show size(G1)
@show size(G2)
@show issymmetric(G1)
@show issymmetric(G2)
````

### Analiza

Dla `A` o rozmiarze `m × n`:

- `A' * A` ma wymiar `n × n`,
- `A * A'` ma wymiar `m × m`.

Obie macierze są hermitowskie/symetryczne i dodatnio półokreślone.


## **10. Układ kwadratowy**

Rozwiązujemy:

\[
Ax=b
\]

gdzie `A` jest kwadratowa.


````julia
A = [
    3.0 1.0
    1.0 2.0
]

b = [9.0, 8.0]

x = A \ b

@show x
@show A * x
@assert A * x ≈ b
````

### Ważne

Operator `\` jest idiomatycznym sposobem rozwiązania układu liniowego w Julia.


## **11. Układ nadokreślony — tall matrix**

Jeżeli `A` ma więcej wierszy niż kolumn:

\[
m > n
\]

zwykle mamy więcej równań niż niewiadomych.


````julia
A_tall = [
    1.0 1.0
    1.0 2.0
    1.0 3.0
    1.0 4.0
]

b_tall = [1.2, 1.9, 3.2, 3.9]

x_ls = A_tall \ b_tall

@show x_ls
````

## **12. Metoda najmniejszych kwadratów**

Dla układu nadokreślonego zwykle nie istnieje dokładne rozwiązanie `A*x == b`.

Julia zwraca rozwiązanie minimalizujące normę residual:


````julia
residual = A_tall * x_ls - b_tall

@show residual
@show norm(residual)
````

Matematycznie szukamy:

\[
\min_x \|Ax-b\|_2
\]


### Ważne

Nie trzeba ręcznie tworzyć równań normalnych:

```julia
(A' * A) \ (A' * b)
```

Bezpośrednie:

```julia
A \ b
```

jest zwykle stabilniejsze numerycznie.


## **13. Układ niedookreślony — wide matrix**

Jeżeli:

\[
m < n
\]

mamy mniej równań niż niewiadomych.


````julia
A_wide = [
    1.0 0.0 1.0
    0.0 1.0 1.0
]

b_wide = [1.0, 1.0]

x_wide = A_wide \ b_wide

@show x_wide
@show A_wide * x_wide
````

## **14. Rozwiązanie o minimalnej normie**

Dla pełnego rzędu wierszowego układu niedookreślonego istnieje nieskończenie wiele rozwiązań.

`A \ b` zwraca rozwiązanie zgodne z algorytmem Julii, typowo odpowiadające rozwiązaniu minimalnej normy w odpowiednim przypadku.


````julia
@show norm(x_wide)
@assert A_wide * x_wide ≈ b_wide
````

### Ważne

Nie interpretuj niedookreślonego układu jak układu kwadratowego.

Brak unikalności jest własnością matematyczną problemu, a nie błędem Julii.


## **15. Układ z niepełnym rzędem**

Macierz może mieć liniowo zależne kolumny lub wiersze.


````julia
A_rd = [
    1.0 2.0
    2.0 4.0
    3.0 6.0
]

@show rank(A_rd)
````

Jej rząd wynosi `1`, mimo że ma dwie kolumny.


````julia
b_rd = [1.0, 2.0, 3.0]

x_rd = A_rd \ b_rd

@show x_rd
@show norm(A_rd * x_rd - b_rd)
````

### Ważne

W układach rank-deficient rozwiązanie może być nieunikalne, a zachowanie zależy od użytego algorytmu/faktoryzacji.

Do dokładnej diagnozy często wykorzystuje się SVD.


## **16. Residual — jak ocenić rozwiązanie?**

Residual:

\[
r = Ax-b
\]

mierzy, jak dobrze rozwiązanie spełnia układ.


````julia
A = rand(8, 3)
x_true = [2.0, -1.0, 0.5]
b = A * x_true + 0.01 .* randn(8)

x_est = A \ b
r = A * x_est - b

@show norm(r)
````

### Dobra praktyka

Po rozwiązaniu problemu numerycznego często warto sprawdzić:

- `norm(A*x-b)`,
- względny residual,
- conditioning,
- rząd macierzy.


````julia
relative_residual = norm(r) / norm(b)

@show relative_residual
````

## **34. Typowe błędy początkujących**

| Błąd | Przyczyna | Poprawne podejście |
|---|---|---|
| `A .* B` zamiast `A * B` | pomylenie broadcastingu | `*` dla algebry liniowej |
| `A' == transpose(A)` dla Complex | adjoint sprzęga | rozróżniaj operacje |
| `inv(A)*b` | jawna odwrotność niepotrzebna | `A \ b` |
| równania normalne dla least squares | gorsza stabilność | `A \ b` |
| zakładanie unikalności wide system | za mało równań | analizuj przestrzeń rozwiązań |
| ignorowanie rank deficiency | macierz liniowo zależna | `rank`, `svd` |
| brak kontroli residual | rozwiązanie bez weryfikacji | `norm(A*x-b)` |
| konwersja wszystkiego do gęstej `Matrix` | utrata struktury | `Diagonal`, `Symmetric`, `Hermitian` |

## **36. Mini-projekt — regresja liniowa od podstaw**

### Cel

Zbudujemy prostą regresję liniową bez zewnętrznego pakietu statystycznego.

Model:

\[
y = \beta_0 + \beta_1 x + \varepsilon
\]


````julia
xdata = collect(1.0:10.0)
ydata = [2.2, 4.1, 6.0, 8.3, 10.2, 11.8, 14.1, 16.2, 17.9, 20.3]

X = hcat(ones(length(xdata)), xdata)

β = X \ ydata

@show β
````

### Predykcja


````julia
ŷ = X * β

@show ŷ
````

### Residual


````julia
r = ydata - ŷ

@show r
@show norm(r)
````

### Średni błąd kwadratowy


````julia
mse = sum(abs2, r) / length(r)

@show mse
````

### Analiza

W kilku liniach wykonaliśmy kompletny pipeline:

1. zbudowaliśmy macierz projektu `X`,
2. rozwiązaliśmy problem least squares przez `X \ y`,
3. obliczyliśmy predykcję,
4. policzyliśmy residual,
5. obliczyliśmy MSE.

To doskonały przykład praktycznego zastosowania algebry liniowej.
