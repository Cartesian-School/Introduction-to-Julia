## **1. Czym jest funkcja?**

### Teoria

Funkcja jest nazwanym lub anonimowym fragmentem programu, który:

1. może przyjmować dane wejściowe — **argumenty**;
2. wykonuje określone operacje;
3. zwraca wynik.

Funkcje pomagają:

- eliminować powtórzenia kodu;
- dzielić program na mniejsze części;
- testować logikę;
- ponownie wykorzystywać rozwiązania;
- budować abstrakcje.


### Przykład

Zamiast wielokrotnie pisać:

```julia
3 + 5
10 + 20
100 + 200
```

możemy stworzyć funkcję:


````julia
function add(x, y)
    return x + y
end

@show add(3, 5)
@show add(10, 20)
@show add(100, 200)
````

## **2. Deklaracja przez `function`**

### Składnia

```julia
function nazwa(argumenty)
    ciało
    return wynik
end
```


````julia
function greet(name)
    message = "Witaj, $name!"
    return message
end

println(greet("Anna"))
````

### Analiza

- `greet` — nazwa funkcji;
- `name` — parametr;
- `"Anna"` — argument przekazany podczas wywołania;
- `return` — zwraca wynik i kończy wykonanie funkcji.


## **3. Funkcje jednowierszowe**

### Teoria

Dla krótkich funkcji Julia oferuje składnię przypominającą równanie matematyczne:

To bardzo idiomatyczny zapis dla małych funkcji.


### Przykład z dwoma argumentami

Funkcja anonimowa nie musi mieć własnej nazwy.


### Więcej argumentów

### Kiedy używać?

Funkcje anonimowe są szczególnie wygodne jako krótkie argumenty dla:

- `map`,

- innych funkcji wyższego rzędu.


## **5. Parametry i argumenty**

### Teoria

**Parametr** to nazwa w definicji funkcji.

**Argument** to konkretna wartość przekazana przy wywołaniu.


````julia
function power(base, exponent)
    return base^exponent
end

@show power(2, 8)
````

W definicji:

- `base`, `exponent` — parametry.

W wywołaniu:

- `2`, `8` — argumenty.


## **6. Argumenty domyślne**

### Teoria

Parametr może mieć wartość domyślną:


````julia
function greet_user(name = "Gość")
    return "Witaj, $name!"
end

@show greet_user()
@show greet_user("Siergiej")
````

### Ważne

Argument domyślny pozwala wywołać funkcję z mniejszą liczbą argumentów.


## **7. Argumenty nazwane**

### Teoria

Argumenty nazwane zapisujemy po średniku `;`.


````julia
function create_message(name; language="pl", uppercase_text=false)
    text = language == "pl" ? "Witaj, $name!" : "Hello, $name!"
    return uppercase_text ? uppercase(text) : text
end

@show create_message("Anna")
@show create_message("Anna"; language="en")
@show create_message("Anna"; language="en", uppercase_text=true)
````

### Dlaczego są użyteczne?

Argumenty nazwane:

- poprawiają czytelność;
- są wygodne dla konfiguracji;
- pozwalają mieć wiele opcjonalnych ustawień.


## **8. Zmienna liczba argumentów — varargs**

### Teoria

Operator `...` pozwala funkcji przyjąć dowolną liczbę argumentów.


````julia
function add_all(values...)
    return sum(values)
end

@show add_all(1, 2, 3)
@show add_all(10, 20, 30, 40)
````

### Analiza

Wewnątrz funkcji `values` jest krotką (`Tuple`) przekazanych argumentów.


````julia
function inspect_args(values...)
    @show values
    @show typeof(values)
    return nothing
end

inspect_args(1, "Julia", true)
````

## **9. `return` i ostatnie wyrażenie**

### Teoria

Julia automatycznie zwraca wartość ostatniego wykonanego wyrażenia w funkcji.


````julia
function multiply_explicit(x, y)
    return x * y
end

function multiply_implicit(x, y)
    x * y
end

@assert multiply_explicit(3, 4) == multiply_implicit(3, 4)
````

### Ważne

Jawny `return` jest szczególnie przydatny:

- dla wcześniejszego zakończenia funkcji;
- gdy zwiększa czytelność;
- w bardziej rozbudowanej logice.


````julia
function safe_sqrt(x)
    x < 0 && return nothing
    return sqrt(x)
end

@show safe_sqrt(9)
@show safe_sqrt(-1)
````

## **10. Zwracanie wielu wartości**

### Teoria

Julia może zwrócić kilka wartości jako `Tuple`.


````julia
function minmax_values(values)
    return minimum(values), maximum(values)
end

lo, hi = minmax_values([8, 2, 11, 4])

@show lo hi
````

### NamedTuple jako czytelniejszy wynik


````julia
function stats(values)
    return (
        minimum = minimum(values),
        maximum = maximum(values),
        average = sum(values) / length(values),
    )
end

result = stats([10, 20, 30, 40])

@show result
@show result.average
````

## **11. Duck typing w Julia**

### Teoria

Duck typing oznacza:

> jeśli obiekt obsługuje operacje potrzebne funkcji, funkcja może z nim współpracować.

Nie musimy z góry deklarować konkretnego typu argumentu.


````julia
function combine(x, y)
    return x + y
end

@show combine(3, 5)
@show combine(2.5, 1.25)
````

### Ważne

W Julia „duck typing” nie oznacza braku typów. Julia ma rozbudowany system typów, ale funkcje często pozostają ogólne i wymagają jedynie obsługi konkretnych operacji.


### Przykład z własną semantyką operatora

Dla napisów w Julia konkatenacja używa `*`, a nie `+`, więc funkcja powinna być zaprojektowana zgodnie z operacją, której rzeczywiście potrzebuje.


````julia
function concatenate(x, y)
    return x * y
end

@show concatenate("Hello, ", "Julia!")
````

## **14. Multiple dispatch**

### Teoria

Multiple dispatch wybiera metodę na podstawie typów **wszystkich** argumentów uczestniczących w dispatch.

To jedna z centralnych cech Julii.


````julia
interact(x::Int, y::Int) = "Int + Int"
interact(x::Int, y::AbstractString) = "Int + String"
interact(x::AbstractString, y::Int) = "String + Int"
interact(x::AbstractString, y::AbstractString) = "String + String"

@show interact(1, 2)
@show interact(1, "Julia")
@show interact("Julia", 1)
@show interact("Julia", "Lang")
````

### Analiza

Zachowanie funkcji nie jest zakodowane w jednym wielkim `if`.

Zamiast tego definiujemy osobne, wyspecjalizowane metody.


## **15. Parametry typów i `where`**

### Teoria

Możemy wyrazić zależności pomiędzy typami argumentów.


````julia
function same_type_add(x::T, y::T) where {T<:Number}
    return x + y
end

@show same_type_add(2, 3)
@show same_type_add(2.5, 3.5)
````

### Analiza

`where {T<:Number}` oznacza, że:

- oba argumenty mają ten sam typ `T`;
- `T` musi być podtypem `Number`.


## **16. Funkcje mutujące i konwencja `!`**

### Teoria

W Julia znak `!` w nazwie funkcji jest konwencją sugerującą, że funkcja modyfikuje jeden lub więcej argumentów.


### Funkcja niemutująca


````julia
function doubled(values)
    return values .* 2
end

original = [1, 2, 3]
result = doubled(original)

@show original
@show result
````

### Funkcja mutująca


````julia
function double!(values)
    for i in eachindex(values)
        values[i] *= 2
    end
    return values
end

original = [1, 2, 3]
double!(original)

@show original
````

### Ważne

`!` nie zmienia semantyki języka i nie jest wymuszane przez kompilator. To konwencja nazewnicza bardzo mocno zakorzeniona w ekosystemie Julia.

## **18. Funkcje wyższego rzędu**

### Teoria

Funkcja wyższego rzędu to funkcja, która:

- przyjmuje funkcję jako argument,
- lub zwraca funkcję,
- lub robi obie te rzeczy.


````julia
function transform(f, values)
    return [f(x) for x in values]
end

@show transform(x -> x^2, 1:5)
````

````julia
values = [1, 2, 3, 4]

squares = map(x -> x^2, values)

@show squares
````

## **20. `filter`**

### Teoria

`filter(predicate, collection)` zachowuje tylko elementy, dla których predykat zwraca `true`.

### Predykat anonimowy

## **21. `reduce`**

### Teoria

`reduce` łączy kolejne elementy kolekcji przy użyciu funkcji dwuargumentowej.


````julia
values = [1, 2, 3, 4, 5]

total = reduce(+, values)
product = reduce(*, values)

@show total
@show product
````

### Dobra praktyka

Dla powszechnych operacji często preferujemy wyspecjalizowane funkcje:

```julia
sum(values)
prod(values)

```

Są czytelniejsze i jasno wyrażają intencję.


## **22. Mutujące funkcje wyższego rzędu — `map!`**

### Przykład


````julia
source = [1, 2, 3, 4]
destination = similar(source)

map!(x -> x^2, destination, source)

@show source
@show destination
````

### Analiza

`map!` zapisuje wyniki do istniejącej kolekcji docelowej, co może ograniczać dodatkowe alokacje.


## **23. Closures — domknięcia**

### Teoria

Funkcja anonimowa może „zapamiętać” zmienne z otaczającego ją zakresu.


````julia
function make_multiplier(factor)
    return x -> x * factor
end

times2 = make_multiplier(2)
times10 = make_multiplier(10)

@show times2(7)
@show times10(7)
````

### Analiza

`factor` nie jest argumentem funkcji zwróconej, ale pozostaje dostępny dzięki domknięciu.


## **24. `do` blocks**

### Teoria

Julia oferuje składnię `do`, która jest wygodna, gdy funkcja przyjmuje inną funkcję jako pierwszy argument.


````julia
result = map(1:5) do x
    x^2 + 1
end

@show result
````

### Analiza

Powyższy kod jest odpowiednikiem:


````julia
result2 = map(x -> x^2 + 1, 1:5)

@assert result == result2
````

## **25. Kompozycja funkcji**

### Teoria

Operator `∘` tworzy kompozycję funkcji:

```julia
(f ∘ g)(x) == f(g(x))
```


````julia
f(x) = x + 1
g(x) = x^2

h = f ∘ g

@show h(3)
@assert h(3) == f(g(3))
````

## **26. Operator potokowy `|>`**

### Teoria

`|>` przekazuje wartość po lewej stronie do funkcji po prawej.


````julia
result = 9 |> sqrt |> x -> x + 1

@show result
````

### Czytelność

Potoki są wygodne dla prostych sekwencji transformacji, ale nie należy ich nadużywać, gdy zwykłe wywołania funkcji są bardziej czytelne.


## **27. Rekurencja**

### Teoria

Funkcja rekurencyjna wywołuje samą siebie.

Musi posiadać **warunek bazowy**, który zatrzymuje rekurencję.


````julia
function factorial_recursive(n::Integer)
    n < 0 && throw(ArgumentError("n musi być nieujemne"))
    n <= 1 && return one(n)
    return n * factorial_recursive(n - 1)
end

@show factorial_recursive(5)
````

### Ważne

Rekurencja jest elegancka dla niektórych problemów, ale nie zawsze jest najlepszym rozwiązaniem wydajnościowym.

W wielu zadaniach iteracja będzie prostsza lub bardziej przewidywalna.


## **28. Zakres zmiennych w funkcjach**

### Teoria

Zmienne utworzone wewnątrz funkcji mają lokalny zakres.


````julia
global_value = 100

function scope_demo(x)
    local_value = x + 1
    return global_value + local_value
end

@show scope_demo(5)
````

### Dobra praktyka

Funkcje powinny, gdy to możliwe, otrzymywać dane przez argumenty zamiast polegać na globalnym stanie.

To poprawia:

- testowalność,
- czytelność,
- przewidywalność,
- często także wydajność.


## **29. Dokumentowanie funkcji**

### Teoria

W Julia dokumentację funkcji można umieścić bezpośrednio nad definicją jako docstring.


````julia
"""
    circle_area(r)

Zwraca pole koła o promieniu `r`.

# Argumenty
- `r`: promień koła, musi być nieujemny.

# Przykład
```julia
circle_area(2)
```
"""
function circle_area(r)
    r < 0 && throw(ArgumentError("Promień nie może być ujemny"))
    return π * r^2
end

@show circle_area(2)
````

### Dobra praktyka

Dobra dokumentacja powinna mówić:

- co robi funkcja;
- jakie przyjmuje argumenty;
- co zwraca;
- jakie ma ograniczenia;
- podać krótki przykład, jeśli to pomaga.


## **30. Typowe błędy początkujących**

| Błąd | Przyczyna | Poprawne podejście |
|---|---|---|
| dodawanie adnotacji typów wszędzie | mylenie typowania z optymalizacją | ograniczaj typy, gdy potrzebujesz kontraktu lub dispatch |
| wielki `if x isa ...` | zachowanie zależy od typów | rozważ multiple dispatch |
| brak `return` w złym miejscu | niejasna ścieżka zakończenia | użyj jawnego `return`, gdy poprawia czytelność |
| funkcja mutuje dane bez `!` w nazwie | łamie konwencję ekosystemu | nazwij ją np. `normalize!` |
| używanie `map` tylko dla efektów ubocznych | nieadekwatna semantyka | zwykły `for` |
| rekurencja bez warunku bazowego | nieskończone wywołania | zawsze zdefiniuj przypadek bazowy |
| nadmierne używanie globalnych zmiennych | słaba modularność | przekazuj dane jako argumenty |

### Przykładowe rozwiązanie 31.7


````julia
function increment!(values)
    for i in eachindex(values)
        values[i] += 1
    end
    return values
end

data = [1, 2, 3]
increment!(data)

@assert data == [2, 3, 4]
````

## **32. Mini-projekt — pipeline analizy wyników**

### Dane


````julia
scores = [45, 78, 92, 67, 88, 100, 59, 73]
````

### Cel

Zbudujemy zestaw małych funkcji, które:

1. walidują wynik;
2. klasyfikują wynik;
3. wybierają wyniki zaliczone;
4. liczą średnią.


````julia
is_valid_score(x) = 0 <= x <= 100
is_passed(x) = x >= 60

function grade(score)
    !is_valid_score(score) && return "INVALID"

    if score >= 90
        return "A"
    elseif score >= 80
        return "B"
    elseif score >= 70
        return "C"
    elseif score >= 60
        return "D"
    else
        return "F"
    end
end
````

### Zastosowanie


````julia
valid_scores = filter(is_valid_score, scores)
passed_scores = filter(is_passed, valid_scores)
grades = map(grade, valid_scores)
average = sum(valid_scores) / length(valid_scores)

@show valid_scores
@show passed_scores
@show grades
@show average
````

### Raport


````julia
report = [
    (score = score, grade = grade(score), passed = is_passed(score))
    for score in valid_scores
]

for row in report
    println("score=$(row.score), grade=$(row.grade), passed=$(row.passed)")
end
````

### Analiza

Mini-projekt pokazuje najważniejszą korzyść z funkcji:

zamiast jednego długiego bloku programu tworzymy małe, niezależne elementy o jednej odpowiedzialności.
