### **1️⃣ Использование ключевого слова `if`**  


В Julia условные операторы имеют следующий **синтаксис**: 

```julia

if *условие 1*
    *действие 1*

elseif *условие 2*
    *действие 2*

else
    *действие 3*
end

```
Этот синтаксис позволяет выполнять **разные действия** в зависимости от условий.  

<br>

**Пример:** 

Допустим нам необходимо реализовать тест ***Fizz-Buzz***

Допустим, у нас есть заданное число `N`. 

Нам необходимо:

- Вывести **"Fizz"**, если `N` делится на `3`.
    
- Вывести **"Buzz"**, если `N` делится на `5`.  
  
- Вывести **"FizzBuzz"**, если `N` делится **и на `3`, и на `5`**.  
  
- В противном случае вывести само число `N`.  

<br>

Зададим значение для `N` ниже:

## 🎯 После этого занятия вы сможете

- писать `if` / `elseif` / `else`;
- пользоваться тернарным оператором;
- применять сокращённое вычисление `&&` и `||`;
- выбирать между условной конструкцией и множественной диспетчеризацией.

````julia
N = 5
````

````julia
### **Пример кода для Fizz-Buzz**

N = 5  

if (N % 3 == 0) && (N % 5 == 0)  # `&&` означает "И"; % вычисляет остаток от деления
    println("Fizz-Buzz")
elseif N % 3 == 0
    println("Fizz")
elseif N % 5 == 0
    println("Buzz")
else
    println(N)
end  
````


Этот код проверяет значение `N` и:

- Выводит `"FizzBuzz"`, если `N` делится **и на `3`, и на `5`**.
  
- Выводит `"Fizz"`, если `N` делится **только на `3`**.
  
- Выводит `"Buzz"`, если `N` делится **только на `5`**.
  
- В противном случае просто выводит `N`.

### **2️⃣ Использование тернарных операторов**

Вместо блока `if-else` можно использовать **тернарный оператор** с синтаксисом:


` a ? b : c `  

это эквивалентно следующей конструкции:

<br>

```julia
if a
    b
else
    c
end
```

<br>

💡 Тернарный оператор позволяет записывать простые условные выражения в **более компактной форме**, улучшая читаемость кода.

<br>

Теперь предположим, что мы хотим вернуть большее из двух чисел. Укажите здесь значения `x` и `y`:

````julia
x = 10
y = 30
````

📌 Используя ключевые слова `if` и `else`, мы могли бы написать:

````julia
if x > y
    x
else
    y
end
````

📌 А в виде **тернарного оператора** условие выглядит так:

````julia
(x > y) ? x : y
````

Это эквивалентно следующему условному оператору `if-else`:  

```julia
if x > y
    x
else
    y
end
```

Тернарный оператор проверяет **условие `(x > y)`**:  

- Если **условие истинно**, возвращается `x`.  
  
- Если **условие ложно**, возвращается `y`.  
 
  <br>

💡 Тернарный оператор делает код **короче и удобнее** для простых выражений!

### **3️⃣ Короткое замыкание (Short-circuit evaluation) в Julia**  

**Короткое замыкание** — это концепция в программировании, при которой компилятор **пропускает** выполнение или вычисление некоторых подвыражений в логическом выражении, если результат уже определён.  

📌 **Как это работает?**  

Когда Julia встречает логическое выражение **с несколькими условиями**, она **перестаёт проверять оставшиеся части**, как только становится очевидным, каким будет конечный результат.  

**Например:** 

- В выражении `a && b` **если `a == false`**, то весь результат уже **должен быть `false`**, поэтому **`b` даже не вычисляется**.  
  
- Аналогично, в выражении `a || b` **если `a == true`**, то весь результат уже **будет `true`**, и `b` снова **не вычисляется**.  
  

💡  Это позволяет **экономить вычислительные ресурсы** и **избегать ненужных операций**.  

````julia
# **Пример работы в Julia**

a = false
b = true

result = a && (println("Это не выведется") || b)
````

📌 Как результат вы должны увидеть, что программа **ничего не напечатает**, потому что `a == false`, и Julia **даже не выполняет часть `println(...)`**.  

<br>

##### **Пример с функцией и побочными эффектами**

Иногда второе выражение (`b`) может быть **функцией с побочными эффектами**, например, с `println` или изменением переменных.  

В таком случае **эта функция даже не вызовется**, если её выполнение не влияет на итоговый результат.

````julia
function side_effect()
    println("Эта функция не вызовется, если a == false")
    return true
end

a = false
result = a && side_effect()  # `side_effect()` не выполняется!
````

📌 В результате *ничего не произойдёт*, потому что `a == false`, и Julia **не выполняет `side_effect()`**.

<br>

#### 🔹 **Применение короткого замыкания**


1. **Оптимизация производительности**  
   
   - Условие `b` может быть сложным (например, вызов сложных вычислений), и если `a` уже определяет результат, вычисления `b` можно пропустить.  

2. **Избежание ошибок**  
   
   - Например, если `b` содержит деление на ноль, а `a` гарантирует, что оно не потребуется, программа не выдаст ошибку.  

````julia
#Пример:

x = 0
(x != 0) && (println(10 / x))  # Деление на ноль не произойдёт, так как `x != 0` ложно
````

##### 📌 **Выводы**

✅ **Короткое замыкание** помогает **ускорить код**, **избежать ненужных вычислений** и **предотвратить ошибки**.   

✅ В Julia это работает **автоматически**, когда используются операторы `&&` (логическое И) и `||` (логическое ИЛИ).

В Julia можно использовать оператор `if-else` для проверки чётности числа:

````julia
# Определяем число
N = 7  # Можно заменить на любое число

# Проверяем чётность
if N % 2 == 0
    println(N)  # Выводим число, если оно чётное
else
    println("odd")  # Выводим "odd", если число нечётное
end
````
