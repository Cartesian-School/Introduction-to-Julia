## **1. Czym są metody numeryczne?**

Metody numeryczne służą do znajdowania **przybliżonych rozwiązań** problemów, których:

- nie da się łatwo rozwiązać analitycznie,
- rozwiązanie analityczne jest zbyt kosztowne,
- dane pochodzą z pomiarów,
- rozwiązanie istnieje tylko w postaci algorytmu iteracyjnego.

Przykłady:

- pierwiastki równań,
- całki,
- pochodne,
- układy równań,
- równania różniczkowe,
- optymalizacja.


## **2. Float64 i ograniczona precyzja**


```julia
x = 0.1 + 0.2

@show x
@show x == 0.3
@show isapprox(x, 0.3)
```

## **3. Błąd bezwzględny**

Dla wartości dokładnej `x` i przybliżenia `x̂`:

\[
E_{abs} = |x - \hat{x}|
\]


```julia
exact = sqrt(2)
approx = 1.414

abs_error = abs(exact - approx)

@show abs_error
```

## **4. Błąd względny**

\[
E_{rel} = \frac{|x-\hat{x}|}{|x|}
\]


```julia
rel_error = abs(exact - approx) / abs(exact)

@show rel_error
```

### Ważne

Błąd względny jest często bardziej informacyjny niż bezwzględny, ponieważ uwzględnia skalę wartości.

## **7. Stabilność numeryczna**

Algorytm stabilny numerycznie nie wzmacnia nadmiernie małych błędów zaokrągleń.

To cecha **algorytmu**, nie tylko języka programowania.


### Przykład — alternatywna postać wyrażenia

Dla bardzo małego `x`, wyrażenie:

\[
\sqrt{1+x}-1
\]

może tracić precyzję.

Lepsza algebraicznie równoważna forma:

\[
\frac{x}{\sqrt{1+x}+1}
\]


```julia
x = 1e-12

naive = sqrt(1 + x) - 1
stable = x / (sqrt(1 + x) + 1)

@show naive
@show stable
```

## **8. Conditioning problemu**

### Teoria

Conditioning mówi, jak bardzo wynik problemu reaguje na małą zmianę danych wejściowych.

- dobrze uwarunkowany problem → małe zmiany danych dają małe zmiany wyniku,
- źle uwarunkowany problem → mały błąd danych może mocno zmienić wynik.

To właściwość **problemu matematycznego**, a nie implementacji.


### Przykład z macierzą Hilberta


```julia
using LinearAlgebra

function hilbert_matrix(n)
    [1.0 / (i + j - 1) for i in 1:n, j in 1:n]
end

H = hilbert_matrix(8)

@show cond(H)
```

### Ważne

Stabilny algorytm nie naprawi źle uwarunkowanego problemu, ale może uniknąć dodatkowego pogorszenia wyniku.


## **9. Różnice skończone**

Pochodną można aproksymować przez różnicę w przód:

\[
f'(x) \approx \frac{f(x+h)-f(x)}{h}
\]


```julia
f(x) = sin(x)

function derivative_forward(f, x, h)
    (f(x + h) - f(x)) / h
end

approx = derivative_forward(f, 1.0, 1e-5)
exact = cos(1.0)

@show approx
@show exact
@show abs(approx - exact)
```

## **10. Różnica centralna**

Lepszą aproksymacją jest często:

\[
f'(x) \approx \frac{f(x+h)-f(x-h)}{2h}
\]


```julia
function derivative_central(f, x, h)
    (f(x + h) - f(x - h)) / (2h)
end

approx_central = derivative_central(f, 1.0, 1e-5)

@show approx_central
@show abs(approx_central - exact)
```

## **11. Za mały krok też jest zły**

Zmniejszanie `h` nie poprawia wyniku bez końca.

Dla bardzo małego `h` zaczynają dominować błędy zaokrągleń.


```julia
hs = 10.0 .^ (-1:-1:-15)

errors = [
    abs(derivative_central(sin, 1.0, h) - cos(1.0))
    for h in hs
]

for (h, err) in zip(hs, errors)
    println("h=$h  error=$err")
end
```

## **13. Metoda bisekcji**

### Teoria

Jeżeli funkcja ciągła zmienia znak na przedziale `[a,b]`, możemy wielokrotnie dzielić przedział na pół.


```julia
function bisection(f, a, b; atol=1e-12, maxiter=1000)
    fa = f(a)
    fb = f(b)

    fa == 0 && return a
    fb == 0 && return b
    signbit(fa) == signbit(fb) &&
        throw(ArgumentError("Funkcja musi zmieniać znak na końcach przedziału."))

    left, right = a, b

    for _ in 1:maxiter
        mid = (left + right) / 2
        fm = f(mid)

        if abs(fm) <= atol || abs(right - left) <= atol
            return mid
        end

        if signbit(f(left)) != signbit(fm)
            right = mid
        else
            left = mid
        end
    end

    return (left + right) / 2
end

root_bisect = bisection(x -> x^2 - 2, 1.0, 2.0)

@show root_bisect
@show sqrt(2)
```

### Zaleta

Bisekcja jest wolniejsza niż niektóre metody, ale bardzo przewidywalna, gdy spełnione są jej założenia.


## **14. Metoda Newtona**

Wzór iteracyjny:

\[
x_{n+1}=x_n-\frac{f(x_n)}{f'(x_n)}
\]


```julia
function newton(f, df, x0; atol=1e-12, maxiter=100)
    x = x0

    for _ in 1:maxiter
        fx = f(x)

        abs(fx) <= atol && return x

        dfx = df(x)
        abs(dfx) <= eps(Float64) &&
            throw(ArgumentError("Pochodna zbyt bliska zeru."))

        x_new = x - fx / dfx

        abs(x_new - x) <= atol && return x_new
        x = x_new
    end

    return x
end

froot(x) = x^2 - 2
dfroot(x) = 2x

root_newton = newton(froot, dfroot, 1.0)

@show root_newton
```

## **15. Bisekcja vs Newton**

| Cecha | Bisekcja | Newton |
|---|---|---|

| niezawodność | wysoka przy spełnionych założeniach | zależy od startu i pochodnej |

## **17. Metoda prostokątów**


```julia
function midpoint_rule(f, a, b, n)
    h = (b - a) / n
    total = zero(Float64)

    for i in 0:n-1
        xmid = a + (i + 0.5) * h
        total += f(xmid)
    end

    return h * total
end

I_mid = midpoint_rule(sin, 0.0, π, 1000)

@show I_mid
@show abs(I_mid - 2.0)
```

## **18. Metoda trapezów**


```julia
function trapezoidal_rule(f, a, b, n)
    h = (b - a) / n
    total = (f(a) + f(b)) / 2

    for i in 1:n-1
        total += f(a + i*h)
    end

    return h * total
end

I_trap = trapezoidal_rule(sin, 0.0, π, 1000)

@show I_trap
@show abs(I_trap - 2.0)
```

## **19. Metoda Simpsona**


```julia
function simpson_rule(f, a, b, n)
    iseven(n) || throw(ArgumentError("n musi być parzyste"))

    h = (b - a) / n
    total = f(a) + f(b)

    for i in 1:n-1
        x = a + i*h
        total += (iseven(i) ? 2 : 4) * f(x)
    end

    return h * total / 3
end

I_simpson = simpson_rule(sin, 0.0, π, 100)

@show I_simpson
@show abs(I_simpson - 2.0)
```

## **21. Interpolacja liniowa**

Dla dwóch punktów:

\[
(x_0,y_0),\quad(x_1,y_1)
\]

interpolacja liniowa ma postać:

\[
y(x)=y_0+\frac{x-x_0}{x_1-x_0}(y_1-y_0)
\]


```julia
function lerp(x0, y0, x1, y1, x)
    x0 == x1 && throw(ArgumentError("x0 i x1 muszą być różne"))
    y0 + (x - x0) * (y1 - y0) / (x1 - x0)
end

@show lerp(0.0, 0.0, 10.0, 20.0, 2.5)
```

## **22. Interpolacja wielomianowa — ostrożnie**

Wielomian wysokiego stopnia przechodzący przez wiele punktów może powodować silne oscylacje.

To klasyczny przykład, że „bardziej złożony model” nie musi oznaczać lepszego rozwiązania numerycznego.

W praktyce często preferuje się splajny lub interpolacje lokalne.

## **24. Równania różniczkowe zwyczajne — ODE**

Rozważmy:

\[
\frac{dy}{dt}=-ky
\]

z warunkiem:

\[
y(0)=y_0
\]

Dokładne rozwiązanie:

\[
y(t)=y_0e^{-kt}
\]


## **25. Metoda Eulera**

\[
y_{n+1}=y_n+h f(t_n,y_n)
\]


```julia
function euler(f, tspan, y0, h)
    t0, t1 = tspan
    n = Int(floor((t1 - t0) / h))

    ts = Vector{Float64}(undef, n + 1)
    ys = Vector{Float64}(undef, n + 1)

    ts[1] = t0
    ys[1] = y0

    for i in 1:n
        ts[i+1] = ts[i] + h
        ys[i+1] = ys[i] + h * f(ts[i], ys[i])
    end

    return ts, ys
end

k = 2.0
fode(t, y) = -k*y

ts_e, ys_e = euler(fode, (0.0, 2.0), 1.0, 0.1)

@show ys_e[end]
```

### Porównanie z rozwiązaniem dokładnym


```julia
exact_end = exp(-k * ts_e[end])

@show exact_end
@show abs(ys_e[end] - exact_end)
```

## **26. RK4 — Runge-Kutta 4**

RK4 jest znacznie dokładniejszą metodą jawnego całkowania ODE.


```julia
function rk4(f, tspan, y0, h)
    t0, t1 = tspan
    n = Int(floor((t1 - t0) / h))

    ts = Vector{Float64}(undef, n + 1)
    ys = Vector{Float64}(undef, n + 1)

    ts[1] = t0
    ys[1] = y0

    for i in 1:n
        t = ts[i]
        y = ys[i]

        k1 = f(t, y)
        k2 = f(t + h/2, y + h*k1/2)
        k3 = f(t + h/2, y + h*k2/2)
        k4 = f(t + h, y + h*k3)

        ts[i+1] = t + h
        ys[i+1] = y + h*(k1 + 2k2 + 2k3 + k4)/6
    end

    return ts, ys
end

ts_rk, ys_rk = rk4(fode, (0.0, 2.0), 1.0, 0.1)

@show ys_rk[end]
@show abs(ys_rk[end] - exp(-k*ts_rk[end]))
```

## **27. Błąd lokalny i globalny**

### Teoria

- **błąd lokalny** — błąd po jednym kroku metody,
- **błąd globalny** — skumulowany błąd po wielu krokach.

Dla metody Eulera błąd globalny jest rzędu `O(h)`.

Dla klasycznego RK4 błąd globalny jest rzędu `O(h^4)`.


## **28. Stabilność kroku**

Zbyt duży krok `h` może:

- zwiększyć błąd,

Zmniejszanie kroku zwiększa koszt obliczeń, więc zawsze istnieje kompromis między dokładnością i czasem.

## **30. Typowe błędy**

| Błąd | Problem | Poprawne podejście |
|---|---|---|
| `Float64 == exact` | niepewne porównanie | `isapprox` |
| bardzo małe `h` | dominacja round-off | testuj optimum kroku |
| Newton bez kontroli pochodnej | możliwy podział przez 0 | guard condition |
| bisekcja bez zmiany znaku | brak gwarancji | sprawdź końce |
| `inv(A)` w regresji | słabsza stabilność | `A\b` |
| Euler z dużym krokiem | duży błąd / niestabilność | zmniejsz krok |
| brak walidacji | wynik może być pozornie poprawny | residual / convergence test |

## **32. Mini-projekt — model chłodzenia Newtona**

### Problem

Prawo chłodzenia Newtona:

\[
\frac{dT}{dt}=-k(T-T_{env})
\]

gdzie:

- `T` — temperatura obiektu,
- `T_env` — temperatura otoczenia,
- `k` — współczynnik chłodzenia.


```julia
T_env = 20.0
T0 = 90.0
k = 0.12

cooling(t, T) = -k * (T - T_env)

ts, Ts = rk4(cooling, (0.0, 30.0), T0, 0.1)

@show Ts[1]
@show Ts[end]
```

### Rozwiązanie dokładne

\[
T(t)=T_{env}+(T_0-T_{env})e^{-kt}
\]


```julia
T_exact(t) = T_env + (T0 - T_env) * exp(-k*t)

error_end = abs(Ts[end] - T_exact(ts[end]))

@show error_end
```

### Czas osiągnięcia 30°C

Znajdźmy czas numerycznie przez bisekcję.


```julia
temperature_difference(t) = T_exact(t) - 30.0

t30 = bisection(temperature_difference, 0.0, 100.0)

@show t30
@show T_exact(t30)
```

### Analiza

W jednym projekcie użyliśmy:

- równania różniczkowego,
- RK4,
- rozwiązania analitycznego do walidacji,
- bisekcji do rozwiązania równania nieliniowego,
- analizy błędu.

To jest typowy workflow w obliczeniach naukowych.


## **33. Checkpoint końcowy**

Odpowiedz bez uruchamiania kodu:
