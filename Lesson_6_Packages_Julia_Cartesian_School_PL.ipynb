## **1. Czym jest pakiet?**

### Teoria

Pakiet w Julia to zorganizowany zestaw kodu przeznaczony do ponownego użycia.

Pakiet może zawierać:

- moduły;
- funkcje;
- typy;
- testy;
- dokumentację;
- zależności od innych pakietów;
- metadane projektu.


### Pakiet, moduł i środowisko — nie są tym samym

| Pojęcie | Znaczenie |
|---|---|
| **moduł** | przestrzeń nazw w kodzie Julia |
| **pakiet** | dystrybuowany projekt Julia, zwykle zawierający moduł |
| **środowisko** | zestaw zależności i wersji używanych przez konkretny projekt |


## **2. Ekosystem pakietów Julia**

### Teoria

Julia posiada rozbudowany ekosystem pakietów obejmujący m.in.:

- analizę danych;
- statystykę;
- uczenie maszynowe;
- obliczenia naukowe;
- optymalizację;
- wizualizację;
- równania różniczkowe;
- przetwarzanie sygnałów;
- programowanie GPU;
- interoperacyjność z Python, R, C i innymi językami.

Z punktu widzenia programisty najważniejszym narzędziem do zarządzania zależnościami jest wbudowany menedżer **Pkg**.


### Ważne

Nie instalujemy pakietów „do Julii jako całości” bez zastanowienia.

Profesjonalna praca opiera się na **środowiskach projektowych**, dzięki którym każdy projekt może posiadać własny zestaw zależności.


## **3. REPL i jego tryby**

### Teoria

REPL oznacza:

**Read → Eval → Print → Loop**

To interaktywna konsola Julii.

Po uruchomieniu:

```sh
julia
```

otrzymujemy:

```text
julia>
```


### Główne tryby REPL

| Tryb | Jak wejść | Do czego służy |
|---|---|---|
| Julia mode | domyślny | wykonywanie kodu Julia |
| Pkg mode | `]` | zarządzanie pakietami |
| Help mode | `?` | dokumentacja |
| Shell mode | `;` | polecenia systemowe |


### Julia mode


```julia
julia> 2 + 3
5

julia> sqrt(16)
4.0
```


### Help mode


```text
julia> ?
help?> sqrt
```

Pozwala szybko sprawdzić dokumentację funkcji lub typu.


### Shell mode


```text
julia> ;
shell> pwd
```

Tryb powłoki służy do wykonywania poleceń systemowych bez wychodzenia z REPL.


### Ważne

Shell mode zależy od systemu operacyjnego. Polecenia dostępne w Linux/macOS mogą różnić się od poleceń Windows.


## **4. Tryb pakietów `pkg>`**

### Wejście do trybu

W REPL naciśnij:

```text
]
```

Zobaczysz prompt podobny do:

```text
(@v1.x) pkg>
```


### Najważniejsze komendy

| Polecenie | Zastosowanie |
|---|---|
| `add PackageName` | dodaje pakiet |
| `rm PackageName` | usuwa pakiet |
| `update` | aktualizuje zależności |
| `status` | pokazuje stan środowiska |
| `activate .` | aktywuje środowisko w bieżącym katalogu |
| `instantiate` | odtwarza środowisko |
| `resolve` | ponownie rozwiązuje zależności |
| `precompile` | prekompiluje zależności |
| `pin PackageName` | blokuje wersję pakietu |
| `free PackageName` | usuwa pin / wraca do zwykłego śledzenia |
| `develop PackageName` | używa pakietu w trybie developerskim |


### Wyjście z Pkg mode

Najczęściej:

- `Backspace`,
- albo `Ctrl+C`.


## **5. Moduł `Pkg` i API programistyczne**

### Teoria

Większość operacji z Pkg mode ma odpowiednik w kodzie Julia.

Najpierw:


````julia
using Pkg
````

Przykładowo:

```julia
Pkg.add("Example")
Pkg.rm("Example")
Pkg.status()
Pkg.update()
```

To przydatne w skryptach automatyzujących konfigurację środowiska.


### Ważne

Operacje `Pkg.add`, `Pkg.rm`, `Pkg.update` i podobne **modyfikują stan aktywnego środowiska**.

Dlatego w notebookach dydaktycznych nie wykonujemy ich przypadkowo w środowisku kursu.


## **6. Instalowanie pakietów**

### Pkg mode

```text
pkg> add Plots
```


### API


````julia
using Pkg

# Przykład:
# Pkg.add("Plots")
````

### Instalowanie kilku pakietów

```text
pkg> add CSV DataFrames Plots
```

lub:


````julia
# Pkg.add(["CSV", "DataFrames", "Plots"])
````

### Co robi `add`?

W uproszczeniu:

1. wybiera zgodną wersję pakietu;
2. rozwiązuje zależności;
3. pobiera potrzebne pakiety;
4. aktualizuje `Project.toml`;
5. aktualizuje lub tworzy `Manifest.toml`;
6. może uruchomić prekompilację.


## **7. `using` i `import`**

### Teoria

Po dodaniu pakietu trzeba jeszcze załadować odpowiedni moduł.


### `using`


````julia
using Statistics

values = [1, 2, 3, 4, 5]
@show mean(values)
````

`using ModuleName` udostępnia nazwy eksportowane przez moduł.


### `import`


````julia
import Statistics

@show Statistics.mean([10, 20, 30])
````

### Różnica praktyczna

`import ModuleName` zachęca do jawnego używania kwalifikowanych nazw:

```julia
Statistics.mean(...)
```

`using ModuleName` może udostępnić eksportowane symbole bez prefiksu modułu.


### Import pojedynczej nazwy


````julia
import Statistics: median

@show median([1, 10, 100])
````

### Ważne

W większych projektach jawne nazwy modułów mogą poprawić czytelność i ograniczyć konflikty nazw.


## **11. Środowiska projektowe**

### Teoria

Środowisko mówi Julii:

- jakie pakiety należą do projektu;
- jakie wersje są dozwolone;
- jakie wersje zostały faktycznie rozwiązane.

Najczęściej środowisko projektu znajduje się w katalogu projektu.


### Dlaczego izolacja jest ważna?

Bez izolacji projekt A może przypadkowo zacząć zależeć od pakietu zainstalowanego dla projektu B.

Środowiska eliminują ten problem i umożliwiają reprodukowalną pracę.


## **12. `Project.toml`**

### Teoria

`Project.toml` opisuje projekt na poziomie deklaratywnym.

Może zawierać m.in.:

- nazwę projektu;
- UUID;
- wersję;
- bezpośrednie zależności;
- ograniczenia kompatybilności.


### Przykład


```toml
name = "MyJuliaProject"
uuid = "12345678-1234-1234-1234-123456789abc"
version = "0.1.0"

[deps]
CSV = "336ed68f-0bac-5ca0-87d4-7b16caf5d00b"
DataFrames = "a93c6f00-e57d-5684-b7b6-d8193f3e46c0"

[compat]
julia = "1"
CSV = "0.10"
DataFrames = "1"
```


### Ważne

`Project.toml` opisuje **intencję projektu** — jego bezpośrednie zależności i ograniczenia.


## **13. `Manifest.toml`**

### Teoria

`Manifest.toml` opisuje **konkretny rozwiązany graf zależności**.

Może zawierać informacje o:

- dokładnych wersjach;
- UUID;
- zależnościach pośrednich;
- źródłach pakietów;
- identyfikatorach wersji repozytoriów.


### Różnica

| Plik | Rola |
|---|---|
| `Project.toml` | czego projekt potrzebuje |
| `Manifest.toml` | jakie konkretne wersje zostały rozwiązane |


### Ważne

To właśnie połączenie `Project.toml` + `Manifest.toml` daje bardzo mocny fundament reprodukowalności.


## **14. `activate` — wybór środowiska**

### Bieżący katalog

```text
pkg> activate .
```


### API


````julia
using Pkg

# Pkg.activate(".")
````

### Sprawdzenie aktywnego projektu


````julia
using Pkg

project = Pkg.project()
println("Ścieżka projektu: ", project.path)
````

### Analiza

Aktywne środowisko decyduje, które `Project.toml` i `Manifest.toml` są modyfikowane przez Pkg.


## **15. Środowisko tymczasowe**

### Cel

Do eksperymentów idealne jest środowisko tymczasowe:


````julia
using Pkg

course_env = dirname(Pkg.project().path)
println("Środowisko kursu: ", course_env)

Pkg.activate(; temp=true)

temp_env = dirname(Pkg.project().path)
println("Środowisko tymczasowe: ", temp_env)
````

### Dlaczego to robimy?

Operacje wykonane w tym środowisku:

- nie modyfikują zależności kursu;
- pozwalają bezpiecznie testować pakiety;
- są wygodne w notebookach dydaktycznych.


### Ważne

Po zakończeniu ćwiczeń wrócimy do środowiska kursu.

## **17. `resolve` — ponowne rozwiązanie zależności**

### Teoria

`resolve` aktualizuje zależności w `Manifest.toml` zgodnie z aktualnym stanem `Project.toml`.


```text
pkg> resolve
```


````julia
# Pkg.resolve()
````

### Kiedy używać?

Np. gdy:

- zmieniły się bezpośrednie zależności;
- edytowano `Project.toml`;
- rozwijane lokalnie pakiety zmieniły swoje zależności.


## **18. `precompile`**

### Teoria

Pakiety Julia mogą być prekompilowane, aby przyspieszyć późniejsze ładowanie.


```text

```


````julia
# Pkg.precompile()
````

### Ważne

Prekompilacja nie oznacza, że cały kod aplikacji jest „na zawsze skompilowany”.

Julia nadal może generować specjalizacje metod podczas pracy programu.


## **19. `pin` i `free`**

### `pin`

Blokuje pakiet na konkretnej wersji:


```text
pkg> pin Example
```


### `free`

Usuwa pin i przywraca normalne zarządzanie:


```text
pkg> free Example
```


### Kiedy przydatne?

Pin może być przydatny podczas:

- reprodukcji błędu;
- chwilowego zamrożenia zależności;
- testowania kompatybilności.

Nie zastępuje jednak dobrze zaprojektowanych ograniczeń `[compat]`.

## **24. Ograniczenia kompatybilności `[compat]`**

### Teoria

Sekcja `[compat]` określa, z jakimi wersjami zależności projekt deklaruje kompatybilność.


```toml
[compat]
julia = "1"
DataFrames = "1"
CSV = "0.10"
```


### Dlaczego to ważne?

Bez sensownych ograniczeń kompatybilności aktualizacja jednej zależności może w przyszłości doprowadzić do niespodziewanej niezgodności.


### Dobra praktyka

W bibliotekach `[compat]` jest szczególnie ważne, ponieważ informuje resolver Pkg, z jakimi wersjami pakiet rzeczywiście współpracuje.


## **25. Globalne środowisko a środowisko projektu**

### Teoria

Julia posiada domyślne środowisko użytkownika, np. w stylu:

```text
@v1.x
```

Jest wygodne do szybkich eksperymentów.

Dla poważnego projektu preferujemy jednak:


```text
pkg> activate .
```


### Dlaczego?

Środowisko projektu:

- izoluje zależności;
- łatwiej wersjonować;
- łatwiej odtworzyć;
- ogranicza „działa tylko na moim komputerze”.


## **26. `LOAD_PATH`**

### Teoria

Julia posiada stos środowisk określający, skąd mogą być ładowane pakiety i moduły.


````julia
@show LOAD_PATH
````

### Ważne

Dla początkującego najważniejsze jest:

- aktywuj właściwy projekt;
- nie polegaj przypadkowo na pakietach obecnych tylko w globalnym środowisku.


## **27. Reprodukowalność środowiska**

### Teoria

Reprodukowalny projekt to taki, który inna osoba może odtworzyć na podstawie repozytorium i instrukcji.


### Minimalny workflow

```sh
git clone ...
cd project
julia --project=.
```

a następnie w Julia:


```julia
using Pkg
Pkg.instantiate()
```


### Dlaczego `--project=.`?

Uruchamia Julię od razu z aktywnym środowiskiem bieżącego katalogu.


## **28. Git i pliki środowiska**

### Dobra praktyka dla aplikacji i kursów

Zwykle wersjonujemy:

- `Project.toml`;
- `Manifest.toml`.

Dzięki temu można odtworzyć dokładny stan zależności.


### Biblioteki

W przypadku pakietów/bibliotek polityka dotycząca `Manifest.toml` bywa inna niż w aplikacjach.

Najważniejszy jest poprawny `Project.toml` i `[compat]`, ponieważ użytkownik biblioteki ma własne środowisko.


### Ważne

Nie edytuj `Manifest.toml` ręcznie, jeśli nie masz konkretnego powodu i nie rozumiesz konsekwencji.

Zwykle powinien być zarządzany przez Pkg.


## **29. Notebooki i bezpieczeństwo zależności**

### Problem

Wywołanie:

```julia
Pkg.add("SomePackage")
```

wewnątrz notebooka może zmienić aktywne środowisko kursu lub projektu.


### Bezpieczny wzorzec dydaktyczny

1. zapamiętaj środowisko kursu;
2. przełącz się do środowiska tymczasowego;
3. wykonaj eksperyment;
4. wróć do środowiska kursu.


````julia
using Pkg

course_env_demo = dirname(Pkg.project().path)
println("Aktualne środowisko: ", course_env_demo)

Pkg.activate(; temp=true)
println("Tymczasowe środowisko: ", dirname(Pkg.project().path))
````

W tym miejscu można bezpiecznie eksperymentować z pakietami.


## **30. Typowe błędy początkujących**

| Błąd | Dlaczego jest problemem | Poprawne podejście |
|---|---|---|
| instalowanie wszystkiego globalnie | projekty zaczynają dzielić przypadkowe zależności | osobne środowisko projektu |
| `Pkg.add` w notebooku kursu | modyfikuje pliki środowiska | środowisko tymczasowe |
| usunięcie `Manifest.toml` bez powodu | utrata dokładnego rozwiązania wersji | zachowaj go dla aplikacji/kursu |
| mylenie `using` i instalacji | `using` nie instaluje pakietu | najpierw `add`, potem `using` |
| oczekiwanie, że `update` łamie `[compat]` | resolver respektuje ograniczenia | sprawdź `status` i `[compat]` |
| używanie `develop` jak zwykłego `add` | śledzisz lokalne źródła | `develop` tylko do developmentu |
| ręczna edycja `Manifest.toml` | łatwo uszkodzić rozwiązanie zależności | użyj Pkg |
| brak `instantiate` po klonowaniu | zależności mogą nie istnieć lokalnie | `Pkg.instantiate()` |

## **35. Podsumowanie lekcji**

Najważniejsze zasady Lesson 6:

1. `Pkg` jest wbudowanym menedżerem pakietów Julii.
2. REPL ma osobny Pkg mode uruchamiany klawiszem `]`.
3. Pakiety dodajemy do aktywnego środowiska, a nie „do projektu abstrakcyjnie”.
4. `using` i `import` ładują moduły, ale nie instalują pakietu.
5. `Project.toml` opisuje bezpośrednie zależności i kompatybilność.
6. `Manifest.toml` opisuje konkretne rozwiązanie całego grafu zależności.
7. `activate` wybiera środowisko.
8. `Pkg.activate(; temp=true)` jest idealne do eksperymentów.
9. `instantiate` odtwarza środowisko po klonowaniu projektu.
10. `resolve` ponownie rozwiązuje zależności.
11. `precompile` przygotowuje pakiety do szybszego ładowania.
12. `pin` blokuje wersję, a `free` usuwa blokadę.
13. `develop` służy do aktywnej pracy nad pakietem.
14. `[compat]` jest podstawą kontrolowania zgodności wersji.
15. Środowisko projektu jest lepsze niż przypadkowe poleganie na środowisku globalnym.
16. `Project.toml` i `Manifest.toml` są kluczowe dla reprodukowalności aplikacji i kursów.
17. Notebook dydaktyczny nie powinien bez ostrzeżenia modyfikować środowiska projektu.
