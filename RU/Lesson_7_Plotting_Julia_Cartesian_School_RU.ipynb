## **1. Почему визуализация данных важна?**

### Теория

График позволяет увидеть зависимости, которые трудно заметить в необработанной таблице чисел.

Визуализация помогает, в частности:

- выявлять тренды;
- замечать выбросы;
- сравнивать серии данных;
- исследовать распределения;
- анализировать зависимости между переменными;
- наглядно представлять результаты другим людям.


### Важно

График — это не декоративный элемент.

Хорошая визуализация должна прежде всего:

1. отвечать на конкретный вопрос;
2. быть читаемой;
3. не вводить в заблуждение;
4. содержать корректные подписи осей и единицы измерения.

## **24. Лучшие практики визуализации**

### Принципы

1. **Выбирайте тип графика в соответствии с вопросом.**
2. **Подписывайте оси и единицы измерения.**
3. **Не искажайте диапазоны осей без обоснования.**
4. **Не размещайте слишком много серий на одном графике.**
5. **Не полагайтесь только на цвет для передачи информации.**
6. **Используйте понятную легенду.**
7. **Поддерживайте достаточный контраст.**
8. **Удаляйте элементы, которые не помогают интерпретации.**
9. **График должен быть понятен без длительного угадывания смысла.**


### Пример плохой практики

График, содержащий:

- 20 серий;
- каждую серию в отдельном стиле;
- оси без подписей;
- случайно выбранный масштаб;
- отсутствие легенды,

может быть технически корректным, но аналитически бесполезным.


## **25. Типичные ошибки начинающих**

| Ошибка | Причина | Правильный подход |
|---|---|---|
| отсутствует `using Plots` | пакет не загружен | загрузить модуль |
| автоматический `Pkg.add` в notebook-файле | изменяется окружение | устанавливать пакет осознанно |
| `sin(x)` для вектора | отсутствует broadcasting | использовать `sin.(x)` |
| слишком много серий | график становится нечитаемым | разделить на subplots |
| отсутствуют единицы измерения | интерпретация затруднена | добавить единицы к подписям осей |
| необоснованный 3D | ухудшается читаемость | рассмотреть heatmap/contour |
| чрезмерное количество стилей | визуальный шум | придерживаться минимализма |
| экспорт без проверки backend-а | формат может не поддерживаться | тестировать экспорт |

## **29. Итоги урока**

Основные правила Lesson 7:

1. Визуализация данных служит анализу и коммуникации, а не декорации.
2. `Plots.jl` предоставляет удобный интерфейс для создания различных типов графиков.
3. `plot` создаёт график, а `plot!` изменяет существующий.
4. Линейный график подходит для трендов и непрерывных зависимостей.
5. Scatter plot полезен для независимых наблюдений.
6. Несколько серий можно объединять на одном графике или разделять через subplots.
7. Заголовки, подписи осей и единицы измерения критически важны для интерпретации.
8. Broadcasting часто требуется при вычислении значений функций для массивов.
9. Гистограмма показывает распределение, heatmap — матрицу значений, contour — линии уровня.
10. 3D-графики следует использовать только тогда, когда они действительно улучшают понимание данных.
11. `savefig` позволяет экспортировать графики.
12. Backend отвечает за рендеринг и может влиять на возможности экспорта.
13. Автоматизация построения графиков в цикле естественна при работе с множеством серий.
14. Хорошая визуализация проста, читаема и отвечает на конкретный вопрос.
