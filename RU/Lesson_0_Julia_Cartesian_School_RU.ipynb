## **1. Что такое Julia?**

### Теория

**Julia** — современный язык программирования, разработанный прежде всего для:

- научных вычислений,
- анализа данных,
- математического моделирования,
- симуляций,
- оптимизации,
- машинного обучения,
- программирования общего назначения.

Язык сочетает удобство интерактивной работы с высокой производительностью. Одна из его важнейших особенностей — **multiple dispatch**, то есть выбор реализации функции на основе типов её аргументов.

Julia также предлагает:

- динамическую типизацию,
- JIT-компиляцию на базе LLVM,
- вывод типов,
- развитую поддержку математической нотации и Unicode,
- хорошую совместимость с другими языками и библиотеками.


### Важно

Julia — не «только язык для математики». Это полноценный язык программирования, который также можно использовать для:

- создания инструментов,
- автоматизации,
- обработки данных,
- прототипирования,
- инженерных и исследовательских решений.


### Пример


````julia
println("Hello, Julia!")
````

### Анализ

Это самая простая программа в этом уроке.

- `println(...)` выводит текст;
- строка записывается в двойных кавычках `"..."`;
- после выполнения инструкции курсор переходит на новую строку.


## **2. Краткая история языка Julia и его создатели**

### Теория

Работа над Julia началась в конце первого десятилетия XXI века. Язык создавался с целью объединить:

- простоту и удобство языков высокого уровня,
- высокую производительность, характерную для компилируемых языков,
- естественную запись математического кода.

Первая публичная презентация языка состоялась в **2012 году**, когда был опубликован ставший известным текст **“Why We Created Julia”**.


### Создатели

Создателями языка Julia считаются четыре человека:

| Создатель | Роль |
|---|---|
| **Jeff Bezanson** | соавтор языка, разработчик и исследователь |
| **Stefan Karpinski** | соавтор, популяризатор и один из создателей экосистемы |
| **Viral B. Shah** | соавтор и один из организаторов развития проекта |
| **Alan Edelman** | соавтор, математик и исследователь, связанный с MIT |

Julia развивалась в тесной связи с научным и инженерным сообществом и со временем сформировала зрелую экосистему с собственным менеджером пакетов, документацией и большим сообществом пользователей.


### Почему появилась Julia?

Создатели стремились уменьшить проблему «двух языков», то есть ситуацию, когда:

1. прототип создаётся на языке высокого уровня;
2. критичные по производительности части приходится затем переписывать на C или C++.

Julia должна была позволить писать читаемый код высокого уровня, который при этом способен работать очень быстро.


## **3. REPL и notebook**

### Теория

Чаще всего с Julia работают в двух режимах:

| Режим | Значение |
|---|---|
| **REPL** | интерактивный терминал Julia |
| **Notebook** | документ, объединяющий текст, код и результаты |

**REPL** — сокращение от:

**Read → Eval → Print → Loop**

то есть:

1. прочитать команду;
2. выполнить её;
3. показать результат;
4. ожидать следующую команду.


### Важно

Notebook сохраняет состояние сессии.

Это означает, что если вы выполнили ячейку, создающую переменную `x`, то в следующей ячейке можете использовать эту переменную. Однако если выполнять ячейки в другом порядке, результаты могут отличаться или возникнет ошибка.

Поэтому на практике полезно регулярно выполнять notebook **с начала до конца**.

````julia
VERSION
````

Выражение `VERSION` показывает версию текущего окружения Julia.


## **4. Вывод данных**

### Теория

В этом разделе вы познакомитесь с четырьмя важными инструментами:

| Функция | Назначение |
|---|---|
| `print` | вывод без перехода на новую строку |
| `println` | вывод с переходом на новую строку |
| `show` | техническое представление значения |
| `display` | отображение значения средствами notebook-среды |


### Пример — `print` и `println`


````julia
print("Julia ")
print("— ")
print("это читаемый язык.")
println()

println("Первая строка")
println("Вторая строка")
````

### Анализ

- `print` не добавляет перевод строки;
- `println` автоматически завершает строку;
- поэтому несколько вызовов `print` могут формировать одно сообщение.


### Пример — `show` и `display`


````julia
text = "Julia\nLanguage"

println("println:")
println(text)

println("\nshow:")
show(text)

println("\n\ndisplay:")
display(text)
````

### Анализ

Обратите внимание на различия:

- `println(text)` показывает текст в форме, предназначенной для обычного вывода;
- `show(text)` показывает техническое представление строки;
- `display(text)` передаёт представление объекта системе отображения notebook.


### Пример — интерполяция строк


````julia
name = "Julia"
major = VERSION.major
minor = VERSION.minor

println("Я изучаю язык $name.")
println("Текущая основная версия: $major")
println("Номер версии: $(major).$(minor)")
println("2 + 3 = $(2 + 3)")
````

### Типичная ошибка

Начинающие иногда пытаются писать:

```julia
println("2 + 3 = $2 + 3")
```

Так делать не следует. Если нужно вставить в строку целое выражение, используйте:

```julia
println("2 + 3 = $(2 + 3)")
```


## **5. Переменные и присваивание**

### Теория

Переменная в Julia — это **имя, связанное со значением**.

Предварительно объявлять её тип не требуется. Достаточно присвоить значение:

```julia
x = 10
```


### Пример


````julia
answer = 42
language = "Julia"
is_fast = true

println(answer)
println(language)
println(is_fast)
````

### Пример — проверка типа


````julia
@show typeof(answer)
@show typeof(language)
@show typeof(is_fast)
````

### Важно

Тип относится к **значению**, а не к самому имени переменной. Поэтому одно и то же имя может позднее быть связано со значением другого типа.


````julia
x = 10
@show x typeof(x)

x = 10.5
@show x typeof(x)

x = "десять"
@show x typeof(x)
````

### Анализ

Пример выше показывает, что:

- `x` сначала может обозначать целое число;
- затем число с плавающей точкой;
- а позже строку.

Это типично для динамического языка.

## **8. Комментарии**

### Теория

Julia поддерживает два основных вида комментариев:

| Вид | Синтаксис |
|---|---|
| Однострочный | `# комментарий` |
| Многострочный | `#= ... =#` |


### Пример — однострочный комментарий


````julia
# Этот комментарий не выполняется.
result = 2 + 2  # комментарий также может находиться в конце строки

println(result)
````

### Пример — многострочный комментарий


````julia
#=
Внешний комментарий.

#=
Внутренний комментарий.
=#

Конец внешнего комментария.
=#

println("Код выполняется.")
````

### Важно

Хороший комментарий в первую очередь объясняет, **почему** мы что-то делаем, а не буквально повторяет содержимое кода.

### Теория — логические операции

Основные логические операторы:

| Оператор | Значение |
|---|---|
| `&&` | логическое AND |
| `||` | логическое OR |
| `!` | отрицание |


### Пример


````julia
a = true
b = false

@show a && b
@show a || b
@show !a
````

### Анализ

Операторы `&&` и `||` используют **короткое замыкание**:

- в выражении `A && B`, если `A` равно `false`, Julia не вычисляет `B`;
- в выражении `A || B`, если `A` равно `true`, Julia не вычисляет `B`.


````julia
function positive_message(x)
    x > 0 && println("$x — положительное число")
    return nothing
end

positive_message(5)
positive_message(-2)
````

### Типичная ошибка

Не путайте:

- `=` — присваивание;
- `==` — сравнение.

Это совершенно разные операции.


## **11. `convert`, `parse` и `tryparse`**

### Теория

Начинающие очень часто путают преобразование значения с разбором текста.

| Функция | Назначение |
|---|---|
| `convert(T, x)` | преобразует существующее значение `x` к типу `T`, если такое преобразование определено |
| `parse(T, text)` | интерпретирует текст как значение типа `T` |
| `tryparse(T, text)` | пытается разобрать текст; при ошибке возвращает `nothing` |


### Пример — `convert`


````julia
converted = convert(Float64, 365)
@show converted typeof(converted)
````

### Пример — `parse`


````julia
days_text = "365"
days_number = parse(Int, days_text)

@show days_text typeof(days_text)
@show days_number typeof(days_number)
````

### Пример — `tryparse`


````julia
@show tryparse(Int, "365")
@show tryparse(Int, "365 дней")
````

### Типичная ошибка

Так не следует преобразовывать текст в число:

```julia
convert(Int, "365")
```

Строка `"365"` — это текст, а не готовое числовое значение другого типа. Поэтому следует использовать:

```julia
parse(Int, "365")
```

6. `"J"` и `'J'` — не одно и то же.
7. Unicode — важное преимущество Julia, но использовать его следует разумно.
8. `=` означает присваивание, а `==` — сравнение.
9. `convert` преобразует значения, а `parse` читает данные из текста.
10. `tryparse` безопаснее использовать, когда входные данные могут быть некорректными.
