## **1. Что такое условная конструкция?**

### Теория

Условная конструкция позволяет программе выбирать разные пути выполнения в зависимости от результата логического выражения.

Например:

- если пользователь совершеннолетний → разрешить доступ;
- если пароль неверный → вывести сообщение об ошибке;
- если температура ниже нуля → сообщить о морозе;
- если значение отсутствует → выполнить альтернативную ветвь.


### Базовая идея

```text
если условие истинно
    выполнить A
иначе
    выполнить B
```


## **2. Тип `Bool` и логические условия**

### Теория

Условие в Julia должно приводить к логическому значению:

- `true`;
- `false`.

Тип этих значений — `Bool`.


````julia
@show true typeof(true)
@show false typeof(false)
````

### Пример


````julia
x = 10

condition = x > 5

@show condition
@show typeof(condition)
````

### Важно

Julia не преобразует числа автоматически в логические значения.

Поэтому:

```julia
if 1
    ...
end
```

не является корректным условием.

Условие должно явно возвращать `Bool`.


## **3. Инструкция `if`**

### Синтаксис

```julia
if условие
    инструкции
end
```


### Пример


````julia
temperature = 28

if temperature > 25
    println("Тепло.")
end
````

### Анализ

Если `temperature > 25` возвращает `true`, код внутри блока выполняется.

Если результат равен `false`, блок пропускается.


## **4. `if ... else`**

### Теория

Если нужны ровно два возможных пути выполнения, используется конструкция `if ... else`.


````julia
age = 20

if age >= 18
    println("Совершеннолетний пользователь.")
else
    println("Несовершеннолетний пользователь.")
end
````

### Важно

`else` не имеет собственного условия. Эта ветвь выполняется, если предыдущий `if` оказался ложным.


## **5. `if ... elseif ... else`**

### Теория

Для нескольких возможных случаев используется `elseif`.


````julia
score = 87

if score >= 90
    grade = "A"
elseif score >= 80
    grade = "B"
elseif score >= 70
    grade = "C"
else
    grade = "D"
end

@show grade
````

### Анализ

Julia проверяет условия **сверху вниз**.

После нахождения первого условия со значением `true` выполняется соответствующий блок, а остальные ветви пропускаются.


## **6. Порядок условий имеет значение**

### Типичная ошибка


````julia
score = 95

if score >= 60
    result = "зачтено"
elseif score >= 90
    result = "очень хороший результат"
else
    result = "не зачтено"
end

@show result
````

### Анализ

При `score = 95` первое условие `score >= 60` уже истинно, поэтому `elseif score >= 90` никогда не будет проверено.

Правильная последовательность обычно идёт от более специфичного условия к более общему.


````julia
score = 95

if score >= 90
    result = "очень хороший результат"
elseif score >= 60
    result = "зачтено"
else
    result = "не зачтено"
end

@show result
````

## **10. Тернарный оператор `? :`**

### Теория

Тернарный оператор имеет вид:

```julia
условие ? значение_если_true : значение_если_false
```

Он является компактным эквивалентом простого `if ... else`.


````julia
x = 10
y = 30

maximum_value = x > y ? x : y

@show maximum_value
````

### Эквивалент через `if`


````julia
maximum_value_if = if x > y
    x
else
    y
end

@show maximum_value_if
@assert maximum_value == maximum_value_if
````

### Важно

Тернарный оператор хорошо подходит для **коротких и простых** условий.

Если логика становится сложной, обычный `if` обычно читается лучше.


## **11. Short-circuit evaluation**

### Теория

Операторы `&&` и `||` используют механизм короткого замыкания.

Для:

```julia
A && B
```

если `A` равно `false`, Julia не вычисляет `B`.

Для:

```julia
A || B
```

если `A` равно `true`, Julia не вычисляет `B`.


### Пример


````julia
a = false

result = a && (println("Этот текст не будет выведен."); true)

@show result
````

### Пример с функцией и побочным эффектом


````julia
function side_effect()
    println("Функция side_effect была выполнена.")
    return true
end

a = false
result = a && side_effect()

@show result
````

Поскольку `a == false`, вызов `side_effect()` пропускается.


## **12. Short-circuit как защитный механизм**

### Пример — защита от ошибки


````julia
x = 0

(x != 0) && println(10 / x)

println("Программа продолжает работу.")
````

### Анализ

Если `x == 0`, выражение:

```julia
x != 0
```

имеет значение `false`, поэтому правая часть `&&` не выполняется.

Таким образом можно избежать деления на ноль.


### Пример с `||`


````julia
username = ""

isempty(username) || println("Добро пожаловать, $username!")
````

Если строка пустая, правая часть выражения не будет выполнена.


## **13. FizzBuzz — классический пример условий**

### Цель

Для числа `N`:

- если оно делится на 3 и на 5 → `"FizzBuzz"`;
- если только на 3 → `"Fizz"`;
- если только на 5 → `"Buzz"`;
- иначе → вывести само число.


````julia
N = 15

if (N % 3 == 0) && (N % 5 == 0)
    println("FizzBuzz")
elseif N % 3 == 0
    println("Fizz")
elseif N % 5 == 0
    println("Buzz")
else
    println(N)
end
````

### Важно

Сначала проверяем наиболее специфичный случай:

```julia
N % 3 == 0 && N % 5 == 0
```

Если сначала проверить только делимость на 3, число `15` будет классифицировано как `"Fizz"` и до условия `"FizzBuzz"` выполнение уже не дойдёт.


### FizzBuzz как функция


````julia
function fizzbuzz(n)
    if n % 15 == 0
        return "FizzBuzz"
    elseif n % 3 == 0
        return "Fizz"
    elseif n % 5 == 0
        return "Buzz"
    else
        return string(n)
    end
end

@show fizzbuzz(3)
@show fizzbuzz(5)
@show fizzbuzz(15)
@show fizzbuzz(7)
````

## **14. Условия с `nothing`**

### Теория

`nothing` часто используется как сигнал «значение отсутствует».

Типичный пример — `tryparse`.


````julia
text = "123"
value = tryparse(Int, text)

if value === nothing
    println("Не удалось прочитать число.")
else
    println("Прочитано число: $value")
end
````

### Важно

Для сравнения с singleton-значением `nothing` идиоматично использовать:

```julia
value === nothing
```

или:

```julia
value !== nothing
```

## **16. Условия в функциях**

### Пример


````julia
function classify_number(x)
    if x > 0
        return "положительное"
    elseif x < 0
        return "отрицательное"
    else
        return "ноль"
    end
end

@show classify_number(10)
@show classify_number(-5)
@show classify_number(0)
````

### Хорошая практика — ранний `return`

Иногда функцию легче читать, если завершить её раньше.


````julia
function safe_divide(a, b)
    if b == 0
        return nothing
    end

    return a / b
end

@show safe_divide(10, 2)
@show safe_divide(10, 0)
````

## **17. Условия и multiple dispatch**

### Теория

Не каждый выбор поведения следует реализовывать через `if`.

Julia имеет мощный механизм **multiple dispatch** — выбор метода функции может зависеть от типов аргументов.


### Подход через условные проверки


````julia
function describe_with_if(x)
    if x isa Int
        return "целое число"
    elseif x isa AbstractFloat
        return "число с плавающей точкой"
    elseif x isa AbstractString
        return "строка"
    else
        return "другой тип"
    end
end

@show describe_with_if(10)
@show describe_with_if(3.14)
@show describe_with_if("Julia")
````

### Подход через multiple dispatch


````julia
describe(x::Int) = "целое число"
describe(x::AbstractFloat) = "число с плавающей точкой"
describe(x::AbstractString) = "строка"
describe(x) = "другой тип"

@show describe(10)
@show describe(3.14)
@show describe("Julia")
@show describe(true)
````

### Важно

Используйте:

- `if` — когда решение зависит от **значения**;
- multiple dispatch — часто тогда, когда поведение зависит от **типа**.

Это не абсолютное правило, но важный стиль проектирования в Julia.


## **18. Типичные ошибки начинающих**

| Ошибка | Причина | Правильный подход |
|---|---|---|
| `if 1` | Julia ожидает `Bool` | использовать явное сравнение |
| `x = 5` вместо `x == 5` | присваивание вместо сравнения | использовать `==` |
| плохой порядок `elseif` | общее условие перехватывает более специфичное | проверять специфичные случаи раньше |
| чрезмерный тернарный оператор | ухудшается читаемость | использовать обычный `if` |
| `&` вместо `&&` | другая семантика | использовать `&&` для логических условий |
| `|` вместо `||` | другая семантика | использовать `||` для логических условий |
| длинная цепочка `if x isa ...` | поведение зависит от типа | рассмотреть multiple dispatch |


### Важно — `&&` и `&`

`&&`:

- используется для логических условий;
- поддерживает short-circuit.

`&`:

- является побитовым или поэлементным оператором в зависимости от типов;
- не обеспечивает такое же short-circuit-поведение.

Аналогично различаются `||` и `|`.

## **20. Мини-проект — валидация доступа пользователя**

### Данные

Есть пользователь с параметрами:

- возраст;
- активность аккаунта;
- роль;
- наличие токена.


````julia
user = (
    name = "Анна",
    age = 29,
    active = true,
    role = "editor",
    has_token = true,
)
````

### Правила доступа

Пользователь получает доступ, если:

1. он совершеннолетний;
2. аккаунт активен;
3. роль — `admin` или `editor`;
4. имеется токен.


````julia
allowed_role = user.role == "admin" || user.role == "editor"

can_access =
    user.age >= 18 &&
    user.active &&
    allowed_role &&
    user.has_token

if can_access
    println("Доступ предоставлен пользователю $(user.name).")
else
    println("Доступ запрещён.")
end
````

### Расширенная диагностическая версия

Вместо простого `true/false` можно сообщить конкретную причину отказа.


````julia
if user.age < 18
    reason = "пользователь несовершеннолетний"
elseif !user.active
    reason = "аккаунт неактивен"
elseif !(user.role in ("admin", "editor"))
    reason = "роль не имеет доступа"
elseif !user.has_token
    reason = "отсутствует токен"
else
    reason = nothing
end

if reason === nothing
    println("Доступ предоставлен.")
else
    println("Доступ запрещён: $reason")
end
````

### Анализ

В проекте использованы:

- `if`;
- `elseif`;
- `else`;
- `&&`;
- `||`;
- `!`;
- `in`;
- `nothing`.

Это типичный пример, где читаемая условная логика важнее попытки чрезмерно сократить код.

8. `&&` и `||` используют short-circuit evaluation.
9. Short-circuit можно применять как защитный механизм.
10. `nothing` часто используется для обозначения отсутствия значения.
11. Для коллекций полезны `isempty`, `in` и `haskey`.
12. Ранний `return` может упростить функции.
13. Если выбор поведения зависит от типа, стоит рассмотреть multiple dispatch.
