## **1. Зачем нужны структуры данных?**

### Теория

Одна переменная хранит одно значение. В реальной программе нам часто требуется хранить:

- список температур,
- данные пользователя,
- телефонную книгу,
- матрицу измерений,
- уникальный набор идентификаторов,
- запись с именем, возрастом и городом.

**Структура данных** организует множество значений так, чтобы упростить:

- хранение,
- поиск,
- изменение,
- итерацию,
- обработку.


### Карта основных структур данных

| Структура | Изменяемая? | Упорядоченная? | Типичное применение |
|---|---:|---:|---|
| `Tuple` | нет | да | небольшой фиксированный набор значений |
| `NamedTuple` | нет | да | лёгкая запись с именованными полями |
| `Dict` | да | логически ключ → значение | быстрый доступ по ключу |
| `Set` | да | порядок не является частью контракта | уникальные значения |
| `Vector` | да | да | одномерный массив |
| `Matrix` | да | да | двумерные данные |
| `Array{T,N}` | да | да | N-мерный массив |


## **2. Кортежи — `Tuple`**

### Теория

Кортеж — это упорядоченная, **неизменяемая** коллекция значений.

Создаётся с помощью круглых скобок:


````julia
my_favorite_animals = ("пингвины", "кошки", "сахарные поссумы")

@show my_favorite_animals
@show typeof(my_favorite_animals)
````

### Индексирование

Julia индексирует коллекции начиная с **1**.


````julia
@show my_favorite_animals[1]
@show my_favorite_animals[2]
@show my_favorite_animals[end]
````

### Важно

`Tuple` неизменяем. Нельзя выполнить:

```julia
my_favorite_animals[1] = "выдры"
```

Если нужна изменённая версия, создайте новый кортеж.


### Пример — гетерогенный кортеж


````julia
person = ("Anna", 30, true)

@show person
@show typeof(person)
````

### Анализ

Кортеж может хранить элементы разных типов без использования `Any`. Тип кортежа сохраняет информацию о типе каждого элемента.


### Одноэлементный кортеж

Это тонкий, но важный случай:


````julia
not_a_tuple = (42)
one_element_tuple = (42,)

@show typeof(not_a_tuple)
@show typeof(one_element_tuple)
````

### Типичная ошибка

Одни только круглые скобки не создают одноэлементный кортеж.

Нужна запятая:

```julia
(42,)
```


### Деструктуризация кортежа

Элементы кортежа можно сразу присвоить нескольким переменным:


````julia
coordinates = (10, 20)
x, y = coordinates

@show x y
````

## **3. `NamedTuple` — кортеж с именованными полями**

### Теория

`NamedTuple` похож на обычный `Tuple`, но каждому элементу сопоставлено имя.

Это удобная структура для небольших записей данных.


````julia
animal_profile = (
    bird = "пингвин",
    mammal = "кошка",
    marsupial = "сахарный поссум",
)

@show animal_profile
@show typeof(animal_profile)
````

### Доступ по индексу и имени


````julia
@show animal_profile[1]
@show animal_profile.bird
@show animal_profile.mammal
````

### Пример — данные человека


````julia
person = (
    name = "Alice",
    age = 30,
    city = "New York",
)

println(person.name)
println(person.age)
println(person.city)
````

### Создание из существующих переменных


````julia
name = "Bob"
age = 25
city = "Los Angeles"

person2 = (; name, age, city)

@show person2
````

### Важно

`NamedTuple` неизменяем.

Нельзя выполнить:

```julia
person.age = 31
```

Вместо этого можно создать новую запись.


### Обновление через `merge`


````julia
person_updated = merge(person, (age = 31,))

@show person
@show person_updated
````

### Использование в функциях


````julia
function describe_person(p)
    return "$(p.name) — $(p.age) лет, живёт в $(p.city)."
end

println(describe_person(person_updated))
````

## **4. Словари — `Dict`**

### Теория

`Dict` хранит пары:

**ключ → значение**

Пример:

```text
"Anna" → "123-456"
```

Словарь изменяем и хорошо подходит для данных, к которым нужно обращаться по уникальному ключу.


### Создание словаря


````julia
phonebook = Dict(
    "Anna" => "111-222",
    "Bartosz" => "333-444",
    "Siergiej" => "555-666",
)

@show phonebook
@show typeof(phonebook)
````

### Доступ по ключу


````julia
@show phonebook["Anna"]
````

### Типичная ошибка

Обращение к отсутствующему ключу:

```julia
phonebook["Неизвестный"]
```

приведёт к `KeyError`.

Если ключ может отсутствовать, используйте более безопасные способы доступа.


## **5. Безопасный доступ к `Dict`**

### `haskey`


````julia
@show haskey(phonebook, "Anna")
@show haskey(phonebook, "Неизвестный")
````

### `get`

`get(dict, key, default)` возвращает значение по ключу либо значение по умолчанию.


````julia
number = get(phonebook, "Неизвестный", "номер отсутствует")

@show number
````

### `get!`

`get!` также может записать значение по умолчанию в словарь.


````julia
settings = Dict{String, Int}()

timeout = get!(settings, "timeout", 30)

@show timeout
@show settings
````

## **6. Изменение словаря**

### Добавление и обновление


````julia
phonebook["Ewa"] = "777-888"
phonebook["Anna"] = "999-000"

@show phonebook
````

### Удаление через `delete!`


````julia
temporary = Dict("a" => 1, "b" => 2, "c" => 3)

delete!(temporary, "b")

@show temporary
````

### `pop!` — удалить и вернуть


````julia
contacts = Dict(
    "Anna" => "111",
    "Siergiej" => "222",
)

removed_number = pop!(contacts, "Siergiej")

@show removed_number
@show contacts
````

### Ключи, значения и пары


````julia
example = Dict("a" => 10, "b" => 20)

@show collect(keys(example))
@show collect(values(example))
@show collect(pairs(example))
````

### Итерация по словарю


````julia
for (name, number) in phonebook
    println("$name -> $number")
end
````

### Типы ключей и значений

Каждый `Dict` имеет тип ключа и тип значения:

```julia
Dict{K, V}
```

Например:


````julia
strict_phonebook = Dict{String, String}(
    "Anna" => "1234",
    "Boris" => "5678",
)

@show typeof(strict_phonebook)
````

### Важно

В `Dict{String,String}` нельзя просто записать целое число как значение.


````julia
error_detected = try
    strict_phonebook["Emergency"] = 911
    false
catch e
    e isa MethodError
end

@show error_detected
````

### Гетерогенный словарь


````julia
flexible_phonebook = Dict{String, Any}(
    "Евгений" => 8675309,
    "Константин" => "555-2368",
)

flexible_phonebook["Emergency"] = 911

@show flexible_phonebook
````

### Важно

`Dict{String,Any}` обеспечивает гибкость, но лишает компилятор части полезной информации о типах.

Если набор допустимых типов невелик, часто лучше использовать:

```julia
Dict{String, Union{Int, String}}
```


````julia
typed_flexible_phonebook = Dict{String, Union{Int, String}}(
    "Евгений" => 8675309,
    "Константин" => "555-2368",
)

typed_flexible_phonebook["Emergency"] = 911

@show typed_flexible_phonebook
@show typeof(typed_flexible_phonebook)
````

## **7. Множества — `Set`**

### Теория

`Set` хранит **уникальные** значения.

Он полезен, когда нас прежде всего интересуют:

- принадлежность элемента множеству,
- удаление дубликатов,
- операции над множествами.


````julia
numbers = Set([1, 2, 2, 3, 3, 3])

@show numbers
@show length(numbers)
````

### Проверка принадлежности


````julia
@show 2 in numbers
@show 10 in numbers
````

### Добавление и удаление


````julia
push!(numbers, 4)
delete!(numbers, 1)

@show numbers
````

### Операции над множествами


````julia
a = Set([1, 2, 3])
b = Set([3, 4, 5])

@show union(a, b)
@show intersect(a, b)
@show setdiff(a, b)
````

обычно создаёт `Vector{Float64}`.

Если вы действительно хотите принудительно использовать `Any`, можно написать:

```julia
Any[1, "Julia", true]
```

## **19. Comprehensions**

### Теория

Comprehension — компактный способ создавать коллекции по заданному правилу.


````julia
squares = [x^2 for x in 1:10]

@show squares
````

### Comprehension с условием


````julia
even_squares = [x^2 for x in 1:20 if iseven(x)]

@show even_squares
````

### Матрица через comprehension


````julia
multiplication_table = [i * j for i in 1:4, j in 1:4]

@show multiplication_table
````

## **20. `map`, `filter` и `reduce`**

### `map`

Преобразует каждый элемент.


````julia
values = [1, 2, 3, 4]

doubled = map(x -> 2x, values)

@show doubled
````

### `filter`

Оставляет элементы, удовлетворяющие условию.


````julia
even_values = filter(iseven, values)

@show even_values
````

### `reduce`

Сворачивает коллекцию в одно значение.


````julia
total = reduce(+, values)

@show total
````


## **22. `Vector{Any}` — гибкость и цена**

### Теория

`Vector{Any}` может хранить практически любые значения:


````julia
mixed = Any[
    42,
    "Julia",
    true,
    [1, 2, 3],
]

@show mixed
````

### Что мы теряем?

Очень широкий тип элементов означает меньше информации для компилятора.

Это может привести к:

- более частому динамическому dispatch;
- boxing;
- дополнительным аллокациям;
- менее предсказуемой производительности.

В коротком учебном коде это не всегда существенно, но в численном и производительном коде имеет значение.


### Лучшая альтернатива — небольшой `Union`


````julia
id_or_name = Union{Int, String}[1, "Alice", 2, "Bob"]

@show id_or_name
@show eltype(id_or_name)
````

### Ещё более подходящая альтернатива — семантическая структура

Если элементы имеют постоянный смысл, часто лучше использовать:

- `Tuple`,
- `NamedTuple`,
- собственный `struct`.

Не применяйте `Any` только потому, что это самый простой вариант.


## **23. Присваивание не всегда означает копирование**

### Теория

Для изменяемых коллекций:


````julia
fibonacci = [1, 1, 2, 3, 5, 8]
some_more_numbers = fibonacci

some_more_numbers[1] = 999

@show fibonacci
@show some_more_numbers
````

### Анализ

Оба имени указывают на **один и тот же объект**.

Это называется aliasing.


### `copy`


````julia
fibonacci = [1, 1, 2, 3, 5, 8]
independent = copy(fibonacci)

independent[1] = 999

@show fibonacci
@show independent
````

## **24. `copy` и `deepcopy`**

### Теория

Для плоского массива `copy` обычно достаточно.

Для вложенных структур `copy` копирует только внешний контейнер.


````julia
nested = [[1, 2], [3, 4]]
shallow = copy(nested)

shallow[1][1] = 999

@show nested
@show shallow
````

Внутренние векторы по-прежнему остаются общими для обеих структур.


### `deepcopy`


````julia
nested = [[1, 2], [3, 4]]
deep = deepcopy(nested)

deep[1][1] = 999

@show nested
@show deep
````

## **25. Views — `view` и `@view`**

### Теория

Обычный срез часто создаёт копию данных.

View ссылается на фрагмент исходного массива без копирования элементов.


````julia
data = [10, 20, 30, 40, 50]

slice_copy = data[2:4]
slice_view = view(data, 2:4)

@show slice_copy
@show slice_view
````

### Изменение view влияет на исходный массив


````julia
slice_view[1] = 999

@show data
@show slice_view
````

### Макрос `@view`


````julia
data2 = [1, 2, 3, 4, 5]

v = @view data2[2:4]

@show v
````

## **26. Изменение формы массивов**

### `reshape`


````julia
v = collect(1:12)
M = reshape(v, 3, 4)

@show v
@show M
````

### Важно

`reshape` может использовать тот же буфер данных вместо создания полной копии.


### `vec`


````julia
flat = vec(M)

@show flat
````

### `permutedims`


````julia
A = [
    1 2 3
    4 5 6
]

B = permutedims(A)

@show A
@show B
````

## **27. Итерация по коллекциям**

### Вектор


````julia
for value in [10, 20, 30]
    println(value)
end
````

### Индекс и значение — `enumerate`


````julia
languages = ["Julia", "Python", "Rust"]

for (i, language) in enumerate(languages)
    println("$i -> $language")
end
````

### Словарь


````julia
d = Dict("a" => 1, "b" => 2)

for (key, value) in d
    println("$key -> $value")
end
````

### Матрица

Итерация по матрице проходит по элементам в порядке хранения Julia — **по столбцам (column-major)**.


````julia
M = [
    1 2
    3 4
]

for value in M
    print(value, " ")
end
println()
````

### Важно — порядок column-major

Julia, как Fortran и MATLAB, использует порядок хранения **column-major**.

Это важно для производительности численного кода, особенно во вложенных циклах.


## **28. `@assert` — простая проверка предположений**

### Теория

Макрос:

```julia
@assert условие
```

проверяет, истинно ли выражение.

Если оно равно `false`, Julia выдаёт `AssertionError`.


````julia
flexible_phonebook = Dict{String, Any}(
    "Emergency" => 911,
)

@assert flexible_phonebook["Emergency"] == 911

println("PASS — предположение выполнено.")
````

### Диагностическое сообщение


````julia
x = 10

@assert x > 0 "x должен быть положительным"

println("PASS")
````

### Важно

`@assert` полезен для:

- тестов;
- обучения;
- проверки предположений программиста.

Однако его не следует считать полноценной системой валидации пользовательских данных.


## **29. Типичные ошибки начинающих**

| Ошибка | Причина | Правильный подход |
|---|---|---|
| попытка изменить `Tuple` | кортежи неизменяемы | создать новый кортеж |
| `dict["нет"]` | ключ отсутствует | `get`, `haskey` |
| добавление `Int` в `Dict{String,String}` | тип значения ограничен | выбрать подходящий `V`, например `Union` |
| путаница `Vector{Vector}` и `Matrix` | различаются структура памяти и индексирование | проверить `typeof` и `ndims` |
| ожидание, что `b = a` копирует массив | создаётся alias | использовать `copy` или `deepcopy` |
| использование `Vector{Any}` без необходимости | теряется информация о типах | конкретный тип или небольшой `Union` |
| ненужные срезы для больших данных | создаются копии | `view` / `@view` |
| предположение о row-major | Julia использует column-major | проектировать циклы с учётом столбцового порядка |
