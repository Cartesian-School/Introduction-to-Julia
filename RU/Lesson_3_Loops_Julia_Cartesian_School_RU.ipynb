## **1. Что такое цикл?**

### Теория

Цикл используется для многократного выполнения фрагмента кода.

Типичные применения:

- вывод чисел от 1 до 100;
- обработка всех элементов массива;
- повторение операции до выполнения условия;
- построение матрицы;
- агрегация данных;
- обработка каждой записи словаря.


### Два основных вида циклов в этом уроке

| Конструкция | Когда использовать |
|---|---|
| `while` | когда число итераций заранее неизвестно и зависит от условия |
| `for` | когда выполняется обход конкретного набора значений или итератора |


## **2. Цикл `while`**

### Синтаксис

```julia
while условие
    инструкции
end
```

Тело цикла выполняется до тех пор, пока условие имеет значение `true`.


### Пример


````julia
n = 0

while n < 10
    n += 1
    println(n)
end

@show n
````

### Анализ

1. `n = 0` инициализирует счётчик.
2. `while n < 10` проверяет условие перед каждой итерацией.
3. `n += 1` увеличивает счётчик.
4. Когда `n` достигает `10`, следующая проверка условия возвращает `false`.
5. Цикл завершается.


### Пример — `while` и вектор


````julia
myfriends = ["Ted", "Robyn", "Barney", "Lily", "Marshall"]

i = 1

while i <= length(myfriends)
    println("Привет, $(myfriends[i])!")
    i += 1
end
````

### Важно

При ручном индексировании вы сами должны контролировать:

- начальное значение индекса;
- корректное условие завершения;
- изменение индекса на каждой итерации.

Поэтому для обычного обхода коллекции цикл `for` обычно проще и понятнее.


## **3. Типичные проблемы цикла `while`**

### Типичная ошибка — бесконечный цикл

Если забыть изменить переменную, управляющую условием:

```julia
n = 0

while n < 10
    println(n)
end
```

условие никогда не изменится, и цикл будет выполняться бесконечно.


### Хорошая практика

В каждом цикле `while` проверьте три вещи:

1. что задаёт начальное состояние;
2. что изменяет состояние;
3. что гарантирует завершение цикла.


## **4. Цикл `for`**

### Синтаксис

```julia
for element in iterable
    инструкции
end
```

`iterable` может быть, например:

- диапазоном;
- вектором;
- строкой;
- словарём;
- генератором;
- другим объектом, реализующим протокол итерации.


### Пример — диапазон


````julia
for n in 1:10
    println(n)
end
````

### Пример — вектор


````julia
myfriends = ["Ted", "Robyn", "Barney", "Lily", "Marshall"]

for friend in myfriends
    println("Привет, $friend, рад тебя видеть!")
end
````

### Анализ

В этом примере не требуется вручную управлять индексом.

Julia поочерёдно присваивает каждый элемент `myfriends` переменной `friend`.


## **5. Диапазоны и итераторы**

### Диапазон `1:10`


````julia
r = 1:10

@show r
@show typeof(r)
@show collect(r)
````

Диапазон не обязан сразу создавать в памяти полный массив значений.


### Диапазон с шагом


````julia
for n in 2:2:10
    println(n)
end
````

### Убывающий диапазон


````julia
for n in 10:-2:2
    println(n)
end
````

### `range`


````julia
for x in range(0.0, 1.0; length=5)
    println(x)
end
````

## **6. Итерация по векторам и строкам**

### Вектор


````julia
languages = ["Julia", "Rust", "Python"]

for language in languages
    println(language)
end
````

### Строка


````julia
for c in "Julia 🚀"
    println(c)
end
````

### Важно

Для Unicode-текста итерация:

```julia
for c in text
```

безопаснее, чем ручное обращение к `text[i]` при `i = 1:length(text)`.


## **7. `enumerate` — номер и значение**

### Теория

Если одновременно нужны номер итерации и значение:


````julia
languages = ["Julia", "Rust", "Python"]

for (i, language) in enumerate(languages)
    println("$i -> $language")
end
````

### Важно

`enumerate` нумерует итерации начиная с `1`, но его счётчик не следует отождествлять со всеми возможными индексами произвольной структуры.

Для обычных векторов эти понятия часто совпадают, но семантически это разные вещи.


## **8. `zip` — параллельная итерация по нескольким коллекциям**

### Пример


````julia
names = ["Anna", "Piotr", "Ewa"]
scores = [92, 81, 96]

for (name, score) in zip(names, scores)
    println("$name -> $score")
end
````

### Анализ

`zip` объединяет элементы, находящиеся на соответствующих позициях.

Итерация завершается, когда исчерпан самый короткий итератор.


## **9. `break` — немедленное завершение цикла**

### Пример


````julia
for n in 1:100
    if n == 7
        println("Найдено 7 — завершаю цикл.")
        break
    end
    println(n)
end
````

### Когда использовать?

`break` полезен, например, при:

- поиске первого подходящего элемента;
- ожидании определённого состояния;
- ручном завершении цикла после выполнения условия.


## **10. `continue` — пропуск текущей итерации**

### Пример


````julia
for n in 1:10
    if isodd(n)
        continue
    end
    println(n)
end
````

Цикл выведет только чётные числа.


## **11. Вложенные циклы**

### Теория

Один цикл может находиться внутри другого.


````julia
for row in 1:3
    for col in 1:4
        println("row=$row, col=$col")
    end
end
````

### Сокращённый синтаксис


````julia
for row in 1:3, col in 1:4
    println("row=$row, col=$col")
end
````

Обе формы логически эквивалентны.


## **12. Циклы и матрицы**

### Цель

Построим таблицу сложения:


````julia
m, n = 5, 5
A = fill(0, (m, n))

for j in 1:n
    for i in 1:m
        A[i, j] = i + j
    end
end

A
````

### Анализ

Значение каждого элемента равно сумме индекса строки и индекса столбца:

```julia
A[i, j] = i + j
```


### Column-major order

Julia хранит матрицы по столбцам, как Fortran и MATLAB.

Поэтому в классических вложенных циклах обычно выгоднее:

```julia
for j in столбцы
    for i in строки
        A[i, j] = ...
    end
end
```

то есть изменять индекс строки `i` во внутреннем цикле.


### Сокращённая версия той же петли


````julia
B = fill(0, (m, n))

for j in 1:n, i in 1:m
    B[i, j] = i + j
end

@assert A == B
B
````

## **13. `eachindex` — идиоматичная итерация по индексам**

### Теория

Если индексы действительно нужны, предпочтительно использовать:


````julia
values = [10, 20, 30, 40]

for i in eachindex(values)
    println("индекс=$i, значение=$(values[i])")
end
````

### Почему `eachindex`?

`eachindex`:

- возвращает корректные индексы для конкретной структуры;
- не предполагает определённый способ индексирования;
- является хорошим выбором для обобщённого кода.


## **14. Итерация по словарям**

### Пример


````julia
phonebook = Dict(
    "Анна" => "111-222",
    "Пётр" => "333-444",
    "Ева" => "555-666",
)

for (name, number) in phonebook
    println("$name -> $number")
end
````

### Важно

Не стройте логику на произвольном «номере позиции» записи в словаре.

В первую очередь важна пара:

```text
ключ => значение
```


## **15. Comprehensions — создание коллекций**

### Теория

Comprehension позволяет создавать новую коллекцию с помощью компактного синтаксиса.


````julia
squares = [i^2 for i in 1:10]

@show squares
````

### Comprehension с условием


````julia
even_squares = [i^2 for i in 1:20 if iseven(i)]

@show even_squares
````

### Матрица через comprehension


````julia
C = [i + j for i in 1:m, j in 1:n]

@assert C == A
C
````

### Когда использовать comprehension?

Он отлично подходит для создания новой коллекции на основе выражения.

Если код содержит много побочных эффектов, многошаговую логику или сложное управление потоком, обычный `for` часто будет понятнее.


## **16. Dict comprehension**

### Пример


````julia
squares_dict = Dict(i => i^2 for i in 1:10)

@show squares_dict
@assert squares_dict[10] == 100
````

Это естественный способ построения словаря по заданному правилу.


## **17. `map` против `for`**

### `map`

`map` применяет функцию к последовательным элементам и возвращает результаты.


````julia
values = 1:5
doubled = map(x -> 2x, values)

@show doubled
````

### Сравнение

Если требуется **создать набор результатов**:

```julia
map(f, data)
```

или comprehension часто являются естественным выбором.

Если же основная задача — выполнять **побочные эффекты**, например `println`, обычный `for` обычно понятнее.


### Типичная стилистическая ошибка

Такой код работает:

```julia
map(x -> println(x^2), 1:100)
```

но семантически `map` предназначен для преобразования значений в результаты.

Для простого вывода лучше использовать:


````julia
for x in 1:5
    println(x^2)
end
````

## **18. Генераторы и редукции**

### Сумма квадратов

Вместо создания промежуточного массива:


````julia
sum_of_squares = sum(i^2 for i in 1:100)

@show sum_of_squares
@assert sum_of_squares == 338350
````

### Анализ

Выражение:

```julia
(i^2 for i in 1:100)
```

является генератором.

`sum` может потреблять его без необходимости создавать полный промежуточный массив.


## **19. Производительность циклов в Julia**

### Важно

Во многих языках высокого уровня пользователи избегают циклов из опасений по поводу производительности.

В Julia циклы являются полноценным и эффективным инструментом.

Нет необходимости искусственно «векторизовать» каждую операцию только потому, что используется язык высокого уровня.


### Практические рекомендации

Для производительного кода:

1. размещайте горячие циклы внутри функций;
2. следите за стабильностью типов;
3. избегайте ненужных аллокаций;
4. используйте `eachindex`, когда нужны индексы;
5. при работе с матрицами учитывайте column-major order.


````julia
function sum_squares_loop(n)
    total = zero(Int)
    for i in 1:n
        total += i^2
    end
    return total
end

@show sum_squares_loop(100)
````

## **20. Типичные ошибки начинающих**

| Ошибка | Причина | Правильный подход |
|---|---|---|
| бесконечный `while` | состояние не изменяется | обновлять управляющую переменную |
| ручной `i = 1` без необходимости | усложнение кода | `for element in collection` |
| обход строки через `1:length(s)` | проблемы с Unicode | `for c in s` |
| использование `map` только ради `println` | неверная семантика | обычный `for` |
| создание массива только ради суммы | лишняя аллокация | генератор + `sum` |
| неудачный порядок индексов матрицы | хуже локальность памяти | учитывать column-major |
| отсутствие `break` при поиске первого результата | лишние итерации | завершить цикл после нахождения |

## **22. Мини-проект — анализ результатов слушателей**

### Данные


````julia
students = [
    (name = "Анна", score = 92),
    (name = "Пётр", score = 67),
    (name = "Ева", score = 96),
    (name = "Марек", score = 74),
    (name = "Оля", score = 88),
]
````

### Задание A — вывести всех слушателей


````julia
for student in students
    println("$(student.name) -> $(student.score)")
end
````

### Задание B — выбрать результаты не ниже 80


````julia
passed = [
    student.name
    for student in students
    if student.score >= 80
]

@show passed
````

### Задание C — вычислить средний результат


````julia
average_score = sum(student.score for student in students) / length(students)

@show average_score
````

### Задание D — найти первого слушателя с результатом выше 95


````julia
first_excellent = nothing

for student in students
    if student.score > 95
        first_excellent = student
        break
    end
end

@show first_excellent
````

## **24. Итоги урока**

Основные правила Lesson 3:

1. `while` выполняется, пока условие истинно.
2. `for` итерируется по элементам итератора.
3. Диапазоны являются лёгкими итераторами и не обязательно создают полный массив.
4. Для коллекций обычно лучше использовать `for element in collection`, а не ручной счётчик.
5. `enumerate` возвращает номер итерации и значение.
6. `zip` позволяет параллельно итерироваться по нескольким коллекциям.
7. `break` завершает цикл, а `continue` пропускает текущую итерацию.
8. Вложенные циклы можно записывать в несколько строк или в сокращённой форме.
9. Julia хранит матрицы в column-major order.
10. `eachindex` — подходящий инструмент для безопасной итерации по индексам.
11. Comprehensions удобны для построения новых коллекций.
12. Dict comprehensions позволяют строить словари в одной декларации.
13. `map` предназначен прежде всего для преобразования значений, а не для побочных эффектов.
14. Генераторы помогают избегать части ненужных аллокаций.
15. Циклы в Julia являются полноценным инструментом производительного программирования.
