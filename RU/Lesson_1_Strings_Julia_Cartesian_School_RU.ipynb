## **1. `String` и `Char`**

### Теория

В Julia текст в основном представляется двумя типами:

| Тип | Пример | Значение |
|---|---|---|
| `String` | `"Julia"` | строка, то есть последовательность символов |
| `Char` | `'J'` | один символ Unicode |

Главное различие:

```julia
"J"   # String
'J'   # Char
```


### Пример


````julia
word = "Julia"
letter = 'J'

@show word typeof(word)
@show letter typeof(letter)
````

### Важно

Одинарные кавычки `'...'` предназначены для записи **одного символа**.

Поэтому:

```julia
'Julia'
```

— некорректная запись.


### Типичная ошибка

Начинающие иногда ожидают, что `'текст'` будет строкой — как в некоторых других языках.

В Julia:

- `"текст"` → `String`,
- `'т'` → `Char`.

## **4. Интерполяция строк**

### Теория

Julia позволяет вставлять значения непосредственно в строку с помощью `$`.

Для отдельной переменной:

```julia
"$name"
```

Для целого выражения:

```julia
"$(a + b)"
```


### Пример


````julia
name = "Анна"
num_fingers = 10
num_toes = 10

println("Меня зовут $name.")
println("У меня $num_fingers пальцев на руках и $num_toes пальцев на ногах.")
println("Всего это $(num_fingers + num_toes) пальцев.")
````

### Анализ

Выражение:

```julia
"$(num_fingers + num_toes)"
```

сначала вычисляет сумму, а затем вставляет результат в строку.


### Типичная ошибка

Не записывайте сложное выражение без скобок:

```julia
"$a + $b = $a + $b"
```

Если нужно вычислить результат:

```julia
"$a + $b = $(a + b)"
```

Чаще всего используются:

- `string(...)`,

### Пример — `string`

````julia
left = "Cartesian "
right = "School"

full = left * right
@show full
````

### Важно

В Julia для конкатенации строк используется `*`, а не `+`.

Это осознанный дизайн языка: `*` хорошо соответствует идее композиции/соединения объектов.


### Типичная ошибка

Это не является идиоматичной конкатенацией строк в Julia:

```julia
"Julia" + "Lang"
```

Используйте:

```julia
"Julia" * "Lang"
```

## **8. Unicode и UTF-8**

### Теория

Julia использует для `String` кодировку **UTF-8**.

Это важно, поскольку один символ Unicode может занимать разное количество байтов.

Например:

- ASCII-символ, такой как `A`, обычно занимает 1 байт;
- многие кириллические символы занимают 2 байта;
- многие emoji занимают 4 байта.

Поэтому число **символов** и число **единиц кодирования** не обязаны совпадать.


### Пример


````julia
en = "Hello"
pl = "Zażółć"
ru = "Привет"
emoji = "🚀"

println("Hello  -> znaki: ", length(en), ", bajty: ", ncodeunits(en))
println("Zażółć  -> znaki: ", length(pl), ", bajty: ", ncodeunits(pl))
println("Привет  -> znaki: ", length(ru), ", bajty: ", ncodeunits(ru))
println("🚀      -> znaki: ", length(emoji), ", bajty: ", ncodeunits(emoji))
````

### Важно

- `length(s)` — число символов;
- `ncodeunits(s)` — число единиц кодирования UTF-8.

Эти функции не являются взаимозаменяемыми.


## **9. Индексирование строк**

### Теория

Это один из ключевых разделов урока.

В Julia индекс строки не означает автоматически «номер символа».

Индекс указывает **корректную позицию в представлении UTF-8**.

Для чистого ASCII разница почти незаметна:


````julia
ascii_text = "Hello"

@show ascii_text[1]
@show ascii_text[2]
@show ascii_text[5]
````

Для Unicode-текста последовательные корректные индексы могут не быть последовательными целыми числами.


````julia
ru = "Привет"

@show collect(eachindex(ru))
````

Для `"Привет"` корректные индексы обычно имеют вид:

```text
1, 3, 5, 7, 9, 11
```

Каждая буква здесь занимает два байта.


### Пример — корректный индекс


````julia
ru = "Привет"
@show ru[1]
````

### Пример — намеренно некорректный индекс

Следующий код демонстрирует учебный пример ошибки:

```julia
ru[2]
```

Индекс `2` попадает внутрь многобайтового символа, поэтому Julia выдаст `StringIndexError`.


````julia
ru = "Привет"

bad_index_detected = try
    ru[2]
    false
catch e
    e isa StringIndexError
end

@show bad_index_detected
````

### Важно

Мы не обучаемся путём намеренного «ломания» notebook-файла. Вместо того чтобы оставлять ячейку с необработанным исключением, мы перехватываем его и проверяем его тип.


## **10. Безопасный обход текста**

### Теория

Если нужны только символы, самый простой и безопасный вариант:


````julia
for c in "Привет"
    print(c, " ")
end
println()
````

### Важно

Для произвольного Unicode-текста не используйте:

```julia
for i in 1:length(s)
    println(s[i])
end
```

Такой код может работать для ASCII и перестать работать для польских букв, кириллицы или emoji.


### Когда нужны индексы — `eachindex`


````julia
text = "Zażółć 🚀"

for i in eachindex(text)
    println("индекс ", lpad(i, 2), " -> ", text[i])
end
````

`eachindex(text)` возвращает только те индексы, которые корректны для данной строки.


## **11. `firstindex`, `lastindex`, `nextind`, `prevind`**

### Теория

Для ручной навигации по строке используются функции:

| Функция | Значение |
|---|---|
| `firstindex(s)` | первый корректный индекс |
| `lastindex(s)` | индекс последнего символа |
| `nextind(s, i)` | индекс следующего символа |
| `prevind(s, i)` | индекс предыдущего символа |


````julia
text = "Привет"

@show firstindex(text)
@show lastindex(text)
@show length(text)
@show ncodeunits(text)
````

### Важно

`lastindex(s)` и `length(s)` — разные понятия.

Для текста с многобайтовыми символами их значения могут различаться.


````julia
text = "Привет"

i = firstindex(text)
@show text[i]

i = nextind(text, i)
@show text[i]

i = nextind(text, i)
@show text[i]

i = prevind(text, i)
@show text[i]
````

### Последний символ — `end`


````julia
text = "Привет 🚀"

@show text[end]
@show text[end] == text[lastindex(text)]
````

    ...

## **14. Поиск в тексте**

### Теория

Наиболее часто используются:

| Функция | Назначение |
|---|---|
| `occursin(needle, haystack)` | проверяет наличие фрагмента |
| `startswith(s, prefix)` | проверяет начало строки |
| `endswith(s, suffix)` | проверяет конец строки |

### Пример


````julia
text = "Julia быстрая, и Julia читаемая."

@show occursin("Julia", text)
@show occursin("Python", text)
@show startswith(text, "Julia")
@show endswith(text, "читаемая.")
````

````julia
text = "abc Julia xyz"

position = findfirst("Julia", text)

@show position
````

### Анализ

Если шаблон является строкой, результатом может быть диапазон индексов, соответствующий найденному фрагменту.

## **20. Текст и числа**

### Теория

Это повторение материала Lesson 0, но при работе со строками оно особенно важно.

- `parse` — интерпретирует текст как значение;
- `tryparse` — безопаснее работает с пользовательским вводом;
- `string` — создаёт текстовое представление значения.


````julia
text_number = "123"

number = parse(Int, text_number)
safe_number = tryparse(Int, text_number)
back_to_text = string(number)

@show number typeof(number)
@show safe_number typeof(safe_number)
@show back_to_text typeof(back_to_text)
````

### Типичная ошибка

```julia
parse(Int, "12.5")
```

не сработает, потому что `"12.5"` не описывает целое число.

Правильно:

```julia
parse(Float64, "12.5")
```


````julia
@show parse(Float64, "12.5")
@show tryparse(Int, "12.5")
````

## **22. Типичные ошибки начинающих**

| Ошибка | Почему возникает | Правильный подход |
|---|---|---|
| `'Julia'` | `Char` может содержать только один символ | `"Julia"` |
| `"a" + "b"` | `+` не является оператором конкатенации `String` | `"a" * "b"` |
| `s[i]` при `i = 1:length(s)` | индексы UTF-8 не обязаны быть последовательными | `for c in s` или `eachindex(s)` |
| `s[length(s)]` | `length` не обязан быть корректным последним индексом | `s[end]` |
| попытка `s[1] = 'X'` | `String` неизменяем | создать новую строку |
| `convert(Int, "12")` | текст нужно разобрать | `parse(Int, "12")` |
| отсутствие обработки неверного ввода | `parse` выбрасывает исключение | `tryparse` |


## **23. Мини-проект — нормализация пользовательских данных**

### Цель

Предположим, пользователь ввёл:

```text
"   SIERGEJ.SOBolewski@Example.COM   "
```

Нужно:

1. удалить пробелы по краям;
2. привести текст к нижнему регистру;
3. проверить наличие символа `@`;
4. разделить логин и домен.


````julia
raw_email = "   SIERGEJ.SOBolewski@Example.COM   "

email = lowercase(strip(raw_email))
has_at = occursin("@", email)
parts = split(email, "@")

println("email: ", email)
println("содержит @: ", has_at)

if has_at && length(parts) == 2
    login, domain = parts
    println("логин : ", login)
    println("домен : ", domain)
end
````

### Анализ

В одном коротком примере мы использовали:

- `strip`,
- `lowercase`,
- `occursin`,
- `split`.

Это типичный небольшой pipeline обработки текста.

### Задание 24.3 — Unicode и безопасные индексы

Дана строка:

```julia
phrase = "Привет, Julia! 🚀"
```

Создайте:

- `n_chars` — число символов;
- `n_bytes` — число единиц кодирования;
- `idxs` — все корректные индексы;
- `last_char` — последний символ;
- `bad_index` — `true`, если `phrase[2]` вызывает `StringIndexError`.


````julia
phrase = "Привет, Julia! 🚀"

# Ваше решение:
````

### Пример решения 24.3


````julia
phrase = "Привет, Julia! 🚀"

n_chars = length(phrase)
n_bytes = ncodeunits(phrase)
idxs = collect(eachindex(phrase))
last_char = phrase[end]

bad_index = try
    phrase[2]
    false
catch e
    e isa StringIndexError
end

println("символов        : ", n_chars)
println("единиц UTF-8    : ", n_bytes)
println("индексы         : ", idxs)
println("последний символ: ", last_char)
println("StringIndexError: ", bad_index)

@assert length(idxs) == n_chars
@assert last_char == '🚀'
@assert bad_index === true
````

## **26. Итоги урока**

Ключевые правила Lesson 1:

1. `String` и `Char` — разные типы.
2. Строки записываются в `"..."`, символы — в `'...'`.
3. Интерполяция `$` — основной способ создавать читаемые сообщения.
4. Конкатенация строк выполняется оператором `*`.
5. `repeat` и `^` позволяют повторять текст.
6. `String` неизменяем.
7. Julia хранит строки в UTF-8.
8. Число символов не обязано совпадать с числом единиц кодирования.
9. Корректные индексы Unicode-текста не обязаны быть последовательными числами.
10. Для обхода символов предпочтительно использовать `for c in s`.
11. Если нужны индексы, используйте `eachindex`.
12. `s[end]` — более безопасный способ получить последний символ, чем `s[length(s)]`.
13. `occursin`, `replace`, `split`, `join`, `strip`, `lowercase` и `uppercase` образуют базовый набор инструментов для работы с текстом.
14. `parse` интерпретирует текст, а `tryparse` позволяет безопасно обрабатывать некорректный ввод.
