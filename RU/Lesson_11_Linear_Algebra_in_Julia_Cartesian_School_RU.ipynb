## **8. Adjoint — оператор `'`**

Для вещественных данных:


````julia
A = rand(2, 3)

@show transpose(A)
@show A'
@show transpose(A) == A'
````

Для комплексных данных ситуация отличается.


````julia
Z = [
    1 + 2im  3 - 4im
    5 + 0im  6 + 7im
]

@show transpose(Z)
@show Z'
````

### Важно

- `transpose(Z)` только меняет оси;
- `Z'` выполняет **комплексное сопряжение + транспонирование**.

Поэтому оператор `'` называют adjoint.


## **9. `A' * A` и `A * A'`**

Эти два произведения очень часто встречаются в анализе данных и методах наименьших квадратов.


````julia
A = rand(5, 3)

G1 = A' * A
G2 = A * A'

@show size(G1)
@show size(G2)
@show issymmetric(G1)
@show issymmetric(G2)
````

### Анализ

Для `A` размера `m × n`:

- `A' * A` имеет размер `n × n`;
- `A * A'` имеет размер `m × m`.

Обе матрицы являются эрмитовыми/симметричными и положительно полуопределёнными.


## **10. Квадратная система**

Решаем:

\[
Ax=b
\]

где `A` — квадратная матрица.


````julia
A = [
    3.0 1.0
    1.0 2.0
]

b = [9.0, 8.0]

x = A \ b

@show x
@show A * x
@assert A * x ≈ b
````

### Важно

Оператор `\` — идиоматичный способ решения линейной системы в Julia.


## **11. Переопределённая система — tall matrix**

Если у `A` больше строк, чем столбцов:

\[
m > n
\]

обычно у нас больше уравнений, чем неизвестных.


````julia
A_tall = [
    1.0 1.0
    1.0 2.0
    1.0 3.0
    1.0 4.0
]

b_tall = [1.2, 1.9, 3.2, 3.9]

x_ls = A_tall \ b_tall

@show x_ls
````

## **12. Метод наименьших квадратов**

Для переопределённой системы точного решения `A*x == b` обычно не существует.

Julia возвращает решение, минимизирующее норму residual:


````julia
residual = A_tall * x_ls - b_tall

@show residual
@show norm(residual)
````

Математически ищем:

\[
\min_x \|Ax-b\|_2
\]


### Важно

Не нужно вручную формировать нормальные уравнения:

```julia
(A' * A) \ (A' * b)
```

Прямое решение:

```julia
A \ b
```

обычно численно устойчивее.


## **13. Недоопределённая система — wide matrix**

Если:

\[
m < n
\]

у нас меньше уравнений, чем неизвестных.


````julia
A_wide = [
    1.0 0.0 1.0
    0.0 1.0 1.0
]

b_wide = [1.0, 1.0]

x_wide = A_wide \ b_wide

@show x_wide
@show A_wide * x_wide
````

## **14. Решение минимальной нормы**

Для недоопределённой системы полного строкового ранга существует бесконечно много решений.

`A \ b` возвращает решение в соответствии с алгоритмом Julia, которое в соответствующем случае обычно является решением минимальной нормы.


````julia
@show norm(x_wide)
@assert A_wide * x_wide ≈ b_wide
````

### Важно

Не интерпретируйте недоопределённую систему как квадратную.

Отсутствие единственного решения — математическое свойство задачи, а не ошибка Julia.


## **15. Система неполного ранга**

Матрица может содержать линейно зависимые столбцы или строки.


````julia
A_rd = [
    1.0 2.0
    2.0 4.0
    3.0 6.0
]

@show rank(A_rd)
````

Её ранг равен `1`, хотя она содержит два столбца.


````julia
b_rd = [1.0, 2.0, 3.0]

x_rd = A_rd \ b_rd

@show x_rd
@show norm(A_rd * x_rd - b_rd)
````

### Важно

В системах неполного ранга решение может быть неединственным, а поведение зависит от используемого алгоритма/факторизации.

Для точной диагностики часто применяют SVD.


## **16. Residual — как оценить решение?**

Residual:

\[
r = Ax-b
\]

показывает, насколько хорошо решение удовлетворяет системе.


````julia
A = rand(8, 3)
x_true = [2.0, -1.0, 0.5]
b = A * x_true + 0.01 .* randn(8)

x_est = A \ b
r = A * x_est - b

@show norm(r)
````

### Хорошая практика

После решения численной задачи часто полезно проверить:

- `norm(A*x-b)`;
- относительный residual;
- conditioning;
- ранг матрицы.


````julia
relative_residual = norm(r) / norm(b)

@show relative_residual
````

## **34. Типичные ошибки начинающих**

| Ошибка | Причина | Правильный подход |
|---|---|---|
| `A .* B` вместо `A * B` | путаница broadcasting и линейной алгебры | использовать `*` |
| `A' == transpose(A)` для Complex | adjoint выполняет сопряжение | различать операции |
| `inv(A)*b` | явная обратная матрица не нужна | `A \ b` |
| нормальные уравнения для least squares | хуже численная устойчивость | `A \ b` |
| предположение об уникальности wide system | недостаточно уравнений | анализировать пространство решений |
| игнорирование rank deficiency | линейная зависимость | использовать `rank`, `svd` |
| отсутствие проверки residual | решение не верифицировано | `norm(A*x-b)` |
| преобразование всего в плотную `Matrix` | теряется структура | `Diagonal`, `Symmetric`, `Hermitian` |

## **36. Мини-проект — линейная регрессия с нуля**

### Цель

Построим простую линейную регрессию без внешнего статистического пакета.

Модель:

\[
y = \beta_0 + \beta_1 x + \varepsilon
\]


````julia
xdata = collect(1.0:10.0)
ydata = [2.2, 4.1, 6.0, 8.3, 10.2, 11.8, 14.1, 16.2, 17.9, 20.3]

X = hcat(ones(length(xdata)), xdata)

β = X \ ydata

@show β
````

### Предсказание


````julia
ŷ = X * β

@show ŷ
````

### Residual


````julia
r = ydata - ŷ

@show r
@show norm(r)
````

### Среднеквадратичная ошибка


````julia
mse = sum(abs2, r) / length(r)

@show mse
````

1. построили матрицу проекта `X`;
2. решили задачу least squares через `X \ y`;
3. вычислили предсказание;
4. рассчитали residual;
5. вычислили MSE.

Это отличный пример практического применения линейной алгебры.

## **38. Итоги урока**

Основные правила Lesson 11:

1. Julia представляет векторы и матрицы как специализированные массивы.

3. `A*B` — матричное умножение, `A.*B` — поэлементное.
