## **1. Что такое функция?**

### Теория

Функция — это именованный или анонимный фрагмент программы, который:

1. может принимать входные данные — **аргументы**;
2. выполняет определённую операцию;
3. может возвращать результат.

Функции позволяют:

- уменьшать дублирование кода;
- делить программу на небольшие логические части;
- повторно использовать поведение;
- тестировать отдельные фрагменты системы;
- строить расширяемые API.


### Пример

Вместо многократного написания:

```julia
3 + 5
10 + 20
100 + 200
```

можно создать функцию:


````julia
function add(x, y)
    return x + y
end

@show add(3, 5)
@show add(10, 20)
@show add(100, 200)
````

## **2. Определение через `function`**

### Синтаксис

```julia
function имя(аргументы)
    тело
    return результат
end
```


````julia
function greet(name)
    message = "Привет, $name!"
    return message
end

println(greet("Анна"))
````

### Анализ

- `greet` — имя функции;
- `name` — параметр;
- `"Анна"` — аргумент, переданный при вызове;
- `return` — возвращает результат и завершает выполнение функции.

Это очень идиоматичная форма записи для небольших функций.

## **4. Анонимные функции**

### Теория

Анонимная функция не обязана иметь собственное имя.

Синтаксис:

```julia
x -> выражение
```


````julia
square_lambda = x -> x^2

@show square_lambda(4)
````

### Несколько аргументов


````julia
add_lambda = (x, y) -> x + y

@show add_lambda(10, 20)
````

### Когда использовать?

Анонимные функции особенно удобны как короткие аргументы для:

- `map`;
- `filter`;

## **5. Параметры и аргументы**

### Теория

**Параметр** — это имя в определении функции.

**Аргумент** — конкретное значение, переданное при вызове.

## **7. Именованные аргументы**

### Теория

Именованные аргументы записываются после точки с запятой `;`.


````julia
function create_message(name; language="ru", uppercase_text=false)
    text = language == "ru" ? "Привет, $name!" : "Hello, $name!"
    return uppercase_text ? uppercase(text) : text
end

@show create_message("Анна")
@show create_message("Анна"; language="en")
@show create_message("Анна"; language="en", uppercase_text=true)
````

### Почему это полезно?

Именованные аргументы:

- повышают читаемость;
- удобны для конфигурации;
- позволяют иметь много необязательных параметров.


## **8. Переменное число аргументов — varargs**

### Теория

Оператор `...` позволяет функции принимать произвольное число аргументов.


````julia
function add_all(values...)
    return sum(values)
end

@show add_all(1, 2, 3)
@show add_all(10, 20, 30, 40)
````

### Анализ

Внутри функции `values` является кортежем (`Tuple`) переданных аргументов.


````julia
function inspect_args(values...)
    @show values
    @show typeof(values)
    return nothing
end

inspect_args(1, "Julia", true)
````

## **9. `return` и последнее выражение**

### Теория

Julia автоматически возвращает значение последнего выполненного выражения в функции.


````julia
function multiply_explicit(x, y)
    return x * y
end

function multiply_implicit(x, y)
    x * y
end

@assert multiply_explicit(3, 4) == multiply_implicit(3, 4)
````

### Важно

Явный `return` особенно полезен:

- для раннего завершения функции;
- когда он повышает читаемость;
- в более сложной логике.


````julia
function safe_sqrt(x)
    x < 0 && return nothing
    return sqrt(x)
end

@show safe_sqrt(9)
@show safe_sqrt(-1)
````

## **10. Возврат нескольких значений**

### Теория

Julia может вернуть несколько значений как `Tuple`.


````julia
function minmax_values(values)
    return minimum(values), maximum(values)
end

lo, hi = minmax_values([8, 2, 11, 4])

@show lo hi
````

### `NamedTuple` как более читаемый результат


````julia
function stats(values)
    return (
        minimum = minimum(values),
        maximum = maximum(values),
        average = sum(values) / length(values),
    )
end

result = stats([10, 20, 30, 40])

@show result
@show result.average
````

## **11. Duck typing в Julia**

### Теория

Duck typing означает:

> если объект поддерживает операции, которые нужны функции, функция может с ним работать.

Не обязательно заранее жёстко ограничивать тип каждого аргумента.


````julia
function combine(x, y)
    return x + y
end

@show combine(3, 5)
@show combine(2.5, 1.25)
````

### Важно

В Julia duck typing не означает отсутствие типов. У Julia развитая система типов, но функции часто остаются обобщёнными и требуют лишь поддержки конкретных операций.


### Пример с собственной семантикой оператора

Для строк в Julia конкатенация выполняется оператором `*`, а не `+`, поэтому функцию следует проектировать в соответствии с операцией, которая действительно поддерживается данным типом.


````julia
function concatenate(x, y)
    return x * y
end

@show concatenate("Hello, ", "Julia!")
````

## **14. Multiple dispatch**

### Теория

Multiple dispatch выбирает метод на основе типов **всех** аргументов, участвующих в dispatch.

Это одна из центральных особенностей Julia.


````julia
interact(x::Int, y::Int) = "Int + Int"
interact(x::Int, y::AbstractString) = "Int + String"
interact(x::AbstractString, y::Int) = "String + Int"
interact(x::AbstractString, y::AbstractString) = "String + String"

@show interact(1, 2)
@show interact(1, "Julia")
@show interact("Julia", 1)
@show interact("Julia", "Lang")
````

### Анализ

## **15. Параметры типов и `where`**

### Теория

Можно выразить зависимость между типами аргументов.


````julia
function same_type_add(x::T, y::T) where {T<:Number}
    return x + y
end

@show same_type_add(2, 3)
@show same_type_add(2.5, 3.5)
````

### Анализ

`where {T<:Number}` означает, что:

- оба аргумента имеют один и тот же тип `T`;
- `T` должен быть подтипом `Number`.


## **16. Изменяющие функции и соглашение `!`**

### Теория

В Julia знак `!` в имени функции — это соглашение, указывающее, что функция изменяет один или несколько своих аргументов.


### Неизменяющая функция


````julia
function doubled(values)
    return values .* 2
end

original = [1, 2, 3]
result = doubled(original)

@show original
@show result
````

### Изменяющая функция


````julia
function double!(values)
    for i in eachindex(values)
        values[i] *= 2
    end
    return values
end

original = [1, 2, 3]
double!(original)

@show original
````

### Важно

`!` не изменяет семантику языка и не навязывается компилятором.

Это очень устойчивое соглашение об именовании в экосистеме Julia.

## **18. Функции высшего порядка**

### Теория

Функция высшего порядка — это функция, которая:

- принимает другую функцию как аргумент;
- или возвращает функцию;
- или делает и то и другое.


````julia
function transform(f, values)
    return [f(x) for x in values]
end

@show transform(x -> x^2, 1:5)
````

## **23. Closures — замыкания**

### Теория

Анонимная функция может «запомнить» переменные из окружающей области видимости.

## **25. Композиция функций**

## **27. Рекурсия**

### Теория

Рекурсивная функция вызывает саму себя.

У неё обязательно должно быть **базовое условие**, останавливающее рекурсию.


````julia
function factorial_recursive(n::Integer)
    n < 0 && throw(ArgumentError("n должно быть неотрицательным"))
    n <= 1 && return one(n)
    return n * factorial_recursive(n - 1)
end

@show factorial_recursive(5)
````

### Важно

Рекурсия элегантна для некоторых задач, но не всегда является лучшим решением с точки зрения производительности.

Во многих случаях итерация будет проще или эффективнее.


## **28. Область видимости переменных в функциях**

### Теория

Переменные, созданные внутри функции, имеют локальную область видимости.


````julia
global_value = 100

function scope_demo(x)
    local_value = x + 1
    return global_value + local_value
end

@show scope_demo(5)
````

### Хорошая практика

По возможности функции должны получать данные через аргументы, а не зависеть от глобального состояния.

Это улучшает:

- тестируемость;
- читаемость;
- предсказуемость;
- повторное использование кода.


## **29. Документирование функций**

### Теория

В Julia документацию функции можно разместить непосредственно над определением в виде docstring.


````julia
"""
    circle_area(r)

Возвращает площадь круга радиуса `r`.

# Аргументы
- `r`: радиус круга, должен быть неотрицательным.

# Пример
```julia
circle_area(2)
```
"""
function circle_area(r)
    r < 0 && throw(ArgumentError("Радиус должен быть неотрицательным"))
    return π * r^2
end
````

### Хорошая практика

Хорошая документация должна объяснять:

- что делает функция;
- какие аргументы она принимает;
- что возвращает;
- какие ограничения имеет;
- при необходимости приводить короткий пример.


## **30. Типичные ошибки начинающих**

| Ошибка | Причина | Правильный подход |
|---|---|---|
| аннотации типов повсюду | смешение типизации и оптимизации | ограничивать типы только при реальной необходимости |
| длинный `if x isa ...` | поведение зависит от типа | рассмотреть multiple dispatch |
| отсутствие базового случая в рекурсии | функция не сможет завершиться | определить условие остановки |
| использование изменяющей функции без `!` в имени | нарушение принятого стиля | использовать соглашение `!` |
| побочные эффекты внутри `map` без необходимости | неясное намерение | использовать `for` |
| глобальное состояние вместо аргументов | хуже тестируемость и предсказуемость | передавать данные через параметры |
| слишком длинные функции | несколько ответственностей | делить на небольшие функции |

### Задание 31.7 — изменяющая функция

Напишите:

````julia
# Ваше решение:
````

### Пример решения 31.7


````julia
function increment!(values)
    for i in eachindex(values)
        values[i] += 1
    end
    return values
end

data = [1, 2, 3]
increment!(data)

@assert data == [2, 3, 4]
````

## **32. Мини-проект — pipeline анализа результатов**

### Данные


````julia
scores = [45, 78, 92, 67, 88, 100, 59, 73]
````

### Цель

Построим набор небольших функций, которые:

1. проверяют корректность результата;
2. классифицируют результат;
3. выбирают зачтённые результаты;
4. вычисляют среднее значение.


````julia
is_valid_score(x) = 0 <= x <= 100
is_passed(x) = x >= 60

function grade(score)
    !is_valid_score(score) && return "INVALID"

    if score >= 90
        return "A"
    elseif score >= 80
        return "B"
    elseif score >= 70
        return "C"
    elseif score >= 60
        return "D"
    else
        return "F"
    end
end
````

### Применение


````julia
valid_scores = filter(is_valid_score, scores)
passed_scores = filter(is_passed, valid_scores)
grades = map(grade, valid_scores)
average = sum(valid_scores) / length(valid_scores)

@show valid_scores
@show passed_scores
@show grades
@show average
````

### Отчёт


````julia
report = [
    (score = score, grade = grade(score), passed = is_passed(score))
    for score in valid_scores
]

for row in report
    println("результат=$(row.score), оценка=$(row.grade), зачтено=$(row.passed)")
end
````

### Анализ

Мини-проект показывает главное преимущество функций:

вместо одного длинного блока программы мы создаём небольшие независимые элементы с одной ответственностью.

Основные правила Lesson 5:

1. Функции — базовый инструмент модульности кода.
