## **1. Что такое пакет?**

### Теория

Пакет в Julia — это организованный набор кода, предназначенный для повторного использования.

Пакет может содержать:

- модули;
- функции;
- типы;
- тесты;
- документацию;
- зависимости от других пакетов;
- метаданные проекта.


### Пакет, модуль и окружение — не одно и то же

| Понятие | Значение |
|---|---|
| **модуль** | пространство имён в коде Julia |
| **пакет** | распространяемый проект Julia, обычно содержащий модуль |
| **окружение** | набор зависимостей и версий, используемых конкретным проектом |


## **2. Экосистема пакетов Julia**

### Теория

Julia располагает развитой экосистемой пакетов, охватывающей, в частности:

- анализ данных;
- статистику;
- машинное обучение;
- научные вычисления;
- оптимизацию;
- визуализацию;
- дифференциальные уравнения;
- обработку сигналов;
- программирование GPU;
- интеграцию с Python, R, C и другими языками.

С точки зрения разработчика основным инструментом управления зависимостями является встроенный менеджер пакетов **Pkg**.


### Важно

Не следует бездумно устанавливать пакеты «в Julia вообще».

Профессиональная работа строится на **проектных окружениях**, благодаря которым каждый проект может иметь собственный набор зависимостей.


## **3. REPL и его режимы**

### Теория

REPL означает:

**Read → Eval → Print → Loop**

Это интерактивная консоль Julia.

После запуска:

```sh
julia
```

вы увидите:

```text
julia>
```


### Основные режимы REPL

| Режим | Как войти | Назначение |
|---|---|---|
| Julia mode | режим по умолчанию | выполнение кода Julia |
| Pkg mode | `]` | управление пакетами |
| Help mode | `?` | документация |
| Shell mode | `;` | системные команды |


### Julia mode


```julia
julia> 2 + 3
5

julia> sqrt(16)
4.0
```


### Help mode


```text
julia> ?
help?> sqrt
```

Позволяет быстро посмотреть документацию функции или типа.


### Shell mode


```text
julia> ;
shell> pwd
```

Shell mode позволяет выполнять системные команды, не выходя из REPL.


### Важно

Shell mode зависит от операционной системы. Команды, доступные в Linux/macOS, могут отличаться от команд Windows.


## **4. Режим пакетов `pkg>`**

### Вход в режим

В REPL нажмите:

```text
]
```

Появится prompt вида:

```text
(@v1.x) pkg>
```


### Основные команды

| Команда | Назначение |
|---|---|
| `add PackageName` | добавляет пакет |
| `rm PackageName` | удаляет пакет |
| `update` | обновляет зависимости |
| `status` | показывает состояние окружения |
| `activate .` | активирует окружение в текущем каталоге |
| `instantiate` | восстанавливает окружение |
| `resolve` | повторно разрешает зависимости |
| `precompile` | выполняет precompile зависимостей |
| `pin PackageName` | фиксирует версию пакета |
| `free PackageName` | снимает pin / возвращает обычное управление версией |
| `develop PackageName` | использует пакет в режиме разработки |


### Выход из Pkg mode

Обычно используется:

- `Backspace`;
- либо `Ctrl+C`.


## **5. Модуль `Pkg` и программный API**

### Теория

Большинство операций из Pkg mode имеют эквивалент в коде Julia.

Сначала:


````julia
using Pkg
````

Например:

```julia
Pkg.add("Example")
Pkg.rm("Example")
Pkg.status()
Pkg.update()
```

Это удобно в скриптах, автоматизирующих настройку окружения.


### Важно

Операции `Pkg.add`, `Pkg.rm`, `Pkg.update` и аналогичные **изменяют состояние активного окружения**.

Поэтому в учебных notebook-файлах мы не запускаем их случайно в окружении курса.

### Pkg mode

```text

```


### API


````julia
using Pkg

# Пример:
# Pkg.add("Plots")
````

### Установка нескольких пакетов

```text
pkg> add CSV DataFrames Plots
```

или:


````julia
# Pkg.add(["CSV", "DataFrames", "Plots"])
````

### Что делает `add`?

Упрощённо:

1. выбирает совместимую версию пакета;
2. разрешает зависимости;
3. загружает необходимые пакеты;
4. обновляет `Project.toml`;
5. обновляет или создаёт `Manifest.toml`;
6. может запустить precompile.


## **7. `using` и `import`**

### Теория

После добавления пакета необходимо загрузить соответствующий модуль.


### `using`


````julia
using Statistics

values = [1, 2, 3, 4, 5]
@show mean(values)
````

`using ModuleName` делает доступными экспортируемые модулем имена.


### `import`


````julia
import Statistics

@show Statistics.mean([10, 20, 30])
````

### Практическое различие

`import ModuleName` поощряет явное использование квалифицированных имён:

```julia
Statistics.mean(...)
```

`using ModuleName` может предоставить экспортируемые символы без префикса модуля.


### Импорт отдельного имени


````julia
import Statistics: median

@show median([1, 10, 100])
````

### Важно

В крупных проектах явные имена модулей могут повысить читаемость и уменьшить вероятность конфликтов имён.

### Pkg mode

Удаление пакета обновляет окружение проекта. Это не обязательно означает немедленное удаление всех файлов пакета из Julia depot, поскольку они могут использоваться другими окружениями.


## **9. Обновление пакетов**

### Все пакеты

```text
pkg> update
```


### Конкретный пакет

```text
pkg> update Plots
```


### API


````julia
# Pkg.update()
# Pkg.update("Plots")
````

### Важно

`update` не означает «всегда установить самую новую доступную версию».

Pkg учитывает:

- ограничения совместимости;
- зависимости других пакетов;

## **11. Проектные окружения**

### Теория

Окружение сообщает Julia:

- какие пакеты относятся к проекту;
- какие версии разрешены;
- какие версии фактически были разрешены resolver-ом.

Обычно окружение проекта находится в каталоге проекта.


### Почему изоляция важна?

Без изоляции проект A может случайно начать зависеть от пакета, установленного для проекта B.

Окружения устраняют эту проблему и делают работу воспроизводимой.


## **12. `Project.toml`**

### Теория

`Project.toml` декларативно описывает проект.

Он может содержать, в частности:

- имя проекта;
- UUID;
- версию;
- прямые зависимости;
- ограничения совместимости.


### Пример


```toml
name = "MyJuliaProject"
uuid = "12345678-1234-1234-1234-123456789abc"
version = "0.1.0"

[deps]
CSV = "336ed68f-0bac-5ca0-87d4-7b16caf5d00b"
DataFrames = "a93c6f00-e57d-5684-b7b6-d8193f3e46c0"

[compat]
julia = "1"
CSV = "0.10"
DataFrames = "1"
```


### Важно

`Project.toml` описывает **намерение проекта** — его прямые зависимости и ограничения.


## **13. `Manifest.toml`**

### Теория

`Manifest.toml` описывает **конкретный разрешённый граф зависимостей**.

Он может содержать сведения о:

- точных версиях;
- UUID;
- транзитивных зависимостях;
- источниках пакетов;
- идентификаторах ревизий репозиториев.


### Различие

| Файл | Роль |
|---|---|
| `Project.toml` | что требуется проекту |
| `Manifest.toml` | какие конкретные версии были разрешены |


### Важно

Именно сочетание `Project.toml` + `Manifest.toml` создаёт прочную основу воспроизводимости.

## **15. Временное окружение**

### Цель

Для экспериментов идеально подходит временное окружение:


````julia
using Pkg

course_env = dirname(Pkg.project().path)
println("Окружение курса: ", course_env)

Pkg.activate(; temp=true)

temp_env = dirname(Pkg.project().path)
println("Временное окружение: ", temp_env)
````

### Зачем это нужно?

Операции в таком окружении:

- не изменяют зависимости курса;
- позволяют безопасно тестировать пакеты;
- удобны в учебных notebook-файлах.


### Важно

После завершения упражнений мы вернёмся к окружению курса.


## **16. `instantiate` — восстановление окружения**

Вместо этого:

````julia
# Pkg.instantiate()
````

`instantiate` загружает отсутствующие зависимости в соответствии с файлами окружения.


## **17. `resolve` — повторное разрешение зависимостей**

### Теория

`resolve` обновляет зависимости в `Manifest.toml` в соответствии с текущим состоянием `Project.toml`.


```text
pkg> resolve
```


````julia
# Pkg.resolve()
````

### Когда использовать?

Например, если:

- изменились прямые зависимости;
- вручную редактировался `Project.toml`;
- локально разрабатываемые пакеты изменили свои зависимости.


## **18. `precompile`**

### Теория

Пакеты Julia могут проходить precompile, чтобы ускорить последующую загрузку.


```text
pkg> precompile
```


````julia
# Pkg.precompile()
````

### Важно

Precompile не означает, что весь код приложения «скомпилирован навсегда».

Julia по-прежнему может генерировать специализации методов во время выполнения программы.


## **19. `pin` и `free`**

### `pin`

Фиксирует пакет на определённой версии:


```text
pkg> pin Example
```


### `free`

Снимает pin и возвращает обычное управление версией:


```text
pkg> free Example
```


### Когда это полезно?

Pin может быть полезен при:

- воспроизведении ошибки;
- временной фиксации зависимостей;
- тестировании совместимости.

Однако он не заменяет хорошо спроектированные ограничения `[compat]`.

## **24. Ограничения совместимости `[compat]`**

### Теория

Раздел `[compat]` определяет, с какими версиями зависимостей проект заявляет совместимость.


```toml
[compat]
julia = "1"
DataFrames = "1"
CSV = "0.10"
```


### Почему это важно?

Без разумных ограничений совместимости обновление одной зависимости в будущем может привести к неожиданной несовместимости.


### Хорошая практика

Для библиотек `[compat]` особенно важен, поскольку сообщает resolver-у Pkg, с какими версиями пакет действительно совместим.


## **25. Глобальное окружение и окружение проекта**

### Теория

Julia имеет пользовательское окружение по умолчанию, например:

```text
@v1.x
```

Оно удобно для быстрых экспериментов.

Для серьёзного проекта, однако, предпочтительно:


```text
pkg> activate .
```


### Почему?

Окружение проекта:

- изолирует зависимости;
- проще версионируется;
- проще восстанавливается;
- уменьшает риск ситуации «работает только на моём компьютере».


## **26. `LOAD_PATH`**

### Теория

Julia использует стек окружений, определяющий, откуда могут загружаться пакеты и модули.


````julia
@show LOAD_PATH
````

### Важно

Для начинающего наиболее важно:

- активировать правильный проект;
- не полагаться случайно на пакеты, присутствующие только в глобальном окружении.


## **27. Воспроизводимость окружения**

### Теория

Воспроизводимый проект — это проект, который другой человек может восстановить на основе репозитория и инструкции.


### Минимальный workflow

```sh
git clone ...
cd project
julia --project=.
```

а затем в Julia:


```julia
using Pkg
Pkg.instantiate()
```


### Зачем `--project=.`?

Эта опция запускает Julia сразу с активным окружением текущего каталога.


## **28. Git и файлы окружения**

### Хорошая практика для приложений и курсов

Обычно в системе контроля версий хранятся:

- `Project.toml`;
- `Manifest.toml`.

Это позволяет восстановить точное состояние зависимостей.


### Библиотеки

Для пакетов/библиотек политика в отношении `Manifest.toml` может отличаться от политики для приложений.

Ключевыми остаются корректный `Project.toml` и `[compat]`, поскольку пользователь библиотеки работает в собственном окружении.


### Важно

Не редактируйте `Manifest.toml` вручную без конкретной причины и понимания последствий.

Обычно этим файлом должен управлять Pkg.


## **29. Notebook-файлы и безопасность зависимостей**

### Проблема

Вызов:

```julia
Pkg.add("SomePackage")
```

внутри notebook-файла может изменить активное окружение курса или проекта.


### Безопасный учебный шаблон

1. сохраните путь окружения курса;
2. переключитесь во временное окружение;
3. выполните эксперимент;
4. вернитесь к окружению курса.


````julia
using Pkg

course_env_demo = dirname(Pkg.project().path)
println("Текущее окружение: ", course_env_demo)

Pkg.activate(; temp=true)
println("Временное окружение: ", dirname(Pkg.project().path))
````

В этом месте можно безопасно экспериментировать с пакетами.


## **30. Типичные ошибки начинающих**

| Ошибка | Почему это проблема | Правильный подход |
|---|---|---|
| установка всего в глобальное окружение | проекты начинают случайно разделять зависимости | отдельное окружение проекта |
| `Pkg.add` в notebook-файле курса | изменяет файлы окружения | временное окружение |
| удаление `Manifest.toml` без причины | теряется точное решение версий | сохранять его для приложений/курсов |
| смешение `using` и установки | `using` не устанавливает пакет | сначала `add`, затем `using` |
| ожидание, что `update` игнорирует `[compat]` | resolver учитывает ограничения | проверить `status` и `[compat]` |
| использование `develop` как обычного `add` | отслеживаются локальные исходники | `develop` только для разработки |
| ручное редактирование `Manifest.toml` | легко нарушить решение зависимостей | использовать Pkg |
| отсутствие `instantiate` после клонирования | зависимости могут отсутствовать локально | `Pkg.instantiate()` |

## **35. Итоги урока**

Основные правила Lesson 6:

1. `Pkg` — встроенный менеджер пакетов Julia.
2. REPL имеет отдельный Pkg mode, который включается клавишей `]`.
3. Пакеты добавляются в активное окружение, а не «в абстрактный проект».
4. `using` и `import` загружают модули, но не устанавливают пакеты.
5. `Project.toml` описывает прямые зависимости и совместимость.
6. `Manifest.toml` описывает конкретное решение полного графа зависимостей.
7. `activate` выбирает окружение.
8. `Pkg.activate(; temp=true)` идеально подходит для экспериментов.
9. `instantiate` восстанавливает окружение после клонирования проекта.
10. `resolve` повторно разрешает зависимости.
11. `precompile` подготавливает пакеты к более быстрой загрузке.
12. `pin` фиксирует версию, а `free` снимает фиксацию.
13. `develop` предназначен для активной разработки пакета.
14. `[compat]` — основа контроля совместимости версий.
15. Проектное окружение предпочтительнее случайной зависимости от глобального окружения.
16. `Project.toml` и `Manifest.toml` критичны для воспроизводимости приложений и курсов.
17. Учебный notebook-файл не должен без предупреждения изменять окружение проекта.
