## **1. Что такое численные методы?**

- аналитическое решение слишком дорого вычислять;

## **3. Абсолютная ошибка**

\[

```julia
exact = sqrt(2)
approx = 1.414

abs_error = abs(exact - approx)

@show abs_error
```

## **4. Относительная ошибка**

\[

```julia
rel_error = abs(exact - approx) / abs(exact)

@show rel_error
```

### Важно

Относительная ошибка часто информативнее абсолютной, поскольку учитывает масштаб значения.

## **7. Численная устойчивость**

Численно устойчивый алгоритм не должен чрезмерно усиливать малые ошибки округления.

Это свойство **алгоритма**, а не только языка программирования.


### Пример — альтернативная форма выражения

Для очень малого `x` выражение:

\[
\sqrt{1+x}-1
\]

может терять точность.

Алгебраически эквивалентная, но численно более удачная форма:

\[
\frac{x}{\sqrt{1+x}+1}
\]


```julia
x = 1e-12

naive = sqrt(1 + x) - 1
stable = x / (sqrt(1 + x) + 1)

@show naive
@show stable
```

## **8. Conditioning задачи**

### Теория

Conditioning показывает, насколько чувствителен результат задачи к небольшим изменениям входных данных.

- хорошо обусловленная задача → малые изменения данных приводят к малым изменениям результата;
- плохо обусловленная задача → небольшая ошибка во входных данных может сильно изменить результат.

Это свойство **математической задачи**, а не реализации.


### Пример с матрицей Гильберта


```julia
using LinearAlgebra

function hilbert_matrix(n)
    [1.0 / (i + j - 1) for i in 1:n, j in 1:n]
end

H = hilbert_matrix(8)

@show cond(H)
```

### Важно

Устойчивый алгоритм не исправит плохо обусловленную задачу, но может избежать дополнительного ухудшения результата.


## **9. Конечные разности**

Производную можно аппроксимировать прямой разностью:

\[
f'(x) \approx \frac{f(x+h)-f(x)}{h}
\]


```julia
f(x) = sin(x)

function derivative_forward(f, x, h)
    (f(x + h) - f(x)) / h
end

approx = derivative_forward(f, 1.0, 1e-5)
exact = cos(1.0)

@show approx
@show exact
@show abs(approx - exact)
```

## **10. Центральная разность**

Более точной аппроксимацией часто является:

\[
f'(x) \approx \frac{f(x+h)-f(x-h)}{2h}
\]


```julia
function derivative_central(f, x, h)
    (f(x + h) - f(x - h)) / (2h)
end

approx_central = derivative_central(f, 1.0, 1e-5)

@show approx_central
@show abs(approx_central - exact)
```

## **11. Слишком малый шаг тоже может быть плохим**

Уменьшение `h` не улучшает результат бесконечно.

При очень малом `h` начинают доминировать ошибки округления.


```julia
hs = 10.0 .^ (-1:-1:-15)

errors = [
    abs(derivative_central(sin, 1.0, h) - cos(1.0))
    for h in hs
]

for (h, err) in zip(hs, errors)
    println("h=$h  error=$err")
end
```

## **13. Метод бисекции**

### Теория

Если непрерывная функция меняет знак на интервале `[a,b]`, можно многократно делить этот интервал пополам.


```julia
function bisection(f, a, b; atol=1e-12, maxiter=1000)
    fa = f(a)
    fb = f(b)

    fa == 0 && return a
    fb == 0 && return b
    signbit(fa) == signbit(fb) &&
        throw(ArgumentError("Функция должна менять знак на концах интервала."))

    left, right = a, b

    for _ in 1:maxiter
        mid = (left + right) / 2
        fm = f(mid)

        if abs(fm) <= atol || abs(right - left) <= atol
            return mid
        end

        if signbit(f(left)) != signbit(fm)
            right = mid
        else
            left = mid
        end
    end

    return (left + right) / 2
end

root_bisect = bisection(x -> x^2 - 2, 1.0, 2.0)

@show root_bisect
@show sqrt(2)
```

### Преимущество

Бисекция медленнее некоторых других методов, но очень предсказуема, если выполняются её предпосылки.


## **14. Метод Ньютона**

Итерационная формула:

\[
x_{n+1}=x_n-\frac{f(x_n)}{f'(x_n)}
\]


```julia
function newton(f, df, x0; atol=1e-12, maxiter=100)
    x = x0

    for _ in 1:maxiter
        fx = f(x)

        abs(fx) <= atol && return x

        dfx = df(x)
        abs(dfx) <= eps(Float64) &&
            throw(ArgumentError("Pochodna zbyt bliska zeru."))

        x_new = x - fx / dfx

        abs(x_new - x) <= atol && return x_new
        x = x_new
    end

    return x
end

froot(x) = x^2 - 2
dfroot(x) = 2x

root_newton = newton(froot, dfroot, 1.0)

@show root_newton
```

## **15. Бисекция vs Newton**

| Свойство | Бисекция | Newton |
|---|---|---|
| требует интервал | да | нет |
| требует производную | нет | да |
| скорость | линейная | часто очень высокая |
| надёжность | высокая при выполнении предпосылок | зависит от начальной точки и производной |
| риск расходимости | низкий | выше |

## **17. Метод прямоугольников**


```julia
function midpoint_rule(f, a, b, n)
    h = (b - a) / n
    total = zero(Float64)

    for i in 0:n-1
        xmid = a + (i + 0.5) * h
        total += f(xmid)
    end

    return h * total
end

I_mid = midpoint_rule(sin, 0.0, π, 1000)

@show I_mid
@show abs(I_mid - 2.0)
```

## **18. Метод трапеций**


```julia
function trapezoidal_rule(f, a, b, n)
    h = (b - a) / n
    total = (f(a) + f(b)) / 2

    for i in 1:n-1
        total += f(a + i*h)
    end

    return h * total
end

I_trap = trapezoidal_rule(sin, 0.0, π, 1000)

@show I_trap
@show abs(I_trap - 2.0)
```

## **19. Метод Симпсона**


```julia
function simpson_rule(f, a, b, n)
    iseven(n) || throw(ArgumentError("n должно быть чётным"))

    h = (b - a) / n
    total = f(a) + f(b)

    for i in 1:n-1
        x = a + i*h
        total += (iseven(i) ? 2 : 4) * f(x)
    end

    return h * total / 3
end

I_simpson = simpson_rule(sin, 0.0, π, 100)

@show I_simpson
@show abs(I_simpson - 2.0)
```

## **21. Линейная интерполяция**

Для двух точек:

\[
(x_0,y_0),\quad(x_1,y_1)
\]

линейная интерполяция имеет вид:

\[
y(x)=y_0+\frac{x-x_0}{x_1-x_0}(y_1-y_0)
\]


```julia
function lerp(x0, y0, x1, y1, x)
    x0 == x1 && throw(ArgumentError("x0 и x1 должны различаться"))
    y0 + (x - x0) * (y1 - y0) / (x1 - x0)
end

@show lerp(0.0, 0.0, 10.0, 20.0, 2.5)
```

## **22. Полиномиальная интерполяция — осторожно**

Полином высокой степени, проходящий через множество точек, может демонстрировать сильные осцилляции.

Это классический пример того, что «более сложная модель» не обязательно означает более качественное численное решение.

На практике часто предпочитают сплайны или локальные методы интерполяции.

## **24. Обыкновенные дифференциальные уравнения — ODE**

Рассмотрим:

\[
\frac{dy}{dt}=-ky
\]

с условием:

\[
y(0)=y_0
\]

Точное решение:

\[
y(t)=y_0e^{-kt}
\]


## **25. Метод Эйлера**

\[
y_{n+1}=y_n+h f(t_n,y_n)
\]


```julia
function euler(f, tspan, y0, h)
    t0, t1 = tspan
    n = Int(floor((t1 - t0) / h))

    ts = Vector{Float64}(undef, n + 1)
    ys = Vector{Float64}(undef, n + 1)

    ts[1] = t0
    ys[1] = y0

    for i in 1:n
        ts[i+1] = ts[i] + h
        ys[i+1] = ys[i] + h * f(ts[i], ys[i])
    end

    return ts, ys
end

k = 2.0
fode(t, y) = -k*y

ts_e, ys_e = euler(fode, (0.0, 2.0), 1.0, 0.1)

@show ys_e[end]
```

### Сравнение с точным решением


```julia
exact_end = exp(-k * ts_e[end])

@show exact_end
@show abs(ys_e[end] - exact_end)
```

## **26. RK4 — Runge–Kutta 4**

RK4 — значительно более точный явный метод интегрирования ODE.


```julia
function rk4(f, tspan, y0, h)
    t0, t1 = tspan
    n = Int(floor((t1 - t0) / h))

    ts = Vector{Float64}(undef, n + 1)
    ys = Vector{Float64}(undef, n + 1)

    ts[1] = t0
    ys[1] = y0

    for i in 1:n
        t = ts[i]
        y = ys[i]

        k1 = f(t, y)
        k2 = f(t + h/2, y + h*k1/2)
        k3 = f(t + h/2, y + h*k2/2)
        k4 = f(t + h, y + h*k3)

        ts[i+1] = t + h
        ys[i+1] = y + h*(k1 + 2k2 + 2k3 + k4)/6
    end

    return ts, ys
end

ts_rk, ys_rk = rk4(fode, (0.0, 2.0), 1.0, 0.1)

@show ys_rk[end]
@show abs(ys_rk[end] - exp(-k*ts_rk[end]))
```

## **27. Локальная и глобальная ошибка**

### Теория

- **локальная ошибка** — ошибка одного шага метода;
- **глобальная ошибка** — накопленная ошибка после многих шагов.

Для метода Эйлера глобальная ошибка имеет порядок `O(h)`.

Для классического RK4 глобальная ошибка имеет порядок `O(h^4)`.


## **28. Устойчивость шага**

Слишком большой шаг `h` может:

- увеличить ошибку;
- дестабилизировать решение;
- полностью разрушить физический смысл симуляции.

Уменьшение шага увеличивает вычислительную стоимость, поэтому всегда существует компромисс между точностью и временем вычислений.


## **29. Валидация решения**

Хорошие численные вычисления должны включать контрольный тест.

Возможные стратегии:

1. сравнение с аналитическим решением;
2. сравнение с решением при меньшем шаге;
3. проверка предельного поведения;
4. контроль residual;
5. сравнение двух независимых методов.


```julia
for h in (0.2, 0.1, 0.05, 0.025)
    ts, ys = rk4(fode, (0.0, 2.0), 1.0, h)
    err = abs(ys[end] - exp(-k * ts[end]))
    println("h=$h  error=$err")
end
```

## **30. Типичные ошибки**

| Ошибка | Проблема | Правильный подход |
|---|---|---|
| `Float64 == exact` | ненадёжное сравнение | `isapprox` |
| слишком малое `h` | доминирование round-off | исследовать оптимальный шаг |
| Newton без проверки производной | возможное деление на 0 | защитное условие |
| бисекция без смены знака | нет гарантии сходимости | проверить значения на концах |
| `inv(A)` в регрессии | хуже численная устойчивость | `A\b` |
| Euler с большим шагом | большая ошибка / неустойчивость | уменьшить шаг |
| отсутствие валидации | результат может выглядеть корректным, не будучи таковым | residual / convergence test |

## **32. Мини-проект — модель охлаждения Ньютона**

### Задача

Закон охлаждения Ньютона:

\[
\frac{dT}{dt}=-k(T-T_{env})
\]

где:

- `T` — температура объекта;
- `T_env` — температура окружающей среды;
- `k` — коэффициент охлаждения.


```julia
T_env = 20.0
T0 = 90.0
k = 0.12

cooling(t, T) = -k * (T - T_env)

ts, Ts = rk4(cooling, (0.0, 30.0), T0, 0.1)

@show Ts[1]
@show Ts[end]
```

### Точное решение

\[
T(t)=T_{env}+(T_0-T_{env})e^{-kt}
\]


```julia
T_exact(t) = T_env + (T0 - T_env) * exp(-k*t)

error_end = abs(Ts[end] - T_exact(ts[end]))

@show error_end
```

### Время достижения 30°C

Найдём это время численно методом бисекции.


```julia
temperature_difference(t) = T_exact(t) - 30.0

t30 = bisection(temperature_difference, 0.0, 100.0)

@show t30
@show T_exact(t30)
```
